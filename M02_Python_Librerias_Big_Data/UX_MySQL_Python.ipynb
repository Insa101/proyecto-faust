# Como utilizar MySQL en Python

Dentro del mundo del dato, la capacidad de poder relacionar Bases de datos con programas de calculo y analitica de datos se vuelve fundamental.

En este modulo vamos a visionar como podemos utilizar BBDD relacionales (SQL) para vincularlas a archivos de Python, permitiendo generar una relacion medainte claves API para la comunicacion correcta de informacion para su posterior manipulacion.

La idea es que **terminemos de generar una pequeña base de datos con los resultados de nuestros programas** que utilizaremos para otros proyectos de analitica de datos.

> Esto nos permitira generar un sistema mas robusto para poder realizar analisis posteriores y que la informacion generada no se guarde en cache unicamente.

## Contexto: Que son las BBDD Relacionales (SQL)

La diferencia entre las BBDD relacionales (SQL) y no relacionales (NoSQL) radica en **el modo en el que se almacena dichos datos**

En las bases relacionales existe **un modelo *entidad - relacion*** que estructura la informacion en **tablas**

> Lo que genera por detras es la **interrelacion de las tablas de datos entre si y como se interpretan** dando explicacion a su comportamiento y conexion entre los datos

Aspectos importantes:

- **Cada dato es 1 fila** con un identificador clave
- **Cada una de las columnas es un atributo para el dato**
  
> Por tanto el dato *persona* sera una fila y las columnas seran *nombre, DNI, direccion,...*

- La relacion entre cada tabla puede ser distinta:
  > - **relacion 1:1** : Una entidad (dato) de una tabla sobre puede tener una relacion con otra de otra tabla
  >   
  > > *1 DNI, 1 Persona*
  > 
  > - **relacion n:1 o 1:n**: Muchos datos pueden estar vinculados a 1 o 1 datos esta vinculado a muchos.
  >   
  > > *Una ciudad puede tener 1 pais, pero 1 pais puede tener muchas ciudades*
  > 
  > - **relacion n:n***: Muchos datos puedes estar vinculados a muchos
  >   
  > > *muchas personas conocen a muchas personas*
  > >

### Modelo *Entidad - Relacion*

Es el modelo estandar para representar y consultar datos en una aplicacion.

Basicamente estructura datos en subgrupos que se componen de atributos que las definen y permiten relacionar estos atributos a otros subrupos, generando una red de datos interconectada que da sentido al conjunto de la informacion.

> Ejemplo: Un identificador de un **cliente y su direccion postal**, que tiene un subgrupo del **cliente**, donde estara *nombre, apellido, mail,... direccion postal*, a su vez **direccion postal** dispondra de informacion como *codigo postal, calle, piso,... ciudad*, a su vez **ciudad** dispondra de *nombre, pais* y a su vez **pais** tendra *nombre*.
>
> Siguiendo el modelo, el cliente podra tener mas direcciones postales, pero una direccion postal solo podra tener una ciudad, para entender tipos de relaciones.


#### Concepto de *Transaccion* en BBDD

Con las BBDD de datos el lenguaje que realizaremos junto con el sistema es de consulta de datos, pero esas consultas pueden volverse complejas dependiendo de como las planteemos.

Esto sera debido a que la orden facilitada cumplira una serie de propiedades que afectara a los datos.

A esto lo llamamos ***Transaccion*** (conjunto de operaciones) y las propiedades son los ***ACID***

- ***A de Atomicity***: una transaccion es exitosa o falla, no queda a medias.
  > Es decir, si al mandarle una orden, no termina de ejecutarse el conjunto total, revertira el cambio hasta su estado inicial antes de la transaccion
- ***C de Consistency***: una tranasaccion cumple con las reglas para mantener los datos consistentes.
  > Es decir, mantiene la integridad de los datos impidiendo la duplicidad de IDs
- ***I de Isolation***: una transaccion no afecta a otras mientras el gestor de bases de datos permite el acceso concurrente.
  > Es decir, que cuando se lanza un transaccion, hasta que no terminado no puede verse las modificaciones generadas
- ***D de Durability***: las transacciones se guaran en el almacenamiento persistente.
  > Es decir, las modificaciones deben guardarse

Estas propiedades aseguran la integridad de los datos, simplifica la logica de operaciones y fuerza el uso del almacenamiento confiable.

Los proveedores de acceso (como *MySQL*) **nos proveen de estas propiedades automaticamente al realizar consultas con ellos** permitiendo entre otras cosas la simultaneidad de acceso a los datos

## Como conectar MySQL a Python

Una vez visto que son las BBDD por encima, vamos a explicar el modo en el que se pueden conectar a Python.

Para a ello haremo uso de la liberia ***mysql-connector-python*** para poder conectar BBDD a nuestros programas

> ***pip install mysql-connector-python***
>
> ***import mysql.connector***

### Conexion a una BBDD: funciones *connect() y getpass()*

Una vez que ya tenemos configurada la BBDD de datos en MySQL es hora de dotar de acceso en Python para que pueda acceder a la informacion que ella contiene.

Para ello haremos uso de la libreria ***getpass*** para la autentificacion y en la libreria ***mysql.connnector*** del metodo ***connect*** que nos permitira establecer unos parametros para la vinculacion a la BBDD de MySQL.

> ***nombre variable = connect ( host = "lugar donde se encuentra la BBDD", user = input(...), password = getpass (...), database = "nombre BBDD")***
>

#### Comprobar conexion a BBDD: *.is_connected()*

Para comprobar si hemos logrado la conexion a la BBDD podemos utilizar la funcion ***.is_connected()*** que evalua si se ha vinculado correctamente o no

> ***"nombre variable con acceso a BBDD".is_connected()***

### Consultar datos de SQL en Python

Para poder consultar datos una vez vinculados en la BBDD directamente en Python, **es necesario utilizar la sintaxis de SQL**

Por tanto si por ejemplo queremos seleccionar X datos de la BBDD tendremos que indicarselo mediante ***SELECT "nombre datos" from "nombre tabla"***

> ***"nombre variable con ordenes SQL" = " Sintaxis SQL con peticion"***

### Como realizar una transaccion de MySQL en Python: funcion *.cursor()*

Para realizar una transaccion en Python desde una BBDD de MySQL utilizaremos la funcion ***.cursor()*** que nos va a permitir ejecutar operaciones que sigan la sintaxis de SQL pero directamente en Python y sin tener que re-escribir la orden.

> ***"nombre variable con acceso a BBDD".cursor() as cursor:***

A partir de aqui, hay varias funciones que podemos utilizar para utilizar interactuar con la BBDD.

#### Ejecutar ordenes de MySQL: *.execute()*

Nos permite ejecutar una variable que disponga en esu descripcion **ordenes en sintaxis de SQL para ejecutarlas en Pyhton**

> ***"nombre variable con acceso a BBDD".cursor() as cursor:***
> 
> > ***cursor.execute("nombre de variable con ordenes SQL")***

#### Obtener datos de MySQL: *.fetchall()*

Nos permite obtener los datos seleccionados mediante una orden de SQL y traerlos al entorno de Python.

Nos permitira acceder a todos los datos de modo que es conveniente **utilizar un bucle *for in* para la iteracion de los datos**

> ***"nombre variable con acceso a BBDD".cursor() as cursor:***
> 
> > ***cursor.execute("nombre de variable con ordenes SQL")***
> >
> > ***for x in cursor.fetchall():***
