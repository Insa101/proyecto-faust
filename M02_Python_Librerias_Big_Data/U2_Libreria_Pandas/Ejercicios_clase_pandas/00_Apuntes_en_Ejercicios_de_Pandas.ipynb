## Importamos un archivo y previsualizamos datos.

```python
from pathlib import Path
import pandas as pd

Ruta = Path(r"C:\proyecto-faust\M02_Python_Librerias_Big_Data\UX_Material_Extra_Librerias\U10_ejercicios\consumo_horario_cliente.csv")
consumo = pd.read_csv(Ruta, sep = ";")
consumo.head()
```

## Comprobamos los datos que disponemos mediante .head(), .dtypes(), .info()

```python
consumo.head()
```

```python
consumo.dtypes
```

```python
consumo.info()
```

## Cambiamos datetime a la categoria datetime, ya que esta reconocida como string: to_datetime()

```python
consumo["datetime"] = pd.to_datetime(consumo["datetime"])
```

```python
consumo.dtypes
```

## Extraemos el datos de la fecha, mediante dt.date, dt.hour, dt.year, dt.day_name() y le asignamos una columna adicional al DataFrame mediante .assign()

```python
consumo["datetime"].dt.date
```

```python
consumo["datetime"].dt.year
```

```python
consumo["datetime"].dt.day_name()
```

```python
consumo = consumo.assign( fecha = consumo["datetime"].dt.date, mes = consumo["datetime"].dt.month, dia = consumo["datetime"].dt.day_name())
```

```python
consumo.head()
```

```python
consumo = consumo.assign( coste = (consumo["consumo_kwh"] * consumo["precio_kwh"]).round(3))
consumo.head()
```

## 1º Pregunta: ¿Cual es el consumo maximo que se ha registrado en los datos que tenemos? . max()

```python
print(consumo["consumo_kwh"].max())
```

## 2º Pregunta: Indica los datos de la observacion que ha registrado el consumo maximo. Usaremos .loc() y una mascara booleana

```python
consumo.loc[consumo["consumo_kwh"] == consumo["consumo_kwh"].max()]
```

## 3º Pregunta: Filtrar horas con consumo inferior a 0.02 kWh. Usando .loc + masacara booleana.

```python
consumo.loc[consumo["consumo_kwh"] < 0.02]
```

## 4º Pregunta: Consumo promedio de los martes. Usando .loc + mascara booleana + seleccionando una columna + .mean()

```python
consumo.loc[consumo["dia"] == "Tuesday"]["consumo_kwh"].mean()
```

## 5º Pregunta: Precio_kwh promedio de los martes. Usando .loc + mascara booleana + seleccionando una columna + .mean()

```python
consumo.loc[consumo["dia"] == "Tuesday"]["precio_kwh"].mean()
```

## 6º Pregunta: Precio_kwh y consumo promedio de los martes y viernes. Usando .loc para localizar y .isin() para seleccionar martes y viernes, .groupby() por cada dia de la semana para generar un dataframe con unicamente estos dias + .mean()

```python
consumo.loc[consumo["dia"].isin(["Tuesday","Friday"])].groupby("dia")[["precio_kwh", "consumo_kwh"]].mean()
```

## 7º Pregunta: Precio_kwh y consumo promedio por mes. Usando .groupby() para agrupar por meses para generar un dataframe con unicamente estos dias + .mean()

```python
consumo.groupby(consumo["mes"])[["precio_kwh", "consumo_kwh"]].mean()
```

## 8º Pregunta: Cargamos otro fichero que dispone de costes de consumo de mayoristas. 1º lo cargamos, 2º comprobamos tipos de datos, 3º cambiamos categorias de datos si es necesario.

```python
from pathlib import Path
import pandas as pd

Ruta_2 = Path(r"C:\proyecto-faust\M02_Python_Librerias_Big_Data\UX_Material_Extra_Librerias\U10_ejercicios\precio_md.csv")
consumo_extra = pd.read_csv(Ruta_2, sep = ";")
consumo_extra.head()
```

```python
consumo_extra["datetime"] = pd.to_datetime(consumo_extra["datetime"])
```

## 9º Pregunta: Vincular ambos dataframes para tener un unico DataFrame con todos los datos. Usando .merge() indicando los dataframes y el parametro *on* para seleccionar la columna.

```python
consumo_cruce = pd.merge(consumo, consumo_extra, on = "datetime")
```

```python
consumo_cruce.head()
```

## 10º Pregunta: Calcular una nueva columna con la diferencia entre el precio pagado por el consumidor cada hora y el precio de la electricidad diario. Primero sera necesario cambiar la unidades de conversion ya que uno es Kwh y el otro Mwh (/1000). Usaremos de nuevo .assign() con el calculo.

```python
consumo_df = consumo_cruce.assign( dif_precio = (consumo_cruce["precio_kwh"] - (consumo_cruce["preciomd_eurMw"]/1000)).round(4))
consumo_df.head()
```

## 11º Pregunta: Calcula el promedio de la diferencia de precios por mes. Usaremos .groupby() para agrupar los meses + seleccinando la columna dif_precio calcularemos la media con .mean()

```python
consumo_df.groupby("mes")["dif_precio"].mean()
```
