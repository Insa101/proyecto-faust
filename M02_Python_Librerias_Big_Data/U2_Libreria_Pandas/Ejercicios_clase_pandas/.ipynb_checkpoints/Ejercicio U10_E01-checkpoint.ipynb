Ahora vamos a realizar unas operaciones para comprobar conocimientos.

Fijándote en los ejemplos anteriores, ahora te vamos a pedir que escribas un código en el que asignes a nuestro DataFrame de consumos unas columnas con los siguientes valores:

- coste_eur : con el coste en euros de la energía consumida, multiplicando el consumo por el precio
- fecha : extrayendo la fecha de la columna datetime
- hora : extrayendo la hora de la columna datetime
- dia_semana : extrayendo el día de la semana de la columna datetime


```python
from pathlib import Path
import numpy as np
import pandas as pd

Ruta_ej_1 = Path(r"C:\proyecto-faust\M02_Python_Librerias_Big_Data\UX_Material_Extra_Librerias\U10_ejercicios\consumo_horario_cliente.csv")
consumos = pd.read_csv(Ruta_ej_1, sep = ";")
consumos["datetime"] = pd.to_datetime(consumos["datetime"])
consumos = consumos.assign( coste_eur = lambda x: x["consumo_kwh"]*x["precio_kwh"])
consumos = consumos.assign( fecha = consumos["datetime"].dt.date ,
                           mes = consumos["datetime"].dt.month, 
                           dia_semana = consumos["datetime"].dt.day, 
                           hora = consumos["datetime"].dt.hour)
consumos.head()
```

```python
#1º¿Cuál es el consumo máximo que se ha registrado en los datos que tenemos? Primero realizare una opcion que no me da lo que necesito
    # Utilizamos la funcion .sort_values para localizar los valores con consumos mas altos
        # Esta opcion no nos da unicamente el valor mas alto, sino que ordena los datos en funcion de sus valores

from pathlib import Path
import numpy as np
import pandas as pd

Ruta_ej_1 = Path(r"C:\proyecto-faust\M02_Python_Librerias_Big_Data\UX_Material_Extra_Librerias\U10_ejercicios\consumo_horario_cliente.csv")
consumos = pd.read_csv(Ruta_ej_1, sep = ";")
consumos["datetime"] = pd.to_datetime(consumos["datetime"])
consumos = consumos.assign(coste_eur = lambda x: x["consumo_kwh"]*x["precio_kwh"])
consumos = consumos.assign(fecha = consumos["datetime"].dt.date ,
                           mes = consumos["datetime"].dt.month, 
                           dia_semana = consumos["datetime"].dt.day, 
                           hora = consumos["datetime"].dt.hour)
max_consumo = consumos.sort_values(by="consumo_kwh", ascending =False)
max_consumo.head()
```

```python
#1º¿Cuál es el consumo máximo que se ha registrado en los datos que tenemos? Hay 2 opciones
    # Aqui buscamos el valor .max() y luego extraemos la fila en una variable donde se ubica el valor mas alto, para poder referenciar los datos
        # Esta opcion localiza el valor mas alto, extraemos la informacion anexa y la exponemos como queremos.

from pathlib import Path
import numpy as np
import pandas as pd

Ruta_ej_1 = Path(r"C:\proyecto-faust\M02_Python_Librerias_Big_Data\UX_Material_Extra_Librerias\U10_ejercicios\consumo_horario_cliente.csv")
consumos = pd.read_csv(Ruta_ej_1, sep = ";")
consumos["datetime"] = pd.to_datetime(consumos["datetime"])
consumos = consumos.assign(coste_eur = lambda x: x["consumo_kwh"]*x["precio_kwh"])
consumos = consumos.assign(fecha = consumos["datetime"].dt.date ,
                           mes = consumos["datetime"].dt.month, 
                           dia_semana = consumos["datetime"].dt.day, 
                           hora = consumos["datetime"].dt.hour)
max_consumo = consumos["consumo_kwh"].max()
fila_max = consumos[consumos["consumo_kwh"] == max_consumo]
print(fila_max[["consumo_kwh", "dia_semana", "hora"]])
```
