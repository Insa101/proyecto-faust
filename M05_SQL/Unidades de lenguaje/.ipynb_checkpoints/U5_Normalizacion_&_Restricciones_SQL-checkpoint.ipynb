# Apuntes 5 SQL: Normalizacion y  Restricciones

## 1. NORMALIZACION

Durante la normalizacion hay que seguir estas normas de **escritura**:

#### 1. **Se escriben siempre en MAYUSCULAS las entidades**

`EQUIPO`

#### 2. **Los atributos/ campos se escriben en minusculas, dentro de un () y/o en una tabulacion**

`EQUIPO(nombre, n_jugadores, fecha_fundacion)`

EQUIPO(
        nombre, 
        n_jugadores, 
        fecha_fundacion,
    )


#### 3. **Se escriben SIN ACENTOS**

`DENOMINACION_ORIGEN`

#### 4. **Evitamos caracteres especiales (ej: la Ñ)**

#### 5. **Los espacios se escriben con _ o la siguiente palabra se inicia con MAYUSCULA**

`EQUIPO(n_jugadores)`

`EQUIPO(nJugadores)`

#### 6. **Identificamos las claves principales asi:**

 * **Se escriben EN ROJO**
 * **Se añade PK**
 * **SIEMPRE SERAN DE TIPO NUMERICO *INT* (ENTEROS)**
 * **SIEMPRE SERAN NOT NULL (No nulas)**
 * **PUEDE INDICARSE QUE SEAN AUTO_INCREMENT**

`EQUIPO(idEquipo_PK, nombre, n_jugadores, fecha_fundacion)`

`EQUIPO(pkCliente, nombre, n_jugadores, fecha_fundacion)`

#### 7. **Las claves ajenas se escriben asi:**

 * **Se añade FK**
 * **Se escribe EN AZUL si la relacion es 1:N**
 * **Se escribe EN VERDE si la relacion es 1:1**
 * **EL TIPO DE DATO SERA SIEMPRE EL MISMO COMO EN LA TABLA ORIGINAL**

`EQUIPO(idEquipo_PK, nombre, n_jugadores, fecha_fundacion)`

`JUGADOR(idjugador_PK, nombre, apellido, fecha_nacimiento)`

---------

EQUIPO(idEquipo_PK, nombre, n_jugadores, fecha_fundacion) - N:1 - JUGADOR(idjugador_PK, nombre, apellido, fecha_nacimiento)

                                 PARTICIPANTE( idEquipo_FK, idjugador_FK, idParticipante_PK)

#### 8. **El orden de la escritura para campos o atributos sera:**

1. **El TIPO DE DATO en Mayusculas**, con su limite de caracteres o maximos.

`nombre VARCHAR (50)`

2. **La RESTRICCIONES APLICADAS en Mayusculas**

`nombre VARCHAR (50), NOT NULL`

3. **Otras normas aplicables en Mayusculas**

`nombre VARCHAR (50), NOT NULL, AUTO_INCREMENT`

## 2. RESTRICCIONES

#### 1. PRIMARY KEY () o PK

`PK`

> Son las claves principales de las tablas o entidades.

- **Sirven para identificar un registro de forma UNICA**
- **Hay que intentar que siempre sean del tipo INT**

> *idCliente INT, PK*

#### 2. FOREIGN KEY () o FK

`FK`

> Son las claves foraneas de las tablas o entidades.

- **Sirven para determinar que atributos se utilizan par establecer relaciones con otras entidades**
- **Hay que intentar que siempre sean del tipo INT**

> *idPais INT, FK*

#### 3. UNIQUE

`UNIQUE`

- **Sirven para no repetir valores en un campo**
- **Pueden existir varios campos con esta restriccion**

> *docIdentificacion VARCHAR(12), UNIQUE*

#### 4. NOT NULL

`NOT NULL`

- **Evitas la omision de valores o OBLIGAR A DAR UN VALOR DEFINIDO**
- **Pueden existir varios campos con esta restriccion**

> *nombre VARCHAR(50), NOT NULL*

#### 5. DEFAULT

`DEFAULT`

- **Permite asociar un valor por defecto a un atributo, para evitra un NULL**
- Entre parentesis se indica el valor por defecto que se almacenara
- **Pueden existir varios campos con esta restriccion**

> *direccion VARCHAR(100), DEFAULT("Sin determinar")*

#### 6. CHECK

`CHECK`

- **Limita el rango de valores que se le pueden asignar a un atributo**
- Entre parentesis se indica **LA CONDICION QUE DEBE CUMPLIR EL VALOR PARA QUE SE ALMACENE**
- **Pueden existir varios campos con esta restriccion**

> *edad INT, CHECK(EDAD >= 18)*

## 3. Caracteres Especiales

Permite establecer reglas que personalizan mas el comportamiento de los datos que se almacenan.

Hay 2 principales a conocer:

#### 1. AUTO_INCREMENT:

`AUTO_INCREMENT`

Es una caracteristica que solo se puede utilizar **con datos tipo numerico, permite incrementar el valor del dato automaticamente**. Estos valores siempre seran unicos.

> *idPais INT, FK, AUTO_INCREMENT*

#### 2. UNSIGNED:

`UNSIGNED`

Es una caracteristica que solo se puede utilizar **con datos tipo numerico INT, impide almacenar datos negativos**. 

> *nPersonas INT, UNSIGNED*
