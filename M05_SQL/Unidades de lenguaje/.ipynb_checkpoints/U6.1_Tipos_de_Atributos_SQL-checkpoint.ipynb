# Apuntes 6.1 SQL: Tipos de atributos

Aqui se indicaran los principales y distintos tipos de atributos que existen en SQL.

## 1. Tipo texto

Registra campos compuestos por **letras, numeros y caracteres especiales**

Estos son los principales que has de conocer:

#### Tipo Alfanumerico

- `CHAR(N)` Permite almacenar N **caracteres alfanumericos entre 0 y 255**. Se puede escribir sin parentesis si no hay limite explicito

> *codPostal CHAR(5)*
>
> *telefono CHAR*

- `VARCHAR (N)` Permite almacenar N **caracteres alfanumericos entre 0 y 65335**. Se puede escribir sin parentesis si no hay limite explicito

> *nombre VARCHAR(50)*
>
> *titulo VARCHAR*

- `TINYTEXT` Permite almacenar **caracteres alfanumericos hasta 255**

> *descripcion TINYTEXT*

- `TEXT` Permite almacenar **caracteres alfanumericos hasta 65335**

> *descripcion TEXT*

- `MEDIUMTEXT` Permite almacenar **caracteres alfanumericos hasta 16777215**

> *descripcion MEDIUMTEXT*

- `LONGTEXT` Permite almacenar **caracteres alfanumericos hasta 4294967295**

> *descripcion LONGTEXT*

#### Tipo Lista o seleccion de valores

- `ENUM` Permite **asociar un valor de un conjunto de valores posibles (es un listado)**, ademas puede almacenar un conjunto de hasta 65335 valores.
- **VIP: ESTOS DATOS LOS PONDREMOS PARA VALORES QUE JAMAS SE VAYAN A MODIFICAR** en otro caso seria recomendable crear una entidad independiente

> *tipoDocumento ENUM("DNI, "NIE, "PASAPORTE, "OTROS")*

- `SET` Permite **asociar uno o mas valores  de un conjunto de valores posibles (es un listado multiple)**, ademas puede almacenar un conjunto de hasta 64 valores.

> *idiomasConocidos SET("Español, "Ingles, "Frances, "Chino")*

#### Tipo Texto Binarios

- `BINARY` Permite almacenar **cadenas binarias de Bytes en lugar de no binarios, similar a CHAR**

- `VARBINARY` Permite almacenar **cadenas binarias de Bytes en lugar de no binarios, similar a VARCHAR**

- `TINYBLOB` Permite almacenar **cadenas binarias de Bytes en lugar de no binarios, similar a TINYTEXT**

- `BLOB` Permite almacenar **cadenas binarias de Bytes en lugar de no binarios, similar a TEXT**

- `MEDIUMBLOB` Permite almacenar **cadenas binarias de Bytes en lugar de no binarios, similar a MEDIUMTEXT**

- `MEDIUMBLOB` Permite almacenar **cadenas binarias de Bytes en lugar de no binarios, similar a LONGTEXT**

## 2. Tipo numerico

Registra campos compuestos por **numeros, enteros y con decimales y de un tamaño determinado, DESTINADOS A REALIZAR OPERACIONES ARITMETICAS**

Estos son los principales que has de conocer:

#### Tipo Enteros

- `TINYINT(N)` Permite almacenar **numeros enteros entre -128 y 127 de 1 byte**. Se puede escribir sin parentesis si no hay limite explicito

> *edad TINYINT(3)*
>
> *edad TINYINT*

- `SMALLINT(N)` Permite almacenar **numeros enteros entre -32768 y 32767, de 2 bytes**. Se puede escribir sin parentesis si no hay limite explicito

> *numeroEmpleados SMALLINT(40)*
>
> *numeroEmpleados SMALLINT*

- `MEDIUMINT(N)` Permite almacenar **numeros enteros entre -8388608 y 8388607, de 3 bytes**. Se puede escribir sin parentesis si no hay limite explicito

> *numeroEmpleados SMALLINT(40)*
>
> *numeroEmpleados SMALLINT*

- `INT(N)` Permite almacenar **numeros enteros entre -2147483648 y 2147483648, de 4 bytes**. Se puede escribir sin parentesis si no hay limite explicito

> *numeroEmpleados INT(40)*
>
> *numeroEmpleados INT*

- `BIGINT(N)` Permite almacenar **numeros enteros de 8 bytes**. Se puede escribir sin parentesis si no hay limite explicito

> *numeroEmpleados BIGINT(40)*
>
> *numeroEmpleados BIGINT*

#### Tipo Decimales

- `FLOAT(N,D)` Permite almacenar **numeros decimales pequeños**. **Opcional** Se puede escribir la cantidad maxima de digidos entre enteros y decimales (N) que se desean y el maximo de decimales (D)


> *temperatura FLOAT(2,2)*
>
> *temperatura FLOAT()*

- `DECIMAL(N,D)` Permite almacenar **numeros decimales mas grandes que el tipo FLOAT**. **DEBES** escribir la cantidad maxima de digidos entre enteros y decimales (N) que se desean y el maximo de decimales (D)

> *peso DECIMAL(5,2)*
>
> *altura DECIMAL(3,2)*

- `DOUBLE(N,D)` Permite almacenar **numeros decimales mas grande que con FLOAT, interesante si queremos guardar numeros con mucha mas carga decimal**. **Opcional** Se puede escribir la cantidad maxima de digidos entre enteros y decimales (N) que se desean y el maximo de decimales (D).

> *temperatura DOUBLE(2,2)*
>
> *temperatura DOUBLE()*

- `REAL(N,D)` Permite almacenar **numeros decimales como FLOAT y DOUBLE**. **Opcional** Depende de como se condigure se parece mas a uno u otro.

> *temperatura REAL(2,2)*


## 3. Tipo Fecha y Hora

Registra campos compuestos por **fechas y horas en distintos formatos**

Estos son los principales que has de conocer:

- `DATE` Permite almacenar **fechas en formato YYYY-MM-DD**, en un rango entre *1000-01-01 y 9999-12-31*

> *fechaNacimiento DATE*

- `TIME` Permite almacenar **horas en formato hh:mm:ss**, en un rango entre *-88:59:59 y 838:59:59*

> *horaInicio TIME*

- `DATETIME` Permite almacenar **fechas y horas en formato YYYY-MM-DD hh:mm:ss**, en un rango entre *1000-01-01 00:00:00 y y 9999-12-31 23:59:59*

> *fechaHoraReserva DATETIME*

- `TIMESTAMP` Permite almacenar **fechas y horas en formato YYYY-MM-DD hh:mm:ss**, en un rango entre *1970-01-01 00:00:00 y y 2038-01-09 03:14:07*

> *fechaHoraReserva DATETIME*

- `YEAR` Permite almacenar **años en formato de 4 diitos**, en un rango entre *1901 y 2155*

> *anyoFabricacion YEAR*

## 4. Atributos Especiales: Claves Artificiales: Principales PK y Foraneas FK

Son atributos que **ayudan a mejorar la estructura de la BBDD**, por lo que no definen al objeto o nacen del objeto en si.

> Suelen ser necesarios para *suplir una falta de atributo propio que permita identificar los registros de forma unica e inequivoca*

> Aunque tecnicamente no son atributos y si RESTRICCIONES, al indicarse en cada entdiad **les convierte en un atributo especial**

Son 2:

- `PK` es la utilizada para CLAVES PRINCIPALES

> *idCliente PK, INT, NOT NULL*

- `FK` es la utilizada para CLAVES FORANEAS

> *idCliente FK, INT, NOT NULL*
