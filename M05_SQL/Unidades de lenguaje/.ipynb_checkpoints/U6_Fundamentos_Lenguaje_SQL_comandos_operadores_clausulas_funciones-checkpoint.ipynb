## 3. Fundamentos Generales: COMANDOS BASICOS DDL, DML, DCL

> **TODAS LAS SENTENCIAS TERMINARAN EN ;**

Existen 3 tipos de comandos:

#### 1. ***DDL***: Para crear estructuras y elementos en el servidor

`CREATE` **PARA CREAR**



`DROP` **PARA ELIMINAR**



> EN ESTE CASO NO ES NECESARIO ESPECIFICAR LOS CAMPOS SI QUEREMOS ELIMINAR UNA TABLA

`ALTER` **PARA MODIFICAR**



#### 2. ***DML***: Para manipular los datos

`INSERT` **PARA INTRODUCIR DATOS**



>**LA INSERCCION DE LOS VALORES SE REALIZARA EN EL ORDEN QUE HEMOS CONFIGURADO/ESCRITO DE LOS ATRIBUTOS SELECCIONADOS**

`UPDATE` **PARA ACTUALIZAR DATOS**

V

> **LA ACTUALIZACION SE REALIZARA CON *SET* INDICANDO EL ATRIBUTO A MODIFICAR**

`DELETE` **PARA ELIMINAR DATOS**



> **HABRA QUE INDICAR DESDE DONDE QUEREMOS ELEMINAR Y QUE ELEMENTO, SELECCIONANDO UN FILTRADO CONDICIONAL *WHERE***

`SELECT` **PARA SELECCIONAR/FILTRAR DATOS**



> **PODREMOS SELECCIONAR LOS ELEMENTOS QUE QUERAMOS ( * ES PARA SELECCIONAR TODOS) DESDE LA TABLA QUE QUERAMOS**

#### 3. ***DCL***: Para gestionar los permisos de usuarios.

`GRANT` **PARA GARANTIZAR O DAR PERMISOS A USUARIOS**

`REVOKE` **PARA QUITAR O DENEGAR PERMISOS A USUARIOS**

### 3.5 COMANDOS BÁSICOS EXTRA:

`USE` **TE PERMITE SELECCIONAR UNA BBDD CON LA QUE TRABAJAR EN LA SESION ACTUAL**



`SHOW` **TE MUESTRA INFORMACION DE LEMENTOS DEL SERVIDOR**

Existen distintas sintaxis:

- `SHOW ENGINES;` = MOSTRAR MOTOR DE ALMACENAMIENTO
- `SHOW CHARSET;` = MOSTRAR CONJUNTO DE CARACTERES
- `SHOW DATABASES;` = MOSTRRAR LAS BBDD QUE HAY EN EL SERVIDOR
- `SHOW CREATE DATABASE nombreBBDD;` = MOSTRAR LAS CONSULTAS UTILIZADAS PARA CREAR UNA BBDD
- `SHOW TABLES;` = MOSTRAR LAS ENTIDADES
- `SHOW CREATE TABLE nombreTabla;` = MOSTRAR UNA ENTIDAD EN CONCRETO

`AS`  TIENE 2 USOS PRINCIPALES:

- 1. **TE PERMITE RENOMBRAR UNA COLUMNA AL REALIZAR UN FILTRADO O SELECCION**
 
Ej: `SELECT ROUND (PI(),2) AS "Numero PI";`



- 2. **TE PERMITE ACORTAR EL NOMBRE DE UNA COLUMNA**
 
Ej: `SELECT empleados.nombre AS "Nombre" FROM empleados as e;`



> **Evitar utilizar el mismo alias para VARIAS ENTIDADES DISTINTAS ya que puede provocar un ERROR**

`SET` **TE PERMITE REALIZAR DISTINTAS ACCIONES EN EL SERVIDOR**





## 4. Fundamentos Generales: OPERADORES

#### 1. Operadores Aritmeticos: +, -, *, /, %

```python
2+2;
5%7;
12/5;
```

#### 2. Operadores de Comparacion: >, <, >=, <=, =, != O <>

```python
edad <18;
edad > 18;
edad = 18;
edad =¡ 18;
```

#### 3. Operadores logicos Y u O:  AND, OR

```python
(Condicion 1) AND (Condicion 2);
(Condicion 1) OR (Condicion 2);
```

#### 4. Operadores logicos NO ES, ESTA O NO ESTA, ENTRE: NOT, IN o NOT IN, BETWEEN .... AND ....

```python
NOT (Edad >18)
# NOT es equivalente a !=
pais IN ('ESPAÑA','FRANCIA','ALEMANIA')
pais NOT IN ('ESPAÑA','FRANCIA','ALEMANIA')
edad BETWEEN 18 AND 21;
#Ambos inclusives
```

#### 5. Operadores logicos comparar valor contra un atributo o expresion : LIKE

> **UNA EXPRESION ES UNA VALIDACION QUE INDICA QUE CARACTERES OBLIGATORIOS TIENE QUE CONTENER**

- Los ? indica QUE ES UN CARACTER QUE NO SABEMOS PERO SI CONOCEMOS SU POSICION
- Los * SELECCIONA TODOS LOS ELEMENTOS
- Los % INDICAN CUALQUIER CARACTER

```python
nombrePais LIKE 'ESPAÑA'
# Para encontrar exacta y unicamente ese texto

nombrePais LIKE 'E%'
# Se seleccionaran TODOS LOS DATOS QUE EMPIECEN POR E

nombrePais LIKE '%A'
# Se seleccionaran TODOS LOS DATOS QUE EMPIECEN POR A

nombrePais LIKE '%A%'
# Se seleccionaran TODOS LOS DATOS QUE CONTENGAN UNA A

nombrePais LIKE '??A'
# Se seleccionaran TODOS LOS DATOS QUE EN EL 3º CARACTER TENGAN UNA A
```

#### 6. Operadores logicos BOOLEANOS: ANY, SOME, ALL

> **PERMITEN COMPROBAR EN UNA SELECCION SI ENTRE LOS DATOS SE CUMPLEN LAS CONDICIONES**
>
> **HAY QUE REALIZAR SUBCONSULTAS PARA PODER UTILIZARLO BIEN**

`ANY` / `SOME` + (CONDICION) **TE INDICAN TRUE SI EXISTE ALGUN CAMPO CUMPLE CON LO BUSCADO**

```python
SELECT nombre_producto
FROM productos
WHERE precio > ANY (
    SELECT precio 
    FROM productos 
    WHERE categoria = 'Electrónica'
);
```

`ALL` / (CONDICION) **TE INDICAN TRUE SI TODOS LOS CAMPOS CUMPLEN CON LO BUSCADO**

```python
SELECT nombre_empleado, salario
FROM empleados
WHERE salario > ALL (
    SELECT salario 
    FROM empleados 
    WHERE departamento = 'Ventas'
);
```

## 5. Fundamentos Generales: CLAUSULAS

> **Son sentencias para añadir mas informacion, algunas son obligatorias y otras van a ser opcionales**

- `FROM` **CLAUSULA OBLIGATORIA EN CONSULTAS** `SELECT` **NOS INDICA DE QUE ENTIDAD SE OBTIENEN LOS DATOS**



- `WHERE` **ESTABLECE LAS CONDICIONES DE UN** `SELECT` 



- `GROUP BY` **AGRUPA RESULTADOS**

- `HAVING` **REALIZA UN FILTRO SOBRE LAS AGRUPACIONES** `GROUP BY`

- `ORDER BY` **ORGANIZA LOS RESULTADOS**



Se pueden realizar **ORDENACIONES MIXTAS SI SE COGEN MAS ATRIBUTOS**



- `CASE` **GENERA UNA ESTRUCTURA DE CONTROL DE FLUJO PARECIDO A UN *IF***

Esta sentencia se utiliza con este listado de clausulas:

- `WHEN` = DEFINE CADA CONDICION A EVALUAR
- `THEN` = ESTABLECE EL VALOR QUE SE DEBE DEVOLVER SI SE CUMPLE LA CONDICION
- `ELSE` = ESTABLECE EL VALOR QUE SE DEBE DEVOLVER SI NO SE CUMPLE LA CONDICION
- `END` = MARCA EL FINAL DEL CONDICIONAL

```python
CASE
    WHEN edad < 0 THEN "Dato inválido"
    WHEN edad >= 0 AND edad <=5  THEN 'Primera Infancia'
    WHEN edad >= 6 AND edad <= 11 THEN 'Infancia'
    WHEN edad >= 12 AND edad <= 18 THEN "Adolescencia"
    WHEN edad >= 18 AND edad<= 26 THEN "Jueventud"
    WHEN edad >= 27 AND edad <= 59 THEN "Adultez"
    ELSE "Persona Mayor"
END AS ciclo_vida
```

## 7. Fundamentos Generales: FUNCIONES

> **SE UTILIZAN PARA REPETIR UNA OPERACION RECURRENTE**

`SELECT` **Y EL NOMBRE DE UNA FUNCION** `FUNCION(PARAMETRO1, PARAMETRO2,...)` 

Algunas son:

- `CEILING(NUMERO)` **DEVUELVE EL NUMERO ENTERO MAS PEQUEÑO IGUAL O SUPERIOR AL DADO**
- `FLOOR(NUMERO)` **REDONDEA AL NUMERO ENTERO MAS PEQUEÑO IGUAL O INFERIOR AL DADO**
- `MOD (NUMERO A DIVIDIR, DIVISOR)` **DA EL RESTO DE UNA DIVISION DE 2 NUMEROS DADOS**
- `RAND()` **NOS DA UN NUMERO ALEATORIO ENTRE 0 Y 1**
- `ROUND(NUMERO DE DECIMALES)` **REDONDEA A UN NUMERO CON DECIMALES INDICADO**
- `MAX (ATRIBUTO)` **NOS DA EL VALOR MAXIMO DE UNA COLUMNA**
- `COUNT (ATRIBUTO)` **NOS DA LA CUENTA DE LOS VALORES DE UNA COLUMNA**
- `AVG (ATRIBUTO)` **NOS DA EL VALOR MEDIO DE UNA COLUMNA**
- `CONCAT (VALOR O ATRIBUTO, VALOR QUE SE AÑADIRA, VALOR QUE SE AÑADIRA"` **NOS CONCATENA CADENAS DE TEXTO**
- `CONCAT_WS (" ", VALOR, VALOR,...)` **AÑADE UN CARACTER ENTRE VALORES DADOS**, por ejemplo meter espacios entre palabras.
- `CURRENT_DATE()` **NOS INDICA LA FECHA ACTUAL**
