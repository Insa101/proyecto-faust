En esta unidad nos vamos a centrar en el diseño conceptual que se puede entender como **un esquema o diagrama donde se representa el analisis previo realizado a traves de la entidad-relacion.**

> Recuerda que la primera fase de analisis busca conocer como es la informacion, como se relaciona y el objetivo de la BBDD que desarrollara el resto de etapas.

Esta fase es fundamental ya que:

- Nos permite visualizar si **en la parte de analisis nos hemos pasado algo por alto**
- Comprobamos **como se relacionan nuestros datos bajo el concepto entidad-relacion**
- Vemos **que tipo de datos tenemos y que propiedades tienen**

## 1.0 Concepto: Entidad - Relacion (*MER*) ¿Que elementos los componen?

1. ***Entidad***: **Son las TABLAS de una BBDD**, representa los datos, componen la BBDD y esta compuesta por varios atributos/propiedades distintos.

Tienen estas caracteristicas:

 * Son la estructura de informacion que necesitamos utilizar o planificar
 * Son los elementos **que se relacionaran**

2. ***Atributo***: **Son los CAMPOS de las tablas**, es decir, las caracteristicas o propiedades que va a tener una entidad**

3. ***Relacion***: **Es la UNION entre distintas entidades**, se uniran a traves de los atributos que comparten

> **Es necesario que ambas TABLAS unidas COMPARTAN el CAMPO que les vincula**

4. ***Cardinalidad***: **Son LAS COINCIDENCIAS MINIMAS Y MAXIMAS que se generan entre 2 o mas entidades**, por tanto, son valores concretos de cuantas veces puede vincularse esa entidad con otra entidad. Estas son:
    - 0: representa que la entidad **no está obligada a participar en la coincidencia.**
    - 1: representa que la entidad **está obligada a participar en la coincidencia 1 vez.**
    - N, M o *: representa que la entidad **está obligada a participar en la coincidencia muchas veces..**

5. ***Relaciones Especiales:*** Son aquellas que se generan a traves de un proceso distinto a la cardinalidad.

> La mas utilizada es ***la especializacion*** y se genera cuando **una entidad se considera PADRE/MADRE y se vincula a otras como HIJAS**
> 
> > Las entidades hijas **heredan todos los atributos de la entidad padre Y AÑADEN NUEVOS (exclusivos de si)**

## 2.0 Concepto: Entidad - Relacion (*MER*) ¿Como se representan?

Aqui se representan las formas que compondran nuestro diagrama:

#### ENTIDADES

 * **Se representan con la forma de un *RECTANGULO***
 * **Se escriben siempre en MAYUSCULAS**
 * **Se escriben SIN ACENTOS**
 * **Evitamos caracteres especiales (ej: la Ñ)**



#### RELACIONES

 * **Se representan con la forma de un *ROMBO***
 * Dentro del rombo se indicara **el verbo o accion que vincula junto con las INICIALES DE LAS ENTIDADES (Para saber quien hace que)**
 * Las relaciones se indican **mediante un parentesis indicando los posibles n datos compartidos en la relacion por cada entidad**



#### ATRIBUTOS

* **Se representan con la forma de *CIRCULOS* o *ELIPSES***



#### CARDINALIDAD

* **Se representa ENTRE PARENTESIS (minimo: MAXIMO), JUNTO A LA ENTIDAD QUE DELIMITA LA RELACION**
* Para evitar redundancias las relaciones *muchos:muchos* se representa ***(N:M)***, siendo la ***N*** el muchos de una entidad y la ***M*** de la otra, **siendo necesario utilizarlo de este modo en sus propias representaciones**



#### ESPECIALIZACION

* **Se representa con la forma de un TRIANGULO INVERTIDO**
* Dentro del triangulo se incluye la palabra **ES o IS**



#### ENTIDAD ESPECIAL: *TABLAS INTERMEDIAS*

**Son ENTIDADES que surgen de RELACIONES N:M** que se encarga de recoger la union de ambas tablas, vinculandose gracias a los ID's de las 2 ENTIDADES que la han formado.



### 2.1 Tipos de relaciones (cardinalidades)

- ***Relaciones uno a muchos (1:N o N:1)***: Se dan cuando **1 registro de una entidad se puede relacionar con muchos registros de la otra entidad**



- ***Relaciones muchos a muchos (M:N o N:M O N:N)***: Se dan cuando **muchos registros de una entidad se puede relacionar con muchos registros de la otra entidad**



- ***Relaciones uno a muchos (1:1)***: Se dan cuando **1 registro de una entidad se puede relacionar SOLO con 1 registro de la otra entidad**



## 3.0 ¿Como es el proceso de desarrollo?


Ahora detallaremos que pasos y aspectos hay que tener en cuenta para llevarlo a cabo el diseño:

#### 1.0 Analizar la informacion

El primer paso empieza por recurrir a la informacion inicial desde la que partimos. Leer bien cuales son los requisitos, cual es el objetivo y que normas habra que cumplir.

De este analisis **obtendremos las ENTIDADES, LAS RELACIONES, LAS CARDINALIDADES, RELACIONES ESPECIALES y NORMAS**

El segundo paso sera **abrir un archivo de texto y una hoja de diagramas en blanco**

#### 2.0  Definir la representacion de las entidades

En el archivo de texto **indicaremos las entidades con sus atributos** y en el diagrama **CREAREMOS LOS RECUADROS con los nombres de las ENTIDADES**

- Las **Entidades** se escribiran siempre en PLURAL y MAYUSCULA

`EQUIPOS`

- Los atributos e incluiran como un listado

`EQUIPOS`
`los atributos son: nombre, nº jugadores, año fundacion`

#### 3.0 Definir las relaciones entre entidades

En el archivo de texto podemos definir que relaciones nos han indicado y en el diagrama **CREAREMOS LOS ROMBOS con los verbos y las siglas**

#### 4.0 Estableceremos la cardinalidad

Nos preguntaremos y responderemos **para cada ENTIDAD y cada RELACION como son las relaciones CARDINALES**, para posteriormente **indicarlos en el diagrama**
