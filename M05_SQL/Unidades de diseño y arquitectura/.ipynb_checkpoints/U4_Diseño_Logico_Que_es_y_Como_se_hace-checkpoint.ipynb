# Apuntes 4: Diseño Logico en BBDD Relacionales

Como ya vimos con anterioridad a la hora de realizar una BBDD hay que seguir unas fases definidas, siendo estas las siguientes:

* Analisis
* Diseño y moldeamiento
    * Diseño Conceptual
    * **Diseño Logico**
    * Diseño Fisico
* Implementacion
* Uso y gestion de BBDD

En esta unidad nos vamos a centrar en el diseño logico que se puede entender como **un esquema o diagrama donde se representan la normalizacion de las entidades-relacion.**

> Esta fase es la predecesora del diseño conceptual y aqui es importante aplicar las normas que deben de cumplir los enlaces para que funcione correctamente.

Esta fase es fundamental ya que:

- Nos permite **mejorar las conexiones planteadas**
- Aumentar o cambiar aspectos del diseño que previamente no habiamos localizado


## 1.0 ¿Que es el diseño logico?

Aunque ya lo hemos mencionado anteriormente podemos decir que esta fase es **la 3º fase del diseño de BBDD**

Esta fase es fundamental ya que una correcta arquitectura, no solo nos permitira albergar informacion, **sino que permitira realizar una correcta gestion de la misma** a traves de *las consultas*

> Terminar esta parte del diseño supone ***el 50% de la BBDD*** ya que la creacion de la misma es sumamente sencilla, otro ***40% de la BBDD*** es el uso adecuado de las consultas y un ***10% de la BBDD*** seria la propia creacion de la BBDD y las tablas

En ella vamos a transformar el diagrama conceptual de entidad relacion a un diagrama que represente:

1. El modelo relacional
2. El grafo relacional.
3. La normalizacion

Esta fase concluye una vez se ha revisado/confirmado cada uno de las relaciones/normas de la BBDD y **se puede proceder a realizar el diseño fisico**

## 2.0 Modelo relacional

Para poder realizar este diagrama, damos un paso mas en la representacion de los datos.

**Es una representacion total de los campos de los datos a nivel logico y de los objetos obtenidos durante el diseño conceptual**

> En los programas gestores este modelo es conocido como ***vista diseño***



Existen 2 formas de representarlo:

1. Graficamente = ***Diagrama relacional***
2. Esquematicamente = ***Grafo relacional***

Por tanto en este modelo:

- Se representan **todas las entidades en FORMA DE TABLA**



- Se representan **todos los atributos y que tipo son, COMPONIENDO LOS CAMPOS/COLUMNAS DE LAS TABLAS**



- Se representan **todas las relaciones  MEDIANTE LINEAS QUE UNEN LAS TABLAS Y LAS PROPIEDADES VINCULANTES**



- Se represetan **todos los registros u objetos que tienen informacion de los atributos y pertenecen a las tablas, SON LAS FILAS**



- Se representan **todas las normas que se tienen que cumplir en las relaciones**

En definitiva y como conclusion: **una vez terminas de crear este modelo, lo trasladas a codigo de consulta para poder crear las BBDD**

La creacion de este modelo relacional se realizara mediante 2 herramientas: ***el grafo relacional y la normalizacion***

## 3.0 El grafo relacional

**Es otro esquema que representa esquema relacionales complejos, con tipo texto o con diagramas donde se reflejan los campos de las entidades**

La representacion de los grafos se representan tipo texto asi:

`ENTIDAD 1 ( campo1 PK, campo2 FK, ...)`

`ENTIDAD 2 ( campo1, campo2, campo3,...)`

## 4.0 La normalizacion

**Es un proceso de designacion y aplicacion de reglas a atributos de las entidades obtenidas durante la creacion del grafo relacional**

> Su objetivo es **reducir la redundancia de datos y facilitar la gestión**

Beneficios de la normalizacion:

- Disminuir problemas durante la actualizacion de los datos en las tablas
- Proteger la integridad de los datos
- Reducir la redundancia de los datos

Hay que tener en cuenta que ademas:

- Modificar significamente la estructura planteada
- Aumentara el nº de entidades
- Aumentara el nº de atributos

Existen **6 normas de normalizacion, pero las 3 primeras son principales, sin ellas no se considera que esta realizada la BBDD**

### 4.1 Las 3 reglas principales de la normalizacion.

Dentro de la normalizacion existen 3 normas que se aplicaran siempre al crear nuestros modelos.

**Estas 3 normas primarias son las 3 Formas Normales**

#### 1º Norma Formal (1NF)

- **Los atributos seran atomicos, indivisibles y minimos**

Se veran 2 ejemplos de como se puede tratar:

1. Cuando **la informacion del registro no es atomica y hay que GENERAR MAS CAMPOS**, porque la informacion es irrelevante que este replicada en otros registros.



2. Cuando **la informacion del registro no es atomica y hay que GENERAR MAS TABLAS**, porque es informacion asignable a otros registros y es relevante.



- **Cada registro debe tener UN CAMPO UNICO que lo idenfique, haciendo uso de una clave primaria asignada o creada artificialmente**
- **Cada CLAVE PRIMARIA SIN NULOS**
- Los datos si cambian de orden **no deben cambiar su significado**, es decir, cada atributo es independiente entre ellos y no pierden su significado.

#### 2º Norma Formal (2NF)

- Se debe haber aplicado las normas anteriores
- Se **eliminan los valores múltiples, generando con ellos una nueva entidad y estableciendo su relación correspondiente.**
  * Si la relacion es 1:N, **en la entidad del 1, se va a generar una clave artificial FK que sera de la tabla N en la relacion**


#### 3º Norma Formal (3NF)

- Se cumple la 2º norma formal
- **Se eliminan aquellos atributos que no dependan directamente de la entidad**
  * Hace que en una tabla aquellos atributos **que pueden depender de otras entidades, SE RELACIONEN CON LA TABLA MEDIANTE CLAVES FORANEAS**



## 5.0 Proceso de creacion del modelo relacional

Una vez vistos los conceptos del modelo relacional podemos ir desarrollandolo.

Las 2 herramientas principales se iran **creando simultaneamente** para ir dando forma al modelo.

La idea es **aplicando las normas, ir generando el grafo que compone la relaciones de las entidades que representan los datos**


### Ejemplo explicativo.

Tenemos este planteamiento para hacer el diseño:

#### 1. Queremos almacenar estos objetos con estos campos
> **Los nombres de las entidades se escribiran en Mayusculas y sin acentos**

`CLIENTES`

* nombre
* apellidos
* telefono
* email
* fecha_nacimiento
* direccion
* doc_identificacion

`PRODUCTOS`

* nombre
* precio
* categoria
* fecha_registro

#### 2. Las relaciones entre estas entidades sera:

`CLIENTES - N:M - PRODUCTOS`

> **Esto era lo que hacias en la fase de diseño conceptual**

#### 3. Planteamos nuevas tablas intermedias

- Las tablas intermedias se generan cuando tenemos **relaciones N:M**
- Habra que establecer un nombre para la tabla
- Albergara los campos identificativos que vinculan la relacion entre las 2 entidades relacionadas

CLIENTES - N:M - PRODUCTOS

    REGISTRO COMPRA

------------

`REGISTRO COMPRA`

* cliente
* producto
* fecha_compra

#### 4. 1NF: Crear o asignar CLAVES PRINCIPALES (PK) para los registros para cada entidad

> Es el campo encargado de identificar cada registro de maner unica e inequivoca

`CLIENTES`

* nombre
* apellidos
* telefono
* email
* fecha_nacimiento
* direccion
* doc_identificacion
* ***idCliente***

`PRODUCTOS`

* nombre
* precio
* categoria
* fecha_registro
* ***idProductos***

`REGISTRO COMPRA`

* cliente
* producto
* fecha_compra
* ***idCompra***

#### 5. 1NF: Separar datos compuestos:

> Puede pasar que haya campos que dispongan de datos que se puedan separar en mas columnas o atributos

`apellido: Lopez Benedes`

-------------------------

`apellido 1 : Lopez`

`apellido 2 : Benedes`

#### 6. 2NF: Eliminacion de atributos multiples o compuestos:

> Puede suceder que 2 entidades, **compartan un mismo atributo que les define y a su relacion, haciendo que sea redundante y potencialmente equivoco**
>
> Puede suceder que un dato **se pueda escribir de distintas maneras pero solo tiene 1 significado**


`CLIENTE (ciudad, asociacion) - N:N - ASOCIACION (ciudad, nombre)`

En estos casos es necesario **identificarlos y generar una nueva tabla con los campos compuestos**.

- **SE CREARA UNA ENTIDAD CON EL ID QUE IDENTIFIQUE AL CAMPO COMPUESTO**

`CLIENTE (asociacion) - N:M - ASOCIACION (nombre)`

-------------

`CLIENTE (ciudad) - 1:N - CIUDAD (nombre) - N:1- ASOCIACION (ciudad)`


---------

`CLIENTE (asociacion, idCiudad_FK) - 1:N - CIUDAD (idCiudad_PK, nombre) - N:1- ASOCIACION (idCiudad_FK, nombre)`

#### 7. 2NF: Asignar CLAVES AJENAS / FORANEAS (FK) a las entidades segun la relacion:

> **Son claves principales originarias de otras entidades, que se duplican en otras entidades que pasan a ser un atributo heredado de la relacion entre las entidades**

CLIENTES - N:M - PRODUCTOS

    REGISTRO COMPRA

---------------

`REGISTRO COMPRA (idCompra_PK, idCliente_FK, idProductos_FK, fecha_compra)`

##### 8. Creamos las Tablas que no dependen de otras o que tienen N relaciones.

> Estas son las que en el modelo entidad - relacion tienen un N: o :M

**Se genera el codigo normalizado de cada campo con el que crearemos nuestra BBDD**

PROVINCIAS (
            nombre VARCHAR(50),
            nHabitantes INT (7),
            cp,
            idProvincia PK INT NOT NULL AUTO_INCREMENT,
            )

###### 9. Establecemos las claves primarias PK

Se indica asi:

- 1º `nombre atributo INT NOT NULL AUTO_INCREMENT`

- 2º `PRIMARY KEY (atributo que hara de id)`


PROVINCIAS (
            nombre VARCHAR(50),
            nHabitantes INT (7),
            cp,
            idProvincia INT NOT NULL AUTO_INCREMENT,
            PRIMARY KEY (idProvincia)
            )

###### 10. Establecemos las relaciones y clases ajenas FK

Se indicaran asi:

- 1º `nombre atributo id FK INT`

- 2º `FOREIGN KEY (id FK) REFERENCES nombre entidad referenciada (id FK)`

CIUDAD(
            nombre VARCHAR(50),
            nHabitantes INT (200),
            idProvincia INT,
            idCiudad INT NOT NULL AUTO_INCREMENT,
            PRIMARY KEY (idCiudad)
            FOREIGN KEY (idProvincia) REFERENCES (idProvincia)
            )


## 6.0 La representacion en el diagrama

Aqui se indicaran como se han de representar el grafo relacional junto con la normalizacion antes de dar paso al diseño fisico.

#### Representacion de las ENTIDADES / TABLAS

- **SERAN TABLAS Y LA FILA DEL ENCABEZADO SERA EL NOMBRE DE LA ENTIDAD**
- **SE ESCRIBIRA EN PLURAL Y MAYUSCULAS**
- Opcional: Poner en negrita y colores.




#### Representacion de los ATRIBUTOS/CAMPOS

- **SERAN CAMPOS DE UNA TABLA Y EL NOMBRE IRA EN MINUSCULAS, SIN ESPACIOS, NI CARACTERES ESPECIALES**
- Si el nombre es **LARGO, SE ACORTARA**
- Si el nombre incluye un **CARACTER ESPECIAL, SE ESCRIBIRA SIN EL**, modificando la palabra
- **AL LADO DEL NOMBRE SE INCLUIRA SU TIPO DE DATO Y/O RESTRICCION**
- Opcional: Poner en negrita la PK e incluir una linea que lo separe del resto.



#### Representacion de las RELACIONES/CARDINALIDADES

- **SE UTILIZAN LINEAS PARA UNIR LAS ENTIDADES, EN LOS EXTREMOS SE INDICA LA CARDINALIDAD DE LA RELACION**. Dependiendo de la cardinalidad se representa de distintas formas.

###### Relacion (0,1)



###### Relacion (1,1)



###### Relacion (1,N)



###### Relacion (0,N)
