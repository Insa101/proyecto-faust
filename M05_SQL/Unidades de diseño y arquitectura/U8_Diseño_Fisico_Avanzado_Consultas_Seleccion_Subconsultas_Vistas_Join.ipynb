## 1.0 Consultas de Seleccion

> Estas listas son las que estan basadas en la funcion `SELECT` y por tanto son *DML*

**Son todas las consultas que utilizamos para OBTENER INFORMACION DE LAS BASES DE DATOS**

> Cuando usamos el comando SELECT, le decimos al programa que ESCOJA TODO LO QUE VENGA DETRAS DE LA EXPRESION
>

### Estructura de las consultas de seleccion

La estructura de las consultas de seleccion se componen de elementos obligatorios y opcionales, segun el tipo de consulta que se quiera generar.

`SELECT` y `FROM` son **OBLIGATORIAS PARA REALIZAR UNA CONSULTA EN 1 O MAS TABLAS**

* ***SELECT*** permite especificar los valores, atributos o funciones **que se desea mostrar**
> `SELECT atributo1, atributo2`
* ***FROM*** permite indicar desde las entidades que se obtendran los datos
> `FROM nombreTabla`


`WHERE, GROUP BY, HAVING, ORDER BY` son **OPCIONALES**

* ***WHERE*** permite establecer **condiciones que deben cumplirse**
> `WHERE (condicion)`
* ***GROUP BY*** permite realizar agrupaciones de datos
> `GROUP BY atributo1 o atributo2`
* ***HAVING*** permite limitar los resultados de una consulta de agrupacion
> `HAVING (condicion de agrupación)`
* ***ORDER BY*** permite ordenar los resultados de una manera ascendente o descendente
> `ORDER BY atributo1 o atributo2 ASC/DESC ;`

## 1.1 Tipos de Consultas de Seleccion

Existen varios tipos de consultas de seleccion:

- ***Consultas básicas:*** Son las consultas de selección más sencillas. Presentan una estructura simple y no se caracterizan por emplear muchas herramientas del lenguaje SQL.

- ***Consultas de agrupación:*** Son consultas de selección más complejas que las básicas. Utilizan de forma obligatoria la cláusula `GROUP BY` y de forma opcional la cláusula `HAVING.`

- ***Subconsultas:*** Son las consultas más avanzadas, ya que permiten generar consultas dentro de otras consultas, lo que también se conoce como consultas anidadas. Su estructura es más compleja y pueden hacer uso de múltiples herramientas del lenguaje SQL.

- ***Consultas especiales:*** Son consultas cuyo propósito no es seleccionar información desde una o varias tablas de la base de datos. Estas consultas realizan una operación y muestran su resultado, por este motivo, se consideran una excepción de las consultas de selección tradicionales (consultas básicas, de agrupación y subconsultas), llegando incluso a no considerárselas como consultas de selección.

##### 1.2.2 Consultas de seleccion basicas SOBRE VARIAS TABLAS

Las consultas de seleccion de Varias Tablas **pueden replicar lar consultas anteriores** pero tendran que cumplir esta estructura

**Se especificando que atributos y de que tablas se escogen, asi como la relacion de las tablas entre si**

> **ES NECESARIO QUE LAS TABLAS ESTEN RELACIONADAS ENTRE SI PARA PODER REALIZAR CONSULTAS DE SELECCION BASICAS**

`SELECT tabla1.atributo1, tabla2.atributo1 FROM tabla1, tabla2 WHERE (tabla1.fk = tabla2.pk)`

```python
SELECT tabla1.atributo1, tabla2.atributo1
FROM tabla1, tabla2
WHERE (tabla1.fk = tabla2.pk)
```

Otro modo sencillo de realizarlo es **seleccionando los atributos e indicando las tablas mediante Alias**: `SELECT tb1.atributo, tb2.atributo FROM nombreTabla1 AS tb1, nombreTabla2 AS tb2`

```python
SELECT tb1.atributo, tb2.atributo
FROM nombreTabla1 AS tb1, nombreTabla2 AS tb2
```

#### Consultas con condicionales `CASE`

Ahora vamos a ver un ejemplo de como utilizar la clausula `CASE` para **obtener resultados calculados de varias columnas.**

Eso se realiza asi:

> `SELECT` + NombreTabla.nombreAtributo, + `CASE` + `WHEN` + nombreAtributo + Condicional (`IS NULL` / `IS NOT NULL`) + `THEN` + CALCULO + `END AS` + nombreResultados + `FROM` + nombreTabla;

Ej calculo entre fechas:

```python
SELECT 
autores.nombre AS "El pibe", 
CASE
		WHEN autores.fecha_nacimiento IS NULL THEN " Ni puto caso"
		WHEN autores.fDeceso IS NOT NULL THEN CONCAT_WS ( " - ", 
		TIMESTAMPDIFF (YEAR, autores.fecha_nacimiento, autores.fDeceso), "Fallecido")
		ELSE TIMESTAMPDIFF (YEAR, autores.fecha_nacimiento, CURRENT_DATE())
	END AS "Edad"
FROM autores;
```

#### Subconsultas

Son aquellas consultas anidadas dentro de otra consulta de seleccion.

**El caso mas habitual es SELECCIONAR ATRIBUTOS DISTINTOS QUE REQUIEREN HACER UN CALCULO EN LA CONSULTA SUPERIOR**
>Es una consulta de seleccion y con condicionales igual que las anteriores pero que permiten **realizar calculos, mostrando resultados, para generar un condicional, un calculo final,...**

`SELECT atributo1 AS alias, atributo2 AS alias,... FROM (SELECT tabla.atributo1, tabla.atributo2,... FROM nombreTabla) AS aliasResultado;`

```python
SELECT nombre as "Nombre Usuario", edad 
FROM (
    SELECT 
            clientes.nombre, 
            TIMESTAMPDIFF (YEAR, clientes.fecha_nacimiento, CURRENT_DATE()) AS edad
    FROM clientes
) AS datos_clientes;
```

> ES MUY IMPORTANTE INCLUIR UN ALIAS AL FINAL **YA QUE SIN EL ALIAS NO SE EJECUTA LA SUBCONSULTA**

Otro ejemplo es **realizar consultas CON CONDICIONALES COMPLEJOS, CON VARIAS CONDICIONES**

> Se puede realizar una consulta para **GENERAR UN 1º FILTRADO** y una 2º consulta **PARA REALIZAR UN CALCULO DE DATOS FILTRADOS**

```python
SELECT CONCAT_WS(" ", clientes.nombre, clientes.apellido1, clientes.apellido2) AS "Nombre completo"
FROM clientes
WHERE 
    clientes.paisNacimiento = (
            SELECT 
                clientes.nombre, 
                TIMESTAMPDIFF (YEAR, clientes.fecha_nacimiento, CURRENT_DATE()) AS edad
            FROM clientes) 
AND (
    TIMESTAMPDIFF(YEAR, clientes.fNac, CURRENT_DATE()) > (
        SELECT 
            clientes.nombre
        FROM clientes 
        LIKE '%Pedro%'));
```

Es posible utilizar distintas clausulas para poder realizar las subconsultas:

* Se pueden utilizar los operadores `ALL` y `ANY` **PARA REALIZAR LA COMPARACION CON TOTALIDADES O PARTICIONES DEL FILTRADO REALIZADO**

```python
# En esta consulta se busca Mostrar el nombre completo de los clientes extranjeros cuya edad sea superior a la de TODOS  los clientes españoles.

SELECT CONCAT_WS(" ", clientes.nombre, clientes.apellido1, clientes.apellido2) AS "Nombre completo"
FROM clientes
WHERE clientes.paisNacimiento IN (
    SELECT paises.pais_id
    FROM paises
    WHERE paises.nombre <> '%ESPAÑA%'
) 
AND TIMESTAMPDIFF(YEAR, clientes.fNac, CURRENT_DATE()) > ALL(
    SELECT DISTINCT TIMESTAMPDIFF(YEAR, clientes.fNac, CURRENT_DATE())
    FROM paises, clientes
    WHERE (clientes.pais_id = paises.pais_id) AND (paises.nombre = '%Nombre%')
));
```

```python
# En esta consulta se busca Mostrar el nombre completo de los clientes extranjeros cuya edad sea superior a la de ALGUNO DE los clientes españoles.

SELECT CONCAT_WS(" ", clientes.nombre, clientes.apellido1, clientes.apellido2) AS "Nombre completo"
FROM clientes
WHERE clientes.paisNacimiento IN (
    SELECT paises.pais_id
    FROM paises
    WHERE paises.nombre <> '%ESPAÑA%'
) 
AND TIMESTAMPDIFF(YEAR, clientes.fNac, CURRENT_DATE()) > ANY(
    SELECT DISTINCT TIMESTAMPDIFF(YEAR, clientes.fNac, CURRENT_DATE())
    FROM paises, clientes
    WHERE (clientes.pais_id = paises.pais_id) AND (paises.nombre = '%Nombre%')
));
```

* Se pueden utilizar los operadores `EXISTS` y `OR NOT EXISTS` **PARA HACER CONDICIONES DE SUPUESTOS QUE HAYAN SUCEDIDO O NO**

```python
# En esta consulta se  Muestra el nombre completo de los clientes que NO HAYAN RESERVADO NUNCA ningún libro. 

SELECT CONCAT_WS(" ", clientes.nombre, clientes.apellido1, clientes.apellido2) AS "Nombre completo"
FROM clientes
WHERE NOT EXISTS(
    SELECT reservas.cliente
    FROM reservas
WHERE reservas.cliente = cliente.idCliente);
```

#### Consulta de agrupacion

Son mas complejas, utilizan de forma obligatoria el `GROUP BY` y de forma opcional el `HAVING`

> Ej basico: `SELECT entidad1.atributo2, FUNCION(entidad2.atributo1) FROM entidad1, entidad2 GROUP BY (entidad1.atributo2)`

Existen varias formas de plantear estas agrupaciones:

- **Agrupando por Condicionales WHERE** : Se realiza asi cuando se quiere filtrar o **CUANDO QUIERES HACER USO DE UN VALOR QUE ES CLAVE FORANEA EN UNA TABLA O MAS CONDICIONES**, ya que permite identificar el valor de la clave con su tabla de referencia.


> `SELECT nombreTabla1.nombreAtributo AS 'nombreColumna1', count(nombreTabla2.nombreAtributo) AS 'nombreColumna2'`
> 
> `FROM nombreTabla1, nombreTabla2`
> 
> `WHERE (nombreTabla1.nombreAtributoFK = nombreTabla2.nombreAtributoPK)`
> 
> `GROUP BY nombreatributo;`

```python
SELECT paises.nombre AS 'Paises', COUNT(clientes.id_cliente) AS "Cantidad de clientes"
FROM paises, clientes
WHERE (paises.id_pais = clientes.pais_nacimiento)
GROUP BY paises.nombre;
```

Mediante `WHERE`hay 2 elementos adicionales que se pueden utilizar:

- ***Se pueden concatenar condiciones*** haciendo uso de un `AND` 

```python
SELECT 
    paises.nombre AS 'Paises', 
    COUNT(autores.pais_nacimiento) AS 'Paises autores'
FROM paises, clientes, autores
WHERE (paises.id_pais = clientes.pais_nacimiento) AND (paises.id_pais = autores.pais_nacimiento)
GROUP BY paises.nombre;
```

- ***Se pueden incluir CONDICIONES RESTRICTIVAS*** mediante `HAVING COUNT(Atributo) >/< valor`

```python
SELECT paises.nombre AS 'Paises', COUNT(clientes.id_cliente) AS "Cantidad de clientes"
FROM paises, clientes
WHERE (paises.id_pais = clientes.pais_nacimiento)
GROUP BY paises.nombre
HAVING COUNT(paises.id_pais) > 2;
```

- **Agrupando por JOIN** :  **CUANDO QUIERES HACER USO DE UN VALOR QUE ES CLAVE FORANEA EN UNA TABLA SIN CONDICIONANTES**, ya que permite identificar el valor de la clave con su tabla de referencia.

Permite **añadir equivalencia de valores entre atributos de Tablas relacionadas**

> Generalmente esta relacion se puede realizar mediante un `WHERE` + (AtributoFK = AtributoPK) pero **OMITE INFORMACION DE LA CONSULTA** (si no hay valores, nulos,...)

> `SELECT nombreTabla1.nombreAtributo AS 'nombreColumna1', count(nombreTabla2.nombreAtributo) AS 'nombreColumna2'`
> 
> `FROM nombreTabla2`
> 
> `JOIN nombretabla1  ON (nombreTabla1.nombreAtributoFK = nombreTabla2.nombreAtributoPK)`
> 
> `GROUP BY nombreatributo;`

```python
SELECT paises.nombre AS 'Paises', COUNT(clientes.id_cliente) AS "Cantidad de clientes"
FROM paises
JOIN clientes ON (paises.id_pais = clientes.pais_nacimiento)
GROUP BY paises.nombre;
```

## Operador JOIN

Es un metodo alternativo de realizar consultas de seleccion **DE VARIAS TABLAS QUE PUEDAN INCLUIR CONDICIONALES COMPLEJOS SIN ALARGAR LA SENTENCIA DEMASIADO**

Hay que conocer lo siguiente:

- A la hora de indicar `FROM` **se indicara LA TABLA PRINCIPAL O LA QUE MAS DATOS APORTA A LA CONSULTA**
- Cuando indicas `JOIN` **la consulta indica LA TABLA QUE SE VA A UNIR A LA PRINCIPAL**
- Cuando indicas `ON` **hace referencia A QUE ATRIBUTOS SON LOS QUE LOS UNEN**
- Se pueden incluir **VARIOS OPERADORES `JOIN`, PARA CADA ATRIBUTO/TABLA QUE QUERAMOS AÑADIR**
- En el `FROM` debes **UNICAMENTE INDICAR LA TABLA QUE NO INDICAS EN EL JOIN O DA ERROR**

Pero existen diferentes usos de JOIN

#### JOIN

Usaremos el comando `LEFT JOIN` **CUANDO NOS INTERESE QUEDARNOS CON LOS RESULTADOS COMUNES DE LOS OBTENIDOS DE LA UNION DE LA TABLA PRINCIPAL Y SECUNDARIA**




```python
SELECT 
    paises.nombre AS 'Paises', 
    COUNT(clientes.id_cliente) AS "Cantidad de clientes",
FROM paises, autores
JOIN clientes ON (paises.id_pais = clientes.pais_nacimiento)
GROUP BY paises.nombre;
```

#### LEFT JOIN

Usaremos el comando `LEFT JOIN` **CUANDO NOS INTERESE QUEDARNOS CON LOS RESULTADOS DE TODA LA TABLA PRINCIPAL + EL CALCULO/RESULTADOS COMUNES DE LOS OBTENIDOS DE LA UNION DE LA TABLA SECUNDARIA**



> Aqui es fundamental la seleccion de la TABLA PRINCIPAL Y SECUNDARIA, ya que difiere el resultado:
> 
> > ***TODOS LOS PAISES, CON O SIN CLIENTES ES DISTINTO A TODOS LOS CLIENTES, CON O SIN PAIS.***

```python
SELECT 
    paises.nombre AS 'Paises', 
    COUNT(clientes.id_cliente) AS "Cantidad de clientes",
FROM paises
LEFT JOIN clientes ON (paises.id_pais = clientes.pais_nacimiento)
GROUP BY paises.nombre;
```

>***TIP:*** Una consideracion importante de usar `JOIN` es **LA CONCATENACION**

>>  UNA CONCATENACION DE JOINS SOLOS **QUE TODOS LOS JOINS HAN DE CUMPLIRSE SIMULTANEAMENTE PARA OBTENER EL RESULTADO**

```python
# ESTA SINTAXIS DARA ERROR PORQUE NO TIENEN PORQUE CUMPLIRSE LOS 2 JOINS A LA VEZ

SELECT 
    paises.nombre AS 'Paises', 
    COUNT(clientes.id_cliente) AS "Cantidad de clientes",
    COUNT(autores.pais_nacimiento) AS 'Paises autores'
FROM paises
JOIN clientes ON (paises.id_pais = clientes.pais_nacimiento)
JOIN autores ON (paises.id_pais = autores.pais_nacimiento)
GROUP BY paises.nombre;
```

```python
# ESTA SINTAXIS ES CORRECTA PORQUE ESCOGE AQUELLOS QUE SI CUMPLEN LAS CONDICIONES PERO NO SIMULTANEAMENTE

SELECT 
    paises.nombre AS 'Paises', 
    COUNT(clientes.id_cliente) AS "Cantidad de clientes",
    COUNT(autores.pais_nacimiento) AS 'Paises autores'
FROM paises
LEFT JOIN clientes ON (paises.id_pais = clientes.pais_nacimiento)
LEFT JOIN autores ON (paises.id_pais = autores.pais_nacimiento)
GROUP BY paises.nombre;
```

##### RIGHT JOIN

Usaremos el comando `RIGHT JOIN` **CUANDO NOS INTERESE QUEDARNOS CON LOS RESULTADOS DE TODA LA TABLA SECUNDARIAL + EL CALCULO/RESULTADOS COMUNES DE LOS OBTENIDOS DE LA UNION DE LA TABLA PRINCIPAL**



```python
SELECT 
    paises.nombre AS 'Paises', 
    COUNT(clientes.id_cliente) AS "Cantidad de clientes",
    COUNT(autores.pais_nacimiento) AS 'Paises autores'
FROM paises
RIGHT JOIN clientes ON (paises.id_pais = clientes.pais_nacimiento)
RIGHT JOIN autores ON (paises.id_pais = autores.pais_nacimiento)
GROUP BY paises.nombre;
```

#### INNERS y OUTTERS

Por ultimo exsiten una variables de JOINS que son las siguientes:

- `INNER` son aquellas variantes **QUE SOLO INCLUYEN LAS COMUNES, ES DECIR, ES = AL JOIN**
- `OUTERS` son aquellas variantes **QUE SELECCIONAN LAS EXCLUSIONES O NO COMUNES**

Existen 2 variantes :

- Las `FULL OUTER`/ `LEFT OUTER`/ `RIGHT OUTER` que excluyen datos
- La `FULL INNER`que incluyen datos

##### LEFT OUTER JOIN

**UNICAMENTE SELECCIONA REGISTROS DE LA TABLA PRINCIPAL QUE NO SE COMPARTAN CON LA TABLA SECUNDARIA**



> ***VIP:*** Hay que incluir LA RESTRICCION `IS NULL` como condicion.

```python
SELECT paises.nombre AS "Países"
FROM paises
LEFT OUTER JOIN clientes ON (paises.idPais = clientes.paisNacimiento)
WHERE clientes.idCliente IS NULL
GROUP BY paises.nombre;
```

##### RIGHT OUTER JOIN 

**UNICAMENTE SELECCIONA REGISTROS DE LA TABLA SECUNDARIA QUE NO SE COMPARTAN CON LA TABLA PRINCIPAL**



> ***VIP:*** Hay que incluir LA RESTRICCION `IS NULL` como condicion.

```python
SELECT CONCAT_WS(" ", clientes.nombre, clientes.apellido1, clientes.apellido2) AS "Clientes"
FROM reservas
RIGHT OUTER JOIN clientes ON (clientes.idCliente = reservas.cliente)
WHERE reservas.cliente IS NULL
GROUP BY clientes.nombre;
```

##### FULL OUTER JOIN 

**SELECCIONA TODOS REGISTROS DE LA TABLAS AUNQUE NO EXISTA COMUNES, ES DECIR, ES LO MISMO QUE COGER LAS 2 TABLAS SIN CONDICIONALES**



> NO SE PUEDE UTILIZAR EN MYSQL

```python
SELECT * FROM entidadA
FULL OUTER JOIN entidad B ON (entidadA.PK = entidadB.FK);
```

##### FULL INNER JOIN 

**SELECCIONA TODOS REGISTROS DE LA TABLAS NO COMUNES Y EXCLUYE LAS COMUNES**



```python
SELECT * FROM entidadA
FULL OUTER JOIN entidad B ON (entidadA.PK = entidadB.FK)
WHERE (entidadA.PK IS NULL) AND (entidadB.PK IS NULL)
```

## VISTAS

Las vistas son elementos para generar *entidades virtuales* basadas en resultados de una consulta concreta.

**Son tablas visuales sin guardar registros que muestran los resultados de una consulta concreta**

Se utiliza el comando `CREATE` que es un ***DDL***

> `CREATE OR REPLACE VIEW nombreVista AS`
>> `SELECT [datos que se desean seleccionar]`
>> 
>> `FROM [tablas desde la que se desea seleccionar los datos]`
>> 
>> `WHERE [condiciones];`

```python
CREATE VIEW clientesESP AS
SELECT 
    CONCAT_WS(' ', clientes.nombre, clientes.apellido1, clientes.apellido2) AS "Clientes",
    COUNT(reservas.idReserva) AS "Cantidad de libros reservados"
FROM clientes
LEFT JOIN reservas ON clientes.idCliente = reservas.cliente
INNER JOIN paises ON clientes.paisNacimiento = paises.idPais
WHERE paises.nombre = 'España'
GROUP BY clientes.idCliente, clientes.nombre, clientes.apellido1, clientes.apellido2
ORDER BY clientes.nombre ASC;
```
