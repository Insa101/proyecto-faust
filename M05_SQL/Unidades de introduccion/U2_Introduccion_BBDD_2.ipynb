## 1.0 ¿Que son los DBMS y cuales hay?

Los DBMS como hemos visto, son *los gestores de bases de datos* **programas encargados de implementar, almacenar, modificar, gestionar y recuperar los datos de una BBDD**

Sus principales funciones son:

- **Implementar un entorno** en el que se puedan crear BBDD
- **Manipular datos**: incluye mecanismos para analizar y gestionar datos almacenados
- **Administra la BBDD**: a traves de diversas herramientas
- **Propociona mecanismos de seguridad y gestion del acceso a los datos**

### Tipos de DBMS

A continuacion mencionaremos los principales DBMS del mercado:

1. ***MySQL***:  Es el sistema gestor de bases de datos relacionales más utilizado en el desarrollo de páginas web y aplicaciones creadas como software libre.
> No trabaja de forma eficiente con bases de datos muy grandes.

2. ***MariaDB***: basado en *MySQL* cuenta con la mayoria de las cracteristicas e incluye varias extensiones, es compatible con el y no tiene desventajas.
   
3. ***SQLite***: Es una biblioteca escrita en lenguaje C, implementa un sistema gestor y permite transacciones sin necesidades dd configurar un servidor
> No trabaja de forma eficiente con bases de datos muy grandes.
>
> Esta muy extendida ya que es de codigo abierto

4. ***PostgreSQL***: Es un DBMS orientado a objetos.
> Es ineficiente para BBDD pequeñas, esta orientado a gestionar grandes volumenes de datos.

5. ***Microsoft SQL server***: Es un DBMS basado en Transact-SQL, permite hacer uso de grandes cantidades de datos de forma simultanea.

6. ***SQL Server de Oracle***: Es el DBMS mas extendido en el ambito empresarial, tambien el considerado mas robusto y completo.

7. ***MongoDB***: Es un DBMS para BBDD No relacionales orientado a ficheros, es el mas utilizado y extendido.

> Es inadecuado para realizar transacciones complejas.

### Motores de almacenamiento

Otro aspecto a tener en cuenta son este componente de los DBMS, **que se encargan de almacenar, gestionar y recuperar toda la informacion de las tablas que forman las BBDD Relacionales**

> Ademas mejoran el rendimiento y añaden mayor seguridad en los datos.

Se puede conocer que motor utilizan nuestras BBDD a traves del siguiente comando:

`SHOW ENGINES`

Los motores mas utilizados son:

* ***InnoDB:*** Es el motor por defecto. **Es un motor de almacenamiento transaccional (conforme a ACID), con capacidades para confirmar y cancelar transacciones.** Ademas permite el uso de claves foráneas, algo fundamental para mantener la integridad de la base de datos.

> ACID hace alusion a la capacidad de la BBDD de **realizar todas las transacciones de golpe o ninguna, asegurando la integridad de los datos**

* ***MyISAM:*** proporciona un rápido almacenamiento y recuperación de datos. Las tablas que utilizan este motor no permiten el uso de transacciones. **Utilizado en BBDD Relacionales estaticas***

## 2. ¿Que son los lenguajes de consulta y cuales hay?

> Las BBDD utilizan un lenguaje de pseudoprogramacion conocido como **lenguaje de consultas**

Estos lenguajes son aquellos utilizados para **acceder, consultar y gestionar la informacion en BBDD** y diversos sistemas de informacion **a traves de comandos.**

#### Tipos de comandos

Existen 3 grupos principales de comandos en los lenguajes de consulta:

1. ***DDL (Data Definition Language)***: Son los comandos utilizados por el **administrador de la BDD** para la **creacion, modificacion y eliminacion de la estructura** de la BBDD y sus elementos principales.

`CREATE`
`ALTER`
`DROP`

2. ***DML (Data Manipulation Language)***: son los comandos utilizados para la **inserción, modificación, borrado y selección de los datos.**

`INSERT`
`DELETE`
`UPDATE`
`SELECT`

3. ***DCL (Data Control Language)***: son los comandos para utilizar la **gestion de privilegios de usuarios** que pueden acceder a la BBDD.

`GRANT`
`REVOKE`

### Tipos de lenguajes

Ademas existen varios tipos de lenguajes de consultas, a parte de SQL que es el mas extendido, hay varios mas:

* ***SQL***: es el lenguaje de consulta más utilizado por los sistemas gestores de bases de datos relacionales.
* ***XQuery***: lenguaje de consulta para fuentes de datos XML.
* ***OQL***: lenguaje de consulta de objetos utilizado principalmente en bases de datos orientadas a objetos.
* ***GraphOL***: es un lenguaje de código abierto que permite trabajar con APIs.

## 3.0 El ciclo de vida y diseño de una BBDD Relacional

Por ultimo, vamos a ver un vistazo general del proceso de que tendremos que realizar al configurar una BBDD Relacional

> Mas adelante ampliaremos informacion sobre estas cuestiones, pero sirva este apartado para explicar el proceso en general

Daremos respuesta a la pregunta *¿Como se diseña una BBDD?*

Consta de 4 etapas:

1. Etapa: ***Análisis***, define los **objetivos que se desean conseguir** con la creación de la base de datos y **establece los requisitos** que esta debe cumplir.
   
2. Etapa: ***Diseño y moldeamiento:***, en esta etapa **se definirán los elementos que constituirán el sistema de información** a implementar.

> Esta etapa **define los objetos (entidades), atributos, relaciones que se van a utilizar.**

3. Etapa: ***Implementación:*** esta etapa **implementará el diseño y los modelos generados** en la etapa anterior en el sistema gestor de bases de datos, dónde además **realizará varias pruebas** para comprobar que el funcionamiento sea tal y como se había planteado en la fase de análisis.

4. Etapa: ***Uso y gestión de la base de datos:*** etapa final del ciclo de vida. En esta parte, los usuarios podrán hacer uso de la base de datos, mientras que los desarrolladores realizarán las acciones necesarias para mantenerla actualizada y vigente.

### ¿Como se realiza el diseño y moldeamiento en una BBDD Relacional?

De todas las Etapas la que cobra una importancia mas critica es **la 2º etapa: *Diseño y moldeamiento:*** ya que supone el punto donde se define como se va a relacionar la informacion y como se debe almacenar para que el sistema sea eficiente.

Esta etapa se puede diviri en 4 sub etapas, **que vamos a estudiar en profundidad en el resto de modulos**:

1. ***Análisis***: aqui se busca recopilar la informacion necesaria e imprescidndible para crear la BBDD y sobre todo **entender las necesidades que debe satisfacer**
2. ***Diseño Conceptual***: aqui **se genera un diagrama**que representa graficamente los datos antes definidos y sus relaciones
3. ***Diseño Lógico***: aqui se transforma el diagrama obtenido en un **nuevo diagrama que contenga la estructura final** de la BBDD
4. ***Diseño Fisico***: aqui **se llevara el ultimo diagrama a un DBMS para crear la BBDD.**

## 4.0 Principios del diseño de la BBDD

Consta de 3 principios por los que se rige el diseño de una BBDD.

#### 1. Principio: La informacion duplicada es incorrecta

Hay que evitar disponer de datos redundantes, ya que duplica el espacio necesario y aumenta la probablidad de errores e incoherencias.

La norma dice ***que en cada interseccion que genera un registro, ese registro ha de ser UNICO***

> ej: no disponer de 2 valores distintos en 1 misma celda.

#### 2. Principio: La exactitud e integridad es importante

Si la BBDD dispone de datos incorrectos, los informes tambien seran incorrectos.

> ej: disponer de 2 claves principales en una misma entidad para identificar el registro.

#### 3. Principio: Las colummnas no claves seran independientes entre si

Las columnas no clave no pueden generar dependencia entre ellas, unicamente con la clave principal

> ej: una columna descuento depende de otra columna que se llama precio, si cambia el precio, cambia el descuento.
