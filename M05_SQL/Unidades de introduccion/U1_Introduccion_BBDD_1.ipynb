## 1. ¿Que son las BBDD?

La definicion estricta es que son **un conjunto de informacion ordenada de forma sistematica y que pertenece a un mismo contexto**

Tambien se puede considerar como **un conjunto de datos almacenados que permite que la información pueda ser consultada o analizada tras registrarse.**

> Nacen por la necesidad de **preservar y almacenar informacion**, para consultarla y analizarla en el futuro.

En el entorno digital, para poder interactuar con bases de datos (o BBDD) **utilizaremos programas de gestion de BBDD (*DBMS* - Database Management System)**

> Tambien es de tener en cuenta que no todas las BBDD **son gestionadas por estos programas**

### 1.1 Ventajas y Desventajas de las BBDD

Una BBDD tiene una serie de ventajas y desventajas en las organizaciones, frente a otros sistemas de almacenamiento (como sistemas de ficheros) a tener en cuenta:

##### Ventajas

* **Control sobre la redundancia de datos:** Los datos estan integrados con los ficheros y no se almacenan varias copias de los mismos datos. En una BBDD no se puede eliminar la redundancia completamente, ya que en ocasiones es necesaria modelar las relaciones entre datos.
  
* **Consistencia de datos:** Elimina o controla la redundancias. Si un dato se almacena una vez, cualquier actualizacion se debe realizar solo una vez y esta disponible para todos los usuarios inmediatamente. Si esta duplicado, el propio sistema puede encargarse de garantizar que las copias se mantengan consistentes.

* **Compartir datos:** Los datos pertenecen a la organizacion y no a personas individuales, pudiendo ser compartidas.

* **Mantenimiento de estandares:** Facilidad de mantener estandares necesarios, estableciendo un formato de datos para facilitar su intercambio,...

* **Mejora en la integridad de datos:** Hace referencia a la validez y consistencia de los datos. Restringue los datos para evitar que se malversen. Las restricciones y se aplican a los datos, asi como a sus relaciones y es el sistema de gestion BBDD.

* **Mejora en la seguridad:** Es la proteccion de la BBDD frente a usuarios no autorizados.

* **Mejora en la accesibilidad a los datos:** Muchos gestores añaden lenguajes de consulta y visualizadores para consultar sobre los datos, sin necesidad de programar.

* **Mejora en la productividad:** Muchas de las gestion de programacion ya las hace el programa gestor, lo que permite centrarse en otro solicitudes de usuario.

* **Mejora de mantenimiento:** Las descripciones de los datos no dependen de los mismos y por tanto un cambio en la estructura de los mismos, no afecta a las aplicaciones.

* **Aumento de la concurrencia:** Varios usuarios pueden acceder a los mismos archivos simultaneamente manteniendose la integridad de los mismos.

* **Mejora en los servicios de copias de seguridad:** Los sistemas gestores de bases de datos garantizan una mejor recuperacion de los datos en caso de fallo de las copias.

##### Desventajas

* **Tamaño y peso:** Una BBDD requiere mucho espacio en disco, provocando que sea mas lenta e inexacta en las busquedas y recuperacion de la informacion.

* **Complejidad:** Requiere capacitacion, asesoria y un aprendizaje complejo para su manejo.

* **Costo:** Algunos gestores y productos de BBDD suelen ser costosos.

## 2. Clasificacion de BBDD

Las BBDD se pueden clasificar segun **la variabilidad de los datos** ( en funcion de si los datos cambian o no):

#### 2.1 *BBDD Estaticas*

Son BBDD **unicamente de lectura**, los datos no se modifican.

> Se usan para almacenar datos historicos o hechos invariables, para el mundo estadistico,...
>
> ej: BBDD INE

##### Ventajas

* No necesitan mantenimiento
* Las consultas suelen ser sencillan, al ser datos fijos
* Tienen total validez a lo largo del tiempo
  
##### Desventajas

* No sirven para almacenar datos que requieran modificaciones
* No son flexibles y se limitan a un tipo de dato

#### 2.2 *BBDD Dinamicas*

Son BBDD **donde los datos pueden variar, permitiendo operaciones de consulta, actualizar o borrar**. 

> Permiten establecer relaciones entre los registros y sus campos, para poder realizar mejor consultas sobre los datos.
>
> ej: CRM's, ERP's...

##### Ventajas

* Son mas flexibles, permiten adaptar sus datos y la estructura
* Permiten analizar datos de manera agil y sencilla
* Tienen capacidad de adaptarse a los cambios de la informacion
  
##### Desventajas

* Necesitan mas mantenimiento debido a que hay que estar actualizandolas constantemente,provocando errores de datos incompletos, desactualizados o erroneos
* Al actualizarse, resulta dificil analizar datos historicos



### 2.3 Otras clasificaciones de BBDD 

#### 2.3.1 *Segun el contenido registrado*

1. ***Bases de datos de texto completo:*** Son aquellas en las que se permite **realizar consultas sobre el texto íntegro que se encuentra almacenado.**
> Suelen utilizarse para registrar investigaciones.

2. ***Bases de datos bibliográficas:*** Son bases de datos utilizadas como **registros que se encargan de almacenar la información básica y localización** de los elementos a los que están asociadas, pero **no su contenido.**
   
3. ***Bases de datos directorios:*** Son bases de datos de uso cotidiano que permiten ordenar y organizar la información de forma sencilla.

## 3. Diferencias entre sistemas gestores de BBDD y hojas de calculo

Ambas herramientas pueden tener similitudes pero son sus diferencias las que impiden que hacen que no puedan utilizarse del modo que fue concebida la otra. Estas diferencias son:

##### 1. **La forma de registrar y manipular la informacion**

- ***Una hoja de calculo***: Esta diseñada para un pequeño grupo de usuarios, **sin facilitar el acceso a multitud de usuarios** (aunque se pueda)
- ***Una DBMS***: Esta diseñada para un acceso multiusuario, **facilitando la consulta simultanea**

> El principal matiz es que **en una hoja de calculo al modificar o actualizar una celda con informacion, impides que otro usuario interactue con esa celda hasta que has terminado la operacion, mientras que en las DBMS no**
>
> Ademas en las DBMS **hasta que no has acabado la operacion no se modifica el registro, cosa que en una hoja de calculo puede pasar**

##### 2. **La cantidad de datos que se pueden almacenar**

- ***Una hoja de calculo***: Disponen de una capacidad limitada de datos
- ***Una DBMS***: Tienen gran capacidad de almacenamiento de datos


## 4. Modelos de BBDD

Existen 2 tipos de modelos de BBDD: ***Relacionales y No Relacionales***

### 4.1 BBDD Relacionales (SQL)

Es una BBDD que tiene como organizacion de los datos **relaciona los diferentes tipos de datos**

Estas BBDD **son tabulares**

> Los datos en si y las tablas que componen los datos.
>
> Ej: SQL

Los DBMS de este tipo de BBDD utilizan **un lenguaje de consultas estructurado, como es *SQL***

#### Lenguaje SQL

**Es un lenguaje relacional de consultas, que consiste en la definicion del tipo de dato para que luego podamos manipularlo y controlarlo**

> Nos va a permitir insertar, modificar y borrar datos
>
> Nos va a permitir crear diferentes tipos de BBDD, de tablas,...
>
> **Nos va a permitir crear las relaciones entre los datos**



### 4.2 BBDD No Relacionales (NoSQL)

Es una BBDD que permite almacenar grandes cantidades de datos, estructurados o no estructurados para su posterior consulta y analisis

### 4.3 BBDD Orientada a objetos

Son BBDD que utilizan tecnologia de programacion orientada a objetos donde se almacenan el objeto en si en lugar de la informacion sobre si mismo.

### 4.4 BBDD multidimensionales

Son BBDD con datos estructurados almacenados en hipercubos de n dimensiones.

> Se utilizan para el analisis multidimensonal jerarquizando dimensiones

### 4.5 BBDD deductivas

Son BBDD que permiten incorporar deducciones a traves de una inferencia. Sus principios estan fundamentados en la logica matematica

### 4.6 BBDD jerarquicas

En estas BBDD la informacion se almacena en un orden de importancia, como esquemas en forma de arbol.
