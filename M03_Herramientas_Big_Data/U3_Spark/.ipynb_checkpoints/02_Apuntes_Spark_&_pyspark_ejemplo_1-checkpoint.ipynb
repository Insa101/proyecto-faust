## 1 Spark y Python: Libreria *pyspark*

Como decia, para poder hacer uso de Spark en Python, sera necesario instalar la libreria ***pyspark***:

> ***pip install pyspark***

Esta libreria incluye una serie de caracteristicas muy interesantes para la gestion y analitica de datos en Python:

- Posibilidad de computar directamente en memoria

DataFrame[fixed acidity: double, volatile acidity: double, citric acid: double, residual sugar: double, chlorides: double, free sulfur dioxide: double, total sulfur dioxide: double, density: double, pH: double, sulphates: double, alcohol: double, quality: int, type: string]

### 3.3 Seleccionar una muestra de datos y ordenarlos: *.select(), .sort(), .limit()*

1º Podemos seleccionar los datos por sus columnas a traves del metodo ***.select()***:

### 3.4 Explicacion de los pasos realizados para el calculo: *.explain()*

Podemos necesitar comprobar al realizar un calculo **como Spark ejecuta una secuencia de operaciones**, para ello hacemos uso de *.explain()*

> ***nombre variable fichero.groupBy("nombre columna")."otra funcion"("nombre columna").explain()***

Nos otorga una serie de informacion, siempre haciendo el analisis de lo mas interno a lo mas externo:

- *HashAggregate*: corresponde al groupBy, ejecutado en cada nodo por separado, **indica la porción de datos que hay en cada nodo**
- *Exchange*: **Indica el intercambio entre los nodos de los datos de cada grupo**, de manera que todos los conteos parciales de un mismo grupo vaya a parar a un mismo nodo.
- *HashAggregate*: **corresponde a la suma de todos los conteos parciales de cada nodo y obtener el resultado final de cada grupo**



## Explicacion Libreria *pyspark.sql.functions as F* para la transformacion de datos

Para poder avanzar en el ejemplo, es necesario traer a nuestro entorno esta libreria para poder utilziar las funciones de trnaformacion de datos.

> Son las herramientas de transformacion sobre los Dataframes de Spark.

El modo es como el de una libreria estandar en python:

> ***import pyspark.sql.functions as F***

#### Estructuras sintacticas basicas de la libreria: *F.col() y F.lit()*

Antes de ver las funciones complejas, hay que entender las dos unidades básicas que componen casi cualquier expresión en Spark:

- ***F.col("nombre columna")***: Representa una columna existente en el DataFrame. Es una referencia, no el dato en sí.

- ***F.lit(valor)***: Representa un valor literal (constante).

> Si quieres sumar el número 5 a toda una columna, no puedes sumar un entero de Python a una columna de Spark; debes convertir ese 5 en una "columna constante" de 5s usando F.lit(5).

#### Tipos de funciones

Podemos agrupar las cientos de funciones de F en familias lógicas:

- **Funciones de Agregación**: Se usan tras un groupBy. Reducen muchas filas a un solo valor.

> ***F.count(), F.sum(), F.avg(), F.max().***

- **Funciones de String (Cadena)**: Operan sobre columnas de texto.

> ***F.upper(), F.lower()***: Mayúsculas/minúsculas.
>
> **F.substring()***: Extraer trozos de texto.
>
> ***F.split()***: Dividir texto en un array (como el split de Python).

- **Funciones Condicionales (El if-else de Spark)**: En Spark no se usa if/else de Python dentro de un DataFrame. Se usa *F.when().otherwise()*.

> ***F.when(condición, valor_si_cierto).otherwise(valor_si_falso)***

- **Funciones de Ventana (Window Functions)**: Son avanzadas pero vitales. Permiten cálculos sobre "ventanas" de datos (ej. medias móviles, rankings) sin colapsar las filas como en una agregación normal.

> ***F.row_number(), F.rank(), F.lag().***

#### La justificación técnica: Catalyst Optimizer

**¿Por qué usar F.upper(col) en lugar de hacer un bucle for en Python?**

Cuando usas F, estás invocando código escrito en Scala/Java que ya está optimizado. Spark toma todas tus instrucciones F, las pasa por el Catalyst Optimizer (el cerebro de Spark), y genera el código Java más eficiente posible para ejecutarse en paralelo en todos los nodos.

> Si usaras funciones de Python puro (UDFs), Spark tendría que sacar los datos de la memoria optimizada, pasarlos a Python, procesarlos y devolverlos, lo cual es mucho más lento (proceso conocido como serialization overhead).

## 4 Ejemplo con pyspark: Manipulacion de datos

> **A partir de aqui es necesario que hayamos importado la libreria *pyspark.sql.functions as F***

A partir de aqui empezaremos a realizar acciones de transformacion y manipulacion de datos.

### 4.1 Agrupaciones de datos y calculo simplificado: *.groupBy()* + funciones

La mas basica es hacer agrupaciones para realizar calculos o sacar valores estadisticos, para ello utilizaremos ***.groupBy()***

>***nombre variable fichero.groupBy("nombre columna")."otra funcion"("nombre columna").show()***
>
>> Es necesario utilizar *.show()* porque *.groupBy()* devuelve un objeto, pero no la muestra.

Las funciones que podemos aplicar para a los groupby son asi:




### 4.2 Agregaciones de datos agrupados: *.groupBy() + .agg()*

Otras operaciones basicas de Sparks son las agregaciones de grupos de datos mediante **.agg()**.

**Las agreaciones permiten realizar calculos de datos agrupados en groyupby y crear una nueva columna con los reusltados**

> ***nombre variable fichero.groupBy("nombre columna").agg(F."funcion"("nombre columna").alias("nombre nueva columna")).show()***
>
> > Aqui utilizamos la funcion ***.alias()*** para dotar de un nombre concreto a la nueva columna

### 4.3 Creacion de nuevas columnas a partir de otras: *.withColumn() y .withColumnRenamed()*

Otra funcion habitual es la creacion de columnas a partir de otras, **que alberguen operaciones de datos entre columnas existentes**

Esto se puede realizar con 2 funciones:

- ***.withColumn()***: Crea una columna nueva y **modifica los datos** (en base a un calculo o los sobreescribe)

> ***nombre variable fichero.withColumn("nombre de la columna nueva", (F.col("nombre columna")*/+-... ).show()***

- ***.withColumnRenamed()***: Crea una columna nueva, **modifica las etiquetas y no modifica los datos**

> ***nombre variable fichero.withColumnRenamed("nombre columna viejo", "nombre columna nuevo").show()***

### 4.4 Filtracion de datos: *.where() y .filter()*

Aunque ya lo hemos visto antes, **podemos utilizar esta sentencia para filtrar datos o generar condicionales**

Utilizaremos la expresion *.filter()* o su alias ***.where()***

> ***nombre variable fichero.where(F.col("nombre columna a filtrar) , "condicion").show()***
>
> > Se utiliza ***.where()*** como alias ya que se asemeja a la expresion de SQL para la misma funcion. ***WHERE()***

### 4.5 Transformar datos con condicionales: *F.where().otherwise()*

Para terminar, vamos a revisar como funcionan las condicionales en Spark a traves de las funciones ***F.where().otherwise()***

**Generas valores nuevos basandote en condiciones, dentro de una trnasformacion de una columna** (*.withColumn(), .select()*)

> ***nombre variable fichero.withColumn("nombre columna nueva , F.when(F.col("nombre columna") "condicion", "etiqueta si se cumple").otherwise("etiqueta si no se cumple").show()***
