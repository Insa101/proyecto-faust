# Apuntes Spark: Como ejecutar la aplicacion den el cluster de Spark

Hasta ahora hemos estado utilizando Spark a traves de la consola y con la aplicacion ***pyspark*** que emula la arquitectura y el lenguaje a muy alto nivel de Python.

Pero este tipo de ejecucion prepara una *SparkSession* en local para ejecutar los archivos

De modo que ***¿Como podemos utilizar los recursos y elementos de la arquitectura cluster de Spark?***

Lo haremos a invocando a la *aplicacion de Spark*.

## *spark-submit*: El ejecutor de la aplicacion

El comando ***spark-submit*** se encarga de:

- Enviar nuestro programa o aplicación a la plataforma Spark
- Asignar los recursos necesarios para la aplicación
- Distribuir el código entre los nodos del cluster donde se vaya a ejecutar
- Proceder con la ejecución.

Este comando necesita 2 argumentos para configurarse adecuadamente:

- 1º Es necesario indicarle el punto de entrada al cluster de Spark con la opción ***-master.***

> Generalmente, **si no se tiene ningun cluster preparado**, se lanza en *local* de nuestra propia maquina, ***-master local***, sino habria que asigarle una direccion IP o URL.

- 2º Hay que indicarle **la ruta de acceso al fichero con el script en python**

> Por lo que **es necesario tener preparado previamente el script y guardado**

La sentencia entonces tiene un aspecto parecido a este:

> ***spark-submit --master local ./"direccion donde se ubica el script de python"***

### Generar un script de python que pueda ejecutarse a traves de la aplicacion Spark: *SparkSession* y *.write.csv()*

### 1º Generar una sesion en SparkSession

Una caracteristica que hay que incluir en el script de python para que pueda ejecutarlo la aplicaicon de Spar directamente, es **el creador de una SparkSession** 

> La SparkSession es el *punto de acceso del usuario a la plataforma* por lo que es necesario configurarla. **pyspark la genera automaticamente, pero la consola no**

Para ello habra que hacer uso de la libreria ***pyspark.sql*** y el modulo ***SparkSession***

> ***from pyspark.sql import SparkSession***

Con esto lo que conseguimos es poder escribir **una solicitud a Spark para que nos asigne una sesion**

> ***spark = SparkSession.builder.master("el master que vayamos a indicar/ local").appName("nombre de la app que queramos").gettOrCreate()***

### 2º Guardar los resultados de la ejecucion del codigo.

Otro aspecto fundamental es **guardar los resultados de los calculos generados con python a traves de Spark** ya que sino se ejecutaria el archivo en memoria pero no generaria un fichero revisable o utilizable.

> Recuerda que Spark solo ejecuta el script en el cluster, sino en el script no se genera un archivo, **solo calculara en memoria y desaparecera**

Para ello sera necesario utilizar de 2 funciones: ***.coalesce() y .write.csv()***

- ***.coalesce()***: Indica a Spark **el numero de particiones que queremos hacer con el archivo**. Por defecto indicaremos 1.
- ***.write.csv()***: Graba los resultados en **un tipo de archivo CSV**

>***"nombre del modelo evaluado".select("prediction", "otra columna que nos interese",...).coalesce(1).write.csv(r'file://"direccion escritorio donde guardarmos el fichero"/"nombre del fichero nuevo", header = True, sep =";", mode ="overwrite")***

Con los parametros *header = True, sep =";", mode ="overwrite"*, nos aseguramos de:

- ***header = True:*** Los nombres de las columnas sean los mismos del archivo a los del dataframe
- ***sep =";"***: Los datos se separen por ; y se puedan reconocer
- ***mode ="overwrite"***: Se sobrescriba el archivo "en blanco" y asi se graben conforme a los resultados obtenidos
