#### Gestionando valores ausentes o incompletos.

Este mensaje se produce porque el VectorAssembler, que utilizamos para
empaquetar y construir los vectores de variables explicativas a partir
de los datos, se ha encontrado con valores nulos (NaN o NULLs) en alguna
de las columnas que queremos utilizar como regresores. El
VectorAssembler no sabe qué debe hacer con valores nulos o ausentes, y
por defecto genera una excepción.

Sabemos que la columna mpg no tenía valores nulos porque no hubo errores
antes. Podemos ver cuál de las dos nuevas variables tiene nulos:

```python
auto.filter(auto["horsepower"].isNull()).count()
```

`5`

```python
auto.filter(auto["acceleration"].isNull()).count()
```

`0`

En este caso la variable culpable es la de la potencia. Existen varias
opciones para lidiar con datos incompletos. La más sencilla es descartar
aquellas filas que tengan algún valor nulo. Para ello podemos usar el
método na.drop() de un DataFrame. Un ejemplo con nuestro DataFrame auto:

```python
auto_clean = auto.na.drop()

auto_clean.filter(auto["horsepower"].isNull()).count()
```

`0`

Puedes ver más detalles sobre esta y otras funciones para manejar
valores nulos en <http://spark.apache.org/docs/2.1.0/api/python/pyspark.sql.html#pyspark.sql.DataFrameNaFunctions>
