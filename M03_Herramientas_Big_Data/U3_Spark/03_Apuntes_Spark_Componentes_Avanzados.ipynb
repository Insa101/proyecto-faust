## 1.0 Spark SQL

##### 1.1 Definicion

**Es un interprete  *a mas alto nivel* para realizar consultas**, que incorpora una serie de estructuras de datos, a partir del **SchemaRDD**

> Esta estructura nos permite añadir diferentes tipos de datos estructurados y no estructurados
>
> Se genero esta libreria al desarrollar los DataFrame de Spark y te permite realizar funciones sobre estas estructuras.

##### 1.2 Caracteristicas

Tiene las siguientes caracteristicas:

- 1º Trabaja con **DataFrame propios** (datos tabulares, como datasets o dataframes)
- 2º El contenido de un DataFrame de Spark **puede estar repartido entre varias maquinas o nodos del cluster**
- 3º El contenido de un DataFrame de Spark **se guarda en RDDs y adquiere sus propiedades**
- 4º Los DataFrame de Spark **tienen una clave API, que contxtualiza las operaciones y funcionalidades de los datos**

##### 1.3 Funcionalidades

Spark SQL te permite **registrar cualquier DataFrame como una tabla, ejecutar consultas y calculos sobre ella, usando lenguaje SQL**

##### 1.4 Datasets de Spark

A parte de los DataFrames, tambien es posible generar estructuras de Datasets en Spark.

**Son representaciones de registros, donde cada registro es un objeto de una clase definida por el usuario**
> Esto nos permite incluir funcionalidades especificas cuando usamos los datos como objetos, pero con capacidad de usarlos como DataFrames cuando los necesitemos.

Estas estructuras se utilizan con *Scala y Java* unicamente ya que **no son lenguajes tipo dinamico** (el tipo de dato ha de declararse antes de utilizarlo)

## 2.0 SparkMLlib

**Es una herramienta habilitada para realizar machine learning sobre Spark**

> Dispone de implementaciones para aprendizaje estadistico, desde algoritmos de regresion o clasificacion, o algoritmos de clustering, recomednacion y *deep learning*

Ofrece todos los componentes necesarios para hacer a escala big data:

- 1º el preprocesado y preparacion de los datos
- 2º definicion de modelos y entrenamiento
- 3º validacion y ejecucion de modelos

## 3.0 Spark Streaming

**Es una herramienta que posibilita el analisis de un flujo constante de datos, *al vuelo*, a traves de grupos mas pequeños**

Sus caracteristicas son:

- 1º Capacidad de recibir datos **de distintas fuentes en tiempo real**
- 2º Tolerancia a fallos
- 3º Baja latencia para procesar datos conforme llegan, **contestando con eficiencia**
- 4º Se puede realizar las mismas operaciones que sobre un dataframe
- 5º Permite aplicar algoritmos de machine learning

Algunas aplicaciones para esta herramienta son:

- Analisis de *logs* para monitorizar servidores
- Analisis en tiempo real de *feeds* o mensajes en redes sociales
- Procesar señales proporcionadas por sensores de distintos dispositivos conectados a internet (IoT)

## 4.0 Spark GraphX

**Es una herramienta que te permite procesar grafos *al mas alto nivel***

Hay que entender los grafos, no como graficos unicamente, sino como **procesos que cumplen las normas de relacion y direccion de nodos**

> Es decir, en el proceso 1 nodo solo se conecta en una direccion, un proceso empieza en un nodo y termina en oro distinto y es posible recuperar informacion o calcular de nuevo unicamente volviendo atras en el grafo siguiendo los pasos anteriores del nodo.

Es la base de **los motores de busqueda y de los algoritmos de las redes sociales** ya que permiten:

- Conectar los nodos(Usuarios, contenidos, aplicaciones...) mediante enlaces(*me gusta, envio de un mensaje,...*)
- Realizar calculos sobre estas interacciones entre nodos, para determinar contenido relevante, recomendaciones, usuarios influyentes,...

GraphX en definitiva **ofrece una serie de funciones de alto nivel, para aprovechar las caracteristicas de los RDDs y los DAGs, ocultando los detalles de programacion complejos**
