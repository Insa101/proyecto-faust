### Proceso de Machine Learning: Conceptos basicos

La relacion de los conceptos de manera simplificada, es la siguiente:

- Con los datos organizados, **escogemos *features* que permiten generar algoritmos** (las hipotesis de prediccion)
- Estos algoritmos son **entrenados con un conjunto de datos para generar un modelo predictivo**
- **El modelo es puesto a prueba con otros datos de testeo y posteriormente evaluado**
- Una vez confirmado el modelo, **las etiquetas son los datos clasificados**

### Tipos de Aprendizaje de Machine Learning

Existen 3 tipos de entrenamientos, simplificadamente:

- **Aprendizajes No Supervisados**: Se utilizan para algoritmos de **Clustering**, ya que los resultados obtenidos finales vendran dados por patrones no identificables a simple vista

## 2. Scikit - Learn: Libreria en python para Machine Learning

Es una libreria dedicada al desarrollo de modelos de machine learning, se caracteriza:

- Facilita de manera simple **acceso a algoritmos de los distintos tipos** (supervisados y no supervisados)
- Incluye todas las herramientas para procesar datos o gestionar modelos

Se instala a traves de ***pip install scikit-learn***

### Conceptos de la libreria a conocer

Hay varios elementos a conocer pertenecientes a esta libreria:

- ***Estimators*** (o estimadores): Es la representacion de un algoritmo con un modelo entrenado.
> Dispondremos del algoritmo de machine learning y el modelo que podremos ajustar con los datos de entrenamiento

- ***Samples Matrix*** (o matriz de muestra): Son los datos de muestra para entrenar el modelo que se representan como matriz. Filas so registros y columnas atributos (***features***).

### Metodos propios de la libreria

Existen 2 funciones propias de la libreria que ayudan a capacitar los modelos:

- ***.fit()***(funcion de ajuste): Es un metodo para ajustar y entrenar con datos el algoritmo + modelo de machine learning.
> Casi siempre **se reciben 2 argumentos en esta funcion** (son arrays)(Hablamos de algoritmos supervisados porque requieren de intervencion humana):
>
> > Matriz de muestra: conjunto de datos de muestra con todas las caracteristicas
> >
> > Valores objetivo: Seran los datos objetivo.
> >
> > > **Si es un modelo de clasificacion = seran numeros enteros o fijos**
> > >
> > > **Si es un modelo de regresion = seran numero reales para estimar la distribucion de los datos***

- ***.predict()***(funcion de prediccion): Metodo para predecir los datos utilizando el modelo y algoritmos (estimador) entrenado

## Ejemplo de uso de libreria Scikit-Learn

```python
# 1. Importamos de la libreria un algoritmo 
    # RandomForestClassifier es un algoritmo de clasificacion y de tipo arbol de decision

from sklearn.ensemble import RandomForestClassifier
```

```python
# 2. Generamos un estimador y los datos para entrenar el algoritmo.
    # Generamos 2 variables con numpy arrays.
    # Se cumple estas reglas:
        # 1º Siempre deberemos darles los datos de muestra y los datos de clasificacion
        # 2º Debera haber tantos datos de clasificacion como datos de muestra (hay 2 arrays de muestra = 2 valores objetivo)

clf = RandomForestClassifier(random_state=0)
x = [[6, 45, 79], [7, 68, 97]] # Estos seran la matriz de muestra
y = [0,1] # Estos seran los valores objetivos
```

```python
# 3. Ajustamos el modelo con .fit() y los datos

clf.fit(x,y)
```

```python
# 4. Generamos la prediccion con .predict()
    # la prediccion solo recibe un parametro

clf.predict(x)
```
