# Apuntes Cloud Computing y Big Data

En esta unidad conoceremos de manera simplificada que es el Cloud Computing y que aporta en el mundo del big data, que rol desempeña, que tipo de servicios son incluidos en el Cloud Computing y que coste tienen.

## 1.0 ¿Que es Cloud Computing o servicios en la nube?

La denominacion correcta es **el modelo de sistemas informaticos en el que los recursos y servicios de computacion (hardware, procesadores de memoria, almacenamiento, programas) son publicos y comparten acceso a traves de internet**

> Nos permiten ahorrar en costes de infraestructura y mantenimiento de IT para centralizar el servicio mediante un proveedor que configura externamente los servicios y programas que necesitamos.

### Ventajas del uso de servicios de Cloud computing

- **Ahorro en costes**
- **Actualizacion constante**
- **Sencillez de configuracion y mantenimiento**
- **Escalabilidad sencilla en potencia de calculo**
- **Accesibilidad y flexibilidad de servicios**

## Propiedades del Cloud computing

- **Servicio de configuracion "adhoc"**: Puedes configurar el nivel de servicio y recursos que necesites a traves de un configurador online, sin mediacion de ninguna persona.
- **Acceso obicuo por red**: Puedes acceder a estos servicios desde cualquier aparato con conexion a red.
- **Disponibilidad y aprovisionamiento de los recursos**: Podemos acceder a recursos remotos, gestionados por un tercero, con una capacidad de crecimiento horizontal y vertical.
- **Elasticidad**: Podemos decidir de forma dinamica y ajustar los recursos configurados de manera agil y automatica.
- **Traza y Contabilidad de uso**: Permiten llevar un control economico del uso de los servicios que estas utilizando.

## 2.0 Preguntas que surgen al usar Cloud computing en Big Data

### ¿Como se relaciona el cloud computing con nuestro trabajo en Big Data?

Las facilidades de computacion en la nube nos permiten **escalar nuestro trabajo desde un ordenador a un sistema de computacion complejo**(*cluster de servidores*), transformar nuestra capacidad de trabajar con datos, multiplicando sus recursos y herramientas.

### ¿Quien gestiona los recursos y servicios?

Una pregunta que surge al plantearnos configurar un espacio de trabajo de computacion en la nube es quien se encarga de hacerlo.

La respuesta es sencilla: ***Tú mismo***

> *En la mayoria de empresas existe un tecnico de sistemas que es el encargado*

El usuario que contrata el servicio suele ser ademas la persona que administra el servicio, **siendo nosotros mismos quienes se encarga de realizar todas las tareas**



Hay que tener en cuenta pues: ***La administracion y gestion de un cluster de cloud se complica exponencialmente para poder mantener la misma capacidad de trabajo estable**

### ¿Y si los datos crecen rapido?

Otra cuestion a tener en cuenta es inherente a este tipo de datos es **la exponencialidad de datos a tratar y la impredicibilidad de picos de trabajo.**

> Suele ser una cuestion comun que en segun que momentos del año, ocurran fenomenos que lancen las cargas de trabajo que regular en el dia a dia habitualmente, haciendo dificil predecir hasta que punto va a ser neecsario escalar los recursos.
>
> Esto implica tambien que estas situacion conllevaria un gran coste en inversion y tiempo
>

Sin embargo **debido a la elasticidad de los servicios de la nube** podemos escalar los recursos a la situacion real, con ciertas ventajas:

- Se realiza con **inmediatez** una vez solicitado
- **Crecen y decrecen los recursos segun las necesidades** de computacion
- Se puede **reajustar sin limite de veces**
- **Se cobra por los recursos utilizados**, no unicamente los solicitados.
- **Quitas costes** de mantenimiento y configuracion en las adaptaciones

## 3.0 Modelos de Servicios en Cloud Computing: IaSS, PaSS, SaSS

Todo proveedor de servicios en la nube nos ofrecera diferentes paquetes de funcionalidades que nos puede interesar contratar segun nuestras necesidades, recursos y actividad.

Estos servicios se agrupan en 3 familias:

- ***IaaS (Infraestructure as a Service):*** Infraestructura como servicio
- ***PaaS (Platform as a Service):*** Plataforma como servicio
- ***SaaS (Software as a Service):*** Software como servicio



### 3.1 IaSS: Infraestructura como servicio



> Es la capa mas basica de servicios en la nube

**Este tipo de servicios nos ofrecen acceso a capacidades de computacion basicos, tanto en recursos como infraestructuras**

> Estan compuestos por los CPUs, RAMs, HDDs,...

### ¿Cual es la ventaja principal?

Es un servicio de **alquiler de servidores virtuales basados en hardware que podemos decidir y configurar a nuestro gusto**, siguiendo un proceso:

- 1º Definimos y ajustamos las caracteristicas de capacidades (memoria, CPU,..) necesarias de nuestra maquina virtual
- 2º El proveedor habilita el acceso a esta maquina virtual a traves de la red, para configurarlo y adaptarlo a nuestro uso.

### ¿Como se tarifica?

Existen 2 opciones principalmente:

- **Precio Fijo**: Se paga **un coste fijo siempre** la maquina segun la configuracion definida (capacidad de CPU, RAM, HDD,..), **la uses o la uses**
> Aqui hay que tener en cuenta **cuanto tiempo necesitas que este activa y accesible la maquina para su uso**
- **Precio Mixto**: Pagas **un coste fijo menor por la preconfiguracion** de una maquina definida y **se añade un coste variable** en funcion del uso que haces de la maquina (horas de CPU, espacio utilizado en HDD, trafico de datos,...)
> Aqui tienes que tener en cuenta **cuanto uso continuo vamos a hacer de ella**

### 3.2 PaSS: Plataforma como servicio



> En este caso el usuario ya no dispone de control sobre los recursos de hardware o infraestructura

**Este tipo de servicio habilita distintos componentes y recursos de *middleware* para crear aplicaciones o procesamientos de datos**

Aqui se incluyen:

- BBDD, servidores de ficheros o web, lenguajes de programacion, librerias y plataformas de calculo
- Servicios de orquestacion y comunicacion entre componentes (el YARN en Hadoop por ejemplo)

### ¿Cual es la ventaja principal?

Es un servicio **de escalado automatico de recursos y auto gestion** que permite al usuario conectarse unicamente a una maquina virtual.

> En este caso el usuario puede definir unas reglas para que el sistema automaticamente se adapte y decida **si utilizar maquinas virtuales con mayores o menores capacidades y/o si utilizar mas o menos maquinas virtuales**
>
> Generalmente el usuario solo se le pide que defina **el tipo minimo de maquina (potencia y memoria), el nº minimo de maquinas y hasta cuanto queremos que crezca la potencia de trabajo*

Sigue este proceso de uso:

- 1º El usuario debera **realizar una seleccion de componentes *middleware*** seran necesarios para llevar a cabo su trabajo.
  
> Ej: Para una Sistema automatico de recomendacion =  HadoopBase (n nodos) + Hadoop (n nodos) + Motor WS

- 2º El usuario se **conectara al servicio desde cualquier terminal conectado a internet** y operara como si de un programa se tratara.

### ¿Como se tarifica?

Existen distintas opciones segun que contrates:

- **Costes por recursos**: Pagas por acceder a los recursos.
> Parecido a IaSS, posibilidad de pago fijo o mixto segun su uso. (Spark y las BBDD funcionan asi)
- **Costes por componentes**: Pagas por el uso de las herramientas de computacion privadas o no libres.
- **Costes segun disponibilidad**: Pagas por la facilidad de acceso a los recursos y herramientas, segun la frecuencia de uso. Existen 4 formatos:
> - Online: Acceso siempre habilitado
>
> - On demand: Acceso habilitado tras peticion expresa
>
> - Scheduled: Acceso habilitado unicamente en una franja horaria. (para realizar calculos periodicos, habitualmente).
>
> - Bid: Acceso habilitado segun el coste de acceso segun la franja horaria.

### 3.3 SaSS: Software como servicio



> Es la capa de servicios mas alta de los servicios cloud

**Este tipo de servicio nos permite conectarnos a las aplicaciones a traves de internet sin disponer el programa completo instalado**

> Generalmente se interactua con ellas a traves de Apps y toda la informacion de ejecucion esta en el servidor de la nube

Productos tipicos son: correo electronico, mensajeria instantanea, banca remota, ofimatica en nube,.... **en Big Data: seria Jupyter o R Studio**

### ¿Cual es la ventaja principal?

Es un servicio **que no requiere de ningun tipo de instalacion y puede correr a traves de internet unicamente**

Sigues este proceso de uso:

- 1º El usuario debera **conectarnos a nuestro proveedor, seleccionar las herramientas que necesitamos y contratarlas**

- 2º El proveedor **habilitara el acceso al usuario** quien podra hacer uso de los servicios.

### ¿Como se tarifica?

**Es un coste mediante licencias o suscripcion segun las variables de uso decididas:**

- Nº de usuarios con acceso
- Uso personal o profesional de los servicios
- Espacio de almacenamiento
- Version y funcionalidades del servicio

> Aquellos como correo electronico u otras aplicaciones, **no nos cargan nada por hacer uso de ellas**

## 4.0 Modelos de despliegue de servicios de cloud computing

En este punto ponemos el foco en **la gestion de la parte fisica de los servicios de cloud computing**

> Hasta ahora hemos hablado de contratar recursos, aplicaciones,... virtuales y con las que se interactua a traves de internet, pero  ***¿De quien son los elementos fisicos que estamos utilizando?¿Quien los gestiona?¿Quien dispone de acceso?¿Como se comparte la informacion?...***

Bajo esta perspectiva existen 3 opciones, 2 de ellas mayoritarias:

### 1º Cloud Privado

**La infraestructura de hardware y software para dar soporte a la nube, esta dedicada de forma exclusiva al usuario**

> Habitual en empresas, instituciones de formacion, la administracion publica....

##### Caracteristicas

- Todos los recursos de computacion y los sistemas/softwares que corren **no se comporten con otras entidades u organizaciones**
- **Los servidores fisicos pueden ser propios o contratados a un proveedor, pero gestionados con exclusividad**

### 2º Cloud Público

**El acceso a la infraestructura y servicios es abierto a cualquier usuario**

> Siempre que exista una cuenta con el proveedor y este suscrito (gratis o pago) en sus condiciones

##### Caracteristicas

- **No hay restriccion al uso de los recursos**
- **El proveedor dispone de servidores para el procesamiento**
- **El usuario puede configurarse sus maquinas virtuales o *middleware* que necesite**

### 3º Cloud Híbrido

**Es un modelo que parte de un Cloud privado, limitando el total acceso a ciertos usuarios y que posibilita el acceso publico limitado a otra parte soportada por la infraestructura**

> Muy utilizado en empresas, posibilitando acceso ilimitado desde las instalaciones de empresa, pero en remoto solo dispones de acceso a un subconjunto de herramientas 

##### Caracteristicas

- **Se puede compartir parte de la infraestructura y uso de software en 2 ambitos**
- **Protege y limita el acceso de elementos y datos con niveles de seguridad**

## 5.0 Ventajas y Desventajas del Cloud Computing

Finalmente la ultima reflexion final es entender **que implica para bien y para mal el uso de estos servicios con proveedores**

Ahora vamos a ver estos detalles.

### Ventajas del uso de servicios de cloud

- **Inversion inicial menor y ahorro en costes**
> Es clave tener bien dimensionado las necesidades de infraestuctura y plataforma que queremos utilizar **te puedes ahorrar costes muy importantes**
>
> Es una buena solucion para startups o emprendimientos que estan iniciando y requieran servicios en la nube

- **Inmediatez de acceso a los recursos**
> No necesitas comprar las maquinas, adecuarlas, configurarlas y activarlas, **solo contratar por internet que necesitas y esperar a que te habiliten acceso a lo que necesites**

- **Flexibilidad, elasticidad y escalabilidad**
> Puedes iniciar de una manera basica y hacer crecer el servicio **conforme al volumen de trabajo**, ya sea por necesidad de complejidad de las maquinas o aumentar el nº de maquinas.
>
> Ademas este proceso **se puede automatizar para que se auto-ajuste**

- **Mantenimiento**
> Es minimo, se encarga el proveedor de practicamente todo, **especialmente en las partes fisicas y configuraciones de red o actualizaciones de software**

- **Acceso ubicuo**
> Puedes acceder a estos servicios desde practicamente **cualquier dispositivo con conexion a internet**

### Desventajas del uso de servicios de cloud

- **Complejidad de Migracion de datos y servicios, asi como de compatibilidad entre el proveedor y nosotros**
> Implica tener que hacer una planificacion y preparar un trabajo previo antes de empezar a operar con el entorno previo en la infraestructura del proveedor
>
> Es necesario que **ambos sistemas convivan simultaneamente durante un tiempo**

-**Integracion con sistemas *legacy* y *on premise***
> Implica que los servicios de la nube, **se adapten a nuestros programas y entornos que no se pueden migrar**, añadiendo una capa de complejidad en el uso
>
>>*legacy* = Sistemas heredados / *on premise* =  Sistemas hechos adhoc para la empresa

-**Politicas de privacidad y seguridad de datos**
> Asegurar que los datos que se manejan, bajo ningun concepto, **pueden ser accedidos por terceros.**

-**Riesgo de costes variables descontrolados**
> Se pueden disparar los costes si no somos conscientes de la fluctuacion del trabajo
>
> Por tanto **pago por uso en circunstancias donde haya un flujo constante, puede encarecerse mucho mas que si pagas un acceso constante aunque no haya alguien utilizandolo todo el rato**
