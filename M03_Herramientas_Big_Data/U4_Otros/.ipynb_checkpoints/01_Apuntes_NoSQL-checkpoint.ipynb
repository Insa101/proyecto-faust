# Apuntes NoSQL: Tipos de BBDD, caracteristicas, propiedadews y familias

Hablando de las distintas herramientas y elementos del mundo del big data, tenemos que mencionar como las BBDD nos ayudan a albergar y tener disponibles los volumenes de datos que se manejan.

Vamos a recordar y desarrollar algunos conceptos basicos de las BBDD para entender como las bases de datos NoSQL nos puede ayudar, pero principalmente,**vamos a entender que son, que caracteristicas tienen y cuantas existen**

## 1.0 BBDD Relacionales SQL, sus retos con el big data y BBDD NoSQL como solucion.

Las bases de datos relacionales son aquellas **que organizan sus datos tabulada** es decir, cada interseccion de las filas y columnas que lo conforman representa una observacion caracterizada, un dato. Las filas registran los elementos y las columnas las propiedades.

Es necesario **definir la estructura de la tabla para crearla** entre lo que se define:

- Nº de columnas
- Nombre de las columnas
- Tipo de datos de la columna
- Claves unicas de la tabla

### Caracteristicas de las SQL

- Dispone de un lenguaje de consulta estandar (SQL)
- Soporta operaciones relaciones (JOINs)
- Usa esquemas normalizados, sin repetir informacion
- Es consistente y dispone de integridad referencial, para evitar incoherencias en la informacion y rotura de los vinculos entre tablas por error.
- La transmision de informacion entre tablas es atomica y aislada, evita que 2 usuarios modifiquen a la vez los mismos datos.
- Los datos estan centralizados, en un unico servidor, con una unica via de acceso y una limitacion de espacio.
> **Este es el principal escollo para un escalado multiple como los que necesita los datos Big Data**


### Principales problemas para el mundo del Big Data

Como veiamos antes, la dificultad del escalado hace el uso de este tipo de estructuras sea problematico, pero no es el unico problema que presentan a la hora de trabajar con big data. El principal problema es **que las bases de datos relacionales generan un cuello de botella** al trabajar con estos datos, debido a sus caracteristicas:

- Los big data **tienen un crecimiento exponencial** en datos, usuarios y peticiones o trafico a gestionar.
  > Por tanto las capacidades de la BBDD no serian suficientes
- Los datos **tienen estructuras no/semi-estructurados** que son dificiles de trasladas al modelo relacional, son poco eficientes, complejos de gestionar y evolucionar.
  > Lo que implica un nivel de computacion y procesamiento para generar estructuras manejables, que afecta a la latencia del acceso de los datos y a la evolucion de la BBDD que se convierte en insostenible prolongada en el tiempo.


En definitiva la consecuencia si utilizan BBDD SQL en Big data es que **las capacidades de la BBDD se llevan al limite, haciendo que la latencia se eleve y encareciendo la actualizacion de datos**

> Se resume tambien en que existe una imposibilidad de escalado horizontal de estas BBDD, por complejidad en el diseño respetando la estructura y por el elevado coste exponencial de los recursos que necesita para la gestion.

### Solucion para el mundo Big Data: BBDD NoSQL

Las bases de datos No relacionales **si permiten rendir con un sistema escalable con datos big data** por lo que se convierten en las BBDD adecuadas para trabajar con estos datos.

> Debido a que trabajan con un modelo de datos distinto al estandar de relacion tradicional (tablas y asociaciones) entre entidades

Otras caracteristicas que influyen positivamente para esta funcion son:

- Son flexibles en esquemas de datos (no existe *schema-less*), para actualizar las estructuras generadas para albergar los datos que entran.
- Son capaces de gestionar grandes volumenes de datos.
- Tienen una arquitectura distribuida para albergar los datos.
  > **Este es el factor diferencial para la gestion de big data**

#### BBDD distribuidas, la arquitectura necesaria

Como mencion especial, **las BBDD NoSQL nativamente nacen como BBDD distribuidas, con varios nodos (servidores) interconectados** que les permiten una serie de ventajas:

- Escalar horizontalmente de manera sencilla, aumentando capacidades conforme se necesitan.
- Particionar y replicar datos para una alta disponibilidad de los datos y tolerancia a fallos.
  > Esto es ya que al disponer de bloques de informacion replicados y distribuidos entre nodos, al consultarlos, no necesitamos que todos los nodos nos devuelvan la informacion, agilizando el tramite y optimizando los recursos.

## 2.0 Caracteristicas BBDD NoSQL

Ya hemos visto porque nos interesan utilizar BBDD NoSQL para el Big Data, pero estos sistemas de BBDD nos ofrecen otras caracteristicas importantes para el uso de Big data, que hay que conocer:

- ***Son BBDD distribuidas nativamente***: Como deciamos antes, estas BBDD en origen tienen arquitecturas distribuidas que facilitan la integracion de nuevos datos entrantes y su reparticion automatica entre nodos.
> las BBDD SQL **disponen de una estructura en un servidor de forma nativa** que necesita escalar en otros de manera horizontal.
- ***Flexibilidad esquematica***: Tambien visto, indica la facilidad de adaptacion de la estructura de los datos entrantes a su natulareza.
> las BBDD SQL **necesitan que se definan las estructuras de organizacion de los datos antes de hacer uso de ellas**
- ***Integridad referencial***: No requieren que entre los datos y su ordenacion exista vinculacion mediante referencias creadas, ya que al particionar y replicar datos, disponemos de distintas fuentes para los mismos datos.
> las BBDD SQL **recoge toda la informacion en un servidor que vincula los datos entre si para poder relacionarlos y acceder a ellos** necesitando de comprobaciones y establecimiento reglas que realentizan y complican la ordenacion de los datos confomr entran.
- ***Redundancia de datos***: Se particionan y reparten datos entre los nodos, para la tolerancia a fallos y mejorar la latencia.
> las BBDD SQL **solo disponen de una copia de datos en el servidor principal, lo que implica que su acceso es por una unica via**
- ***No existe lenguaje de consulta estandar***: Debido a la naturaleza variable de la tipologia de datos, no es posible establecer un unico lenguaje estandar para la consultas. Tambien debido a las variabilidad de la configuracion de la BBDD NoSQL (modo de particionado, modo de gestion de nodos y su coordinacion,...) implica que cada una de ellas es unica y tenga su porpia version de lenguaje de consultas
> **Para solucionar esto** las BBDD NoSQL, pueden incluir **una capa de software por encima de la BBDD** que les confiere la capacidad de traducir consultas en lenguaje SQL al lenguaje de consultas de la BBDD NoSQL que hayamos generado.


### ¿Como funciona la actualizacion y consistencia en BBDD NoSQL?

Otra caracteristica de estos sistemas que interrelaciona las anteriores caracteristicas, radica en el funcionamiento en general del sistema.

**Una actualizacion del sistema, se actualiza en todos los nodos del cluster, pero una peticion de operacion no se ejecuta a la vez en todos los nodos, se ira actualizando progresivamente en los distintos nodos para asegurar consistencia (pero no inmediatez)**

### Implicaciones del uso de BBDD NoSQL

Otras caracteristicas **de naturaleza limitante** de estas bases de datos como podemos ver es su ***inmediatez***, algunos factores caracteristicos que son afectados a esta falta de inmediatez son:

- No todos los nodos van a estar operativos o alcanzables por red
- No todos los nodos van a estar en un  estado coherente de informacion
- No todos los nodos pueden responder de forma rapida.


### 2.1 Propiedades CAP para trabajar en Big Data

Una vez conocidos los sistemas de BBDD NoSQL y sus caracteristicas, ahora ahondamos en las propiedades inherentes en su estructura **que son necesarias para trabajar con big data**. Son las *propiedades CAP* deseables para cualquier sistema distribuido.

- ***Particion Tolerance***: El sistema **funcione aunque falle la red entre los nodos.**
- ***Availability***: Al realizar una consulta todos los nodos que **esten funcionando, deben responder con una latencia y robustez en la informacion** adecuadas.
- ***Consistency***: Al realizar una consulta todos los nodos del sistema devuelven **la misma informacion con el valor actualizado mas reciente.**




### Teorema CAP y como afecta a las BBDD NoSQL

Este teorema indica que en los sistema distribuidos, estas 3 propiedades **no se pueden garantizar que cumplan perfectamente de forma simultaneamente todo el tiempo**

A continuacion se ve una imagen de como se relacionan estas 3 propiedades y **que riesgos pueden existir si no cohexistieran por la ausencia de 1 de las otras propiedades**



En las BBDD NoSQL **impera la propiedad de *tolerancia al particionado*** sobre el resto.

> Ya que se prioriza que pese a que existan fallos en la red de nodos, el sistema no se caiga.

Por tanto nos tendremos que hacer la pregunta ***¿Que otra propiedad quiero respetar en la configuracion?***
> Al configurar la base de datos, se tienen en cuenta como va a ser la naturaleza de su uso y por tanto discernir que es mas importante en la configuracion.
>
> Ademas **esta priorizacion NO IMPLICA QUE NO SE CUMPLAN LAS PROPIEDADES EN LA BBDD CONFIGURADA** pero en entornos de trabajo que exijan, o que no sean situaciones normales, el sistema intentara garantizar y cumplir las reglas por jerarquia.

- ***Si priorizas la consistencia:*** Esperara a que todos los nodos tengan la misma respuesta, **pero tardara en la respuesta**
- ***Si priorizas la disponibilidad***: Devolvera rapidamente la informacion disponible mas adecuada a la consulta, **pero no necesariamente la informacion mas actualizada**

## 3.0 Familias BBDD NoSQL

El siguiente aspecto a conocer de las BBDD NoSQL tiene que ver con la **dificultad de homogeneizarlas para su compresion y manejo de organizacion de los datos**.

> Las BBDD SQL **siempre tienen estructura de tablas** y facilita la comprension de la organizacion y manejo de datos.

Sin embargo las BBDD NoSQL requieren otra perspectiva para poder agruparlas. Para ello tendremos que tener en cuenta **el modelo de datos en el que se basan** consiguiendo simplificarlas en ***familias*** segun **el esquema general que sigue la estructura de informacion que almacenan**.

Existen 4 familias:

- ***Key-Value***(clave/valor)
- ***Document Oriented***(estructuras de informacion con atributos)
- ***Column Oriented***(formato wide-spread data)
- ***Graph***(grafos)

**A continuacion se incluye un listado de proveedores de servicios especializados para este tipo de familias:**



### 3.1 Familia *Key-Value*

> Es el modelo de datos mas simple que te puedes encontrar.

**En esta familia cada dato/valor se puede acceder a traves de su clave unica**

> Funciona igual que un diccionario en Python.

Tiene 2 elementos:

- Un identificador o etiqueta unicos
- Un valor asociado: La informacion que queremos almacenar



#### ¿Cuando se utiliza y que ventajas tiene esta familia?

**Se utilizan este tipo de BBDD para:**

- Operaciones sencillas de consulta por clave
- Edicion de pares clave/valor (añadir, borrar, modificar)**

**No se utilizan este tipo de BBDD para:**

- Operaciones de consulta sobre los valores

**Ejemplos de uso de estas BBDD son:**

- Cache de datos de acceso de webpages
- Datos de sesiones de usuario
- **Almacenamiento de archivos grandes objetos** (videos, audios,.. multimedia en general)

### 3.2 Familia *Document Oriented*

> Es una extension avanzada del modelo *key-value*

**En esta familia existe una clave vinculada a un elemento que almacena un *documento estructurado***

> La diferencia principal es que la clave es sustituida por *documentos*

Los ***documentos estructurados*** son un modo de representacion de informacion con una estructura definida, generalmente **con parte de la informacion anidada y con atributos**.

> **Estos documentos son almacenados generalmente en formato JSON o XLM**

Tienen estas caracteristicas:

- Los documentos pueden disponer de **estructuras jerarquicas complejas de informacion**
- Pueden no disponer de **esquemas predefinidos en la estructura**
- Se aplican a **multiples usos debido a su facilidad de introduccion de informacion y adaptacion de estructuras**



#### ¿Cuando se utiliza y que ventajas tiene esta familia?

**Se utilizan este tipo de BBDD para:**

- Estructuras de datos que requieran incluir mucha informacion nueva y compleja
- Realizar consultas sobre el contenido (valor) o campos que estamos almacenando.

**No se utilizan este tipo de BBDD para:**

- La latencia de las consultas requiere que sea agil (por ejemplo como en una SQL)

**Ejemplos de uso de estas BBDD son:**

- Informacion de gestion de E-Commerce (Catalogos de productos, fichas de cliente, registro de pedidos,...)
- Gestion de contenido web (comentarios de RRSS, metadatos de actividad como los likes, compartir en RRSS...)
- **Almacenamiento de informacion logs de aparatos IoT o de sistemas utilizados por el usuario** (el registro con la informacion completa)

### 3.3 Familia *Column Oriented*

> Es una extension avanzada del modelo *key-value*
>

**En esta familia los datos se registran en columnas, que representan un atributo asi como una clave y valor en y las filas representan los registros de entrada**

> *Ayuda a pensar en los datos asi como en los wide-spread, donde cada celda es valor defido por su registro y por su columna, pero que no tiene relacion el atributo con el proximo de otra columna*

**La informacion se organiza en columnas y familias de columnas, donde el nombre de la columna sirve de clave y valor/dato**

En esta forma los datos se almacenan en familias de columnas que tienen relacion entre si ya que representan el mismo tipo de informacion, **haciendo que cada columna represente una version distinta del registro**

Las columnas tienen caracteristicas tiene:

- Pueden existir **infinitas familias de columnas**
- **El identificador de la columna es un valor y una clave en si mismo**

**Las filas son registros con datos que disponen informacion de subconjuntos de columnas**

> Estas filas almacenan la informacion **como una matriz dispersa** ya que son estructuras adecuadas cuando hay muchos *nulls* o vacios.



#### ¿Cuando se utiliza y que ventajas tiene esta familia?

**Se utilizan este tipo de BBDD para:**

- Realizar consultas del valor de distintos modos, por columnas o registros o incluso por versiones de tiempo (*timestamp*)
- Familias de datos que estan muy bien jerarquizadas, organizadas y establecidas en sus atributos, pero que pueden variar en la existencia de informacion de todos los atributos
- **Registrar eventos o sucesos *(logs)* de una manera muy rapida**


**No se utilizan este tipo de BBDD para:**

- Informacion muy distinta o variable segun el tipo de dato y estructuras.

**Ejemplos de uso de estas BBDD son:**

- Registros de Mensajeria instantanea
- **Almacenamiento de informacion de señales de aparatos IoT en tiempo real** (analisis de eventos y logs)
- Analitica de redes sociales (gestionar registros de emisores, receptores de eventos generados)

### 3.4 Familia *Graph*


**En esta familia los datos son de diferentes tipos de elementos (nodos) y relaciones entre ellos (enlaces), asi como sus atributos**

> Hay que pensar en los registros de datos como **grafos** que no es otra cosa mas que **una representacion de un proceso por el que transcurre el dato**

Estas BBDD disponen de **operaciones muy especializadas para recorrer grafos, empezando en un nodo y terminando en otro, viendo el camino mas corto**

#### ¿Cuando se utiliza y que ventajas tiene esta familia?

**Se utilizan este tipo de BBDD para:**

- Si los enlaces de los datos se interconecta con pocos nodos y no crecen los enlaces en la misma proporcion.


**No se utilizan este tipo de BBDD para:**

- Si los enlaces generados por un nodo es exponencial y crecen tantos enlaces como nodos

**Ejemplos de uso de estas BBDD son:**

- **Analitica de datos organizados en forma de Red**
- Analitica de redes sociales (gestionar registros de emisores, receptores de eventos generados)

## 4.0 New SQL: Mas alla de SQL

Por ultimo, un update de la situacion actual en el mundo de las BBDD.

El origen de las bases de datos NoSQL fue **dar respuesta a las limitaciones de almacenamiento estructural de datos de distinto tipo** que no encajaban con el formato tabular.

Pero debido al nivel de avance de NoSQL, **esta permitiendo que las BBDD SQL se actualicen** que se caracterizan por:

- Tener unos caracteristicas ACID como las BBDD Relaciones
- Mantener el lenguaje SQL de consulta
- Nacer de manera distribuida y escalar de manera sencilla
