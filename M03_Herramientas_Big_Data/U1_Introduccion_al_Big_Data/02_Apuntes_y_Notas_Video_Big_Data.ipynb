# Apuntes y notas del video de introduccion al Big Data

Utilizo este documento para registrar notas o ideas sobre el mundo del Big Data

- La transformacion digital de la era post covid esta haciendo que als empresas incorporen tecnologias de la informacion para mejorar la eficiencia, el conocimiento interno, desarrollar la innovacion y mejorar la adaptacion a los cambios del mercado
- Es un proceso que afecta a todas las empresas simultaneamente
- Cambia el paradigma y evoluciona los puestos de empleo, debido a la incursion de la IA
- *Si una empresa no da el paso a la tranformacion digital, otra lo hara y le superara*
- Es un cambio que puede definir el nuevo modo de interaccion del mercado y te desactualice
- Este cambio **ha de estar liderado desde la direccion de la empresa** y añadir una cultura de innovacion, para cambiar procesos o productos nuevos.
- Es necesario contar con:
> **personas para liderar el cambio**, que dispongan de conocimientos para llevarlo a cabo y cultura de equipo multidisciplinares.
>
> **datos** que es de donde sacaremos la informacion
>
> **tecnologias** que nos permitan obtener los datos y poder transformarlos para realizar la transformacion digital

Estas tecnologias son las de Industria 4.0 y mas: *Mobile, Cloud, Big Data, AI/ML, IoT, Robotics, AM, Realidad aumentada*

- Una realidad empresarial es que tanto de manera interna como externa, las empresas generan una cantidad ingente de datos, que no son gestionados o controlados en su mayoria y menos aun se les saca redito
- El 1º paso es identificar adecuadamente que fuentes de datos internas tenemos, para poder obtener datos que posteriormente podamos explotar
> **Los datos son la gasolina**

- **El valor estrategico del Big data es la mejora en la capacidad de prediccion (*que va a ocurrir?*) y de prescripcion (*que acciones debo tomar?*)** que añaden un valor economico y de complejidad elevados.
