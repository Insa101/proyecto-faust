## 2. Big Data: Que es y que caracteristicas tiene

Es un poco ambigüo definir el *Big Data* pero se pueden entender como:

> ***Conjuntos de datos cuyo tamaño y complejidad excede la capacidad de los sistemas de software y hardware tradicionales para capturarlos, almacenarlos y procesarlos de forma eficiente y en un tiempo razonable***

Estas cargas de datos, presentan 3 desafios principales:

- **1º La cantidad de datos a gestionar**
- **2º La velocidad a la que se generan**
- **3º La amplia variedad de tipos de datos**

> Estos desafios es a lo que se le llaman ***las 3 V's en Big Data***, cuando se cumple 1 de ellas, ese conjunto de datos se denomina Big Data.

Sin embargo a medida que ha evolucionado el Big Data se ha vuelto mas complejo.... A continuación veremos cada una de las caracteristicas y su implicacion, conocidas como ***Las V's del Big Data***.

> ***Big Data se escribe con V***

### 2.1 Las V's del Big Data: *Volumen*

**La principal caracteristica del Big Data es el tamaño de los datos a manipular**

Aqui se ve una tabla de las capacidades de almacenaje digital, **el Big data consiste en tratar problemas de terabytes, pentabytes,...**



Por tanto el 1º reto es **almacenar los datos**

> Los sistemas tradicionales no permiten escalar el nivel de generacion de datos

Se encontro la manera de hacerlo a traves de *sistemas distribuidos* compuestos por varias maquinas o nodos interconectados por red, **donde los datos se *trocean* y replican para almacenarlos y recuperarlos despues**

> **Esta solucion es *escalable*, ya que si necesitas mas capacidad, conectas mas nodos**

### 2.2 Las V's del Big Data: *Velocidad*

**Existen datos generados de manera continua (*transacciones online*, por ejemplo) haciendo que la velocidad de generacion dificulte su procesamiento**

> Esta velocidad dificulta **los procesos de manipulacion y analisis de datos**

Por tanto, desarrollar soluciones que permitan la ingesta y procesamiento de grandes flujos de datos de manera inmediata es otro factor clave.

### 2.3 Las V's del Big Data: *Variedad*

La fuente de generacion de los datos tambien preesenta retos, ya que pese a disponer de *DataWarehouses* para intentar integrarlos y homogneizarlos, la realidad es que **se generan muchos tipos de datos de muchas fuentes**

> Un *WareHouse* de datos **Integrar datos de varias fuentes (de una empresa u organización) en un único repositorio y organizarlo de forma que toda la información sea fácil de acceder y consultar para un uso muy determinado.**

Pero la variedad no solo es de fuentes, **sino tambien de formatos y estructuras**

> Algunos vendran de fuentes *tradicionales* en formatos excel, csv,... pero otros pueden ser *logs* de maquinas o registros en otros formatos digitales para webs, multimedia...

La solucion es encontrar **sistemas flexibles y adaptativos a los nuevos datos, sin afectar al rendimiento, para los diferentes tipos de datos**

### 2.4 Las V's del Big Data: *Veracidad*

Los datos generalmente **no vienen *higienizados* (faltan datos, valores inexactos,...)**

> Cuando son cientos o miles de datos es afrontable, **si son millones, presentan un problema**

Las soluciones pasan por crear **formas sistematicas para marcar y filtrar datos problematicos para tenerlo en cuenta en analisis**

### 2.5 Las V's del Big Data: *Valor*

**Debemos obtener redito del esfuerzo e inversion en el procesamiento de los datos, mediante el uso de modelos y la evaluacion de sus resultados**

## 3. Tipos de datos: *Estructurados* y *No Estructurados*

Dentro de las caracteristicas del *Big Data* nos encontramos con que los datos con los que trabajaremos **seran distintos y deberemos trabajar por combinarlos, para obtener informacion**

Por tanto resulta esencial, en una primera instancia, **diferenciarlos para posteriormente sistematizar el modo de tratar el dato**

Existen 3 tipos de datos: ***Estructurados, No Estructurados y SemiEstructurados***



### 3.1 Datos Estructurados

**Son aquellos que tienen formato, tamaño y organizacion conocidas**

> Estos seran, *campos numericos, fechas, cadenas de texto simples*....
>
> > **Son tablas de BBDD u Hojas de calculo**

Son comunes en las organizaciones, los datos que componen los *data warehouse*, CRM's o ERP's, pero tambien en formularios webs, *logs* de actividad de servidores o sensores...

**El medio mas conocido para garantizar su persistencia y facilidad de acceso es la consulta de BBDD SQL**, haciendo uso de tablas

> **Cada tabla representa un tipo de dato y esta compuesta por varias columnas que representan atributos de la *entidad* representada**
>
> **Ademas se pueden establecer relaciones entre tablas**

Suelen ser una buena solucion para gestion de datos de muchas aplicaciones, **pero no son escalables en Big Data, ya que no se adaptan ni rinden bien con las cantidades ingentes de datos**

> Otro problema es que la estructura de estps datos ha de ser **tabular y hay datos que no encajan con este patron *(ej, datos jerarquicos tipo arbol)*** obligando a crear representaciones adaptadas que consumen muchos recursos, reduciendo la eficiencia y aumentando la complejidad

### 3.2 Datos No Estructurados

**Son aquellos que no siguen un modelo o esquema definido, ni en tamaño o sin organizacion logica**

> Estos seran, *contenido de texto, como comentarios de un foro, un cuerpo de correo, un contenido web,...* **tambien contenido multimedia,**
>
> > **Este tipo de datos representan el 80% de los datos del mundo actualmente**

**Las soluciones en Big data aplican tareas de reconocimiento de lenguaje (LLM's), traduccion automatica, analisis de sentimiento, procesado y reconocimiento de imagenes,...**

### 3.3 Datos SemiEstructurados

**Son aquellos que no tienen una estructura fija inmutable, como los datos tabulares, pero que siguen cierto esquema o estructura flexible, con elementos diferenciados**

> Permiten representar informacion *con un patron jerarquico*
>
> > **Son datos muy extendidos en el mundo de aplicaciones web y comunicacion en servicios de red. XML y JSON son los principales**

**Las soluciones en Big data simplifican el manejo y acceso, permitiendo trabajar de manera directa con ellos.**

## 4. Sistemas de Big Data: Que son y propiedades

Este concepto habla de la forma en la que se configuran las maquinas para poder llevar a cabo las labores de procesamiento de datos de Big Data.

Estos sistemas **son sistemas distribuidos con distintos nodos interconectados entre si, que son configurados en conjunto de forma nativa y gestionados autonomamente o con poca intervencion del usuario.**

Esto implica que:

- **El usuario no se preocupa en el modo en el que se dividen los datos**
- **El usuario no se preocupa en el modo en el que se distribuyen los datos**
- **El usuario no se preocupa en el modo en el que se replican los datos**

> En los sistemas tradicionales de bases de datos en servidores o *clusters* el problema de fondo es la configuracion de los mismos, la distribucion,... que llevado a datos big data **hacen imposible el procesamiento y la manipulacion de los mismos**

### 4.1 Propiedades de Sistemas Big Data: *Escalabilidad*

**Todo sistema Big Data debe permitir añadir mas recursos a una maquina (*ampliacion*) para poder adaptarse a gestionar mas informacion, manteniendo el rendimiento y su cpacidad de respuesta**

> En los sistemas tradicionales se ***escala verticalmente*** lo que implica que **se añade mas elementos para mejorar su capacidad o cambias la maquina a una mas potente**
> 
> En los sistemas distribuidos, se ***escala horizontalmente***lo que implica **añadir nuevos nodos** para mejorar la capacidad de almacenamiento y calculo. Esto suele requerir ademas **mayores costes y planificacion para ejecutarlo**



Los sistemas en este caso, se conciben como **sistemas distribuidos con ampliaciones *economicas y a gran escala***. De este modo se pueden añadir mas nodos de manera sencilla e incluso automatica.

> La idea era conectar muchas maquinas "mas simples" que permitan hacer la funcion de pocas maquinas complejas

El modo de hacer esta *ampliacion* es unicamente **instalando un software en el servidor principal, configurarlo y conectarlo a la red**. Este software se encargara automaticamente de:

- Reconocer la capacidad del servidor de manejar los datos
- Repartir la carga de datos y procesos mas ajustada a los nodos.
- Replicar y redistribuir los datos entre los nodos.

### 4.2 Propiedades de Sistemas Big Data: *Latencia*

**Los sistemas Big Data procuran mantener *estable* la latencia de calculo en las mismas maquinas donde estan los datos**

> La latencia es **la suma de los retardos o tiempos consumiods por las distintas partes del sistema para completar una tarea**
> 
> La idea es intentar eficientar el proceso de calculo, ya que **procesar en una ubicacion distinta a donde estan los datos, consumo muchos recursos en la transferencia de informacion**

El modo en el que estos sistemas lo consiguen es aplicando tecnologias de ***stream & event processing*** que se encargan de **gestionar flujos de datos que llegan de forma continuada y a gran velocidad, procesandolos y respondiendo en tiempo real** 

### 4.3 Propiedades de Sistemas Big Data: *Tolerancia a fallos*

**Los sistemas Big Data deben permitir disponer de mecanismos para minimizar el tiempo inoperativo por una falla en la estructura de nodos y descargar todo lo posible a los responsables del sistema para mantenerlo en funcionamiento**

> Una falla puede ser **un evento que provoque que un nodo de procesamiento o almacenamiento dejen de funcionar y afecte al sistema**
> 
> En un sistema tradicional **una falla puede provocar que el sistema pierda informacion**

El modo en el que estos sistemas lo consiguen es **incorporando mecanimos para proteger y gestionar estas fallas automaticamente, permitiendo asegurar el funcionamiento, reponerse y ser transparentes para el usuario**

## 5. El ciclo de vida del Big Data

Como en otros ambitos laborales, existen ciclos de vida **que disponen de un evento que inicia un proceso general, se generan procesos o tareas y termina en otro evento distinto al inicial**

El Big data no esta exento de ello, aunque la complejidad de este ciclo puede variar los principales pasos que vamos a ver son los aqui representados.



### 5.1 Primer Paso: *Adquisicion*

**Es el proceso de captura de datos, incluido el preproceso de filtrado y limpieza**

> Suele ser **uno de los procesos mas criticos**

El sistema Big Data debe permitir **la ingesta de grandes volumenes de datos, manteniendo un baja latencia y ser flexible para aceptar estructuras diversas sin afectar al desempeño**

### 5.2 Segundo Paso: *Preparacion y Procesado*

**Es el conjunto de tareas para la limpieza, validacion, normalizacion y filtrado, asi como procesos de precalculo de agregaciones para su uso posterior**

> Tambien se puede entener como **cualquier transformación básica de los datos que nos permita tener los datos listos para su análisis.**
>
> > Hay que tener en cuenta que los datos generalmente, pueden tener *valores incorrectos, ausentes o duplicados asi como disponer de distintas codificaciones*
> >

La complejidad es **realizar estas operaciones con millones de datos**. Ademas resulta esencial ya que en la parte de analisis, **no vas a querer procesar los datos ya que consume mucho tiempo**

> Por ello es necesario **planificar y adelantar preprocesos, para disponer de ellos almacenados, que es mas eficiente y economico que procesar**

El sistema Big Data **estas tareas suelen realizare mediante procesos por lotes que se ejecutan de forma regular sobre un gran conjunto de datos**.

> Estos datos suelen estar organizados en grandes arcivos y almacenados en un sistema de ficheros distribuido, como ***Hadoop***

### 5.3 Tercer Paso: *Analisis y modelizacion*

**Es el conjunto de tareas para explorar, descubir patrones y relaciones entre los datos, sintetizar y crear modelos de clasificacion, prediccion, recomendacion,...**

> Tambien se puede entener como **cualquier transformación básica de los datos que nos permita tener los datos listos para su análisis.**
>

La complejidad es **que estas tareas requieren de calculos complejos, incluido realizar pruebas, simulaciones y validaciones cruzadas hasta llegar al modelo final**

> Son procesos basados en **la ejecucion cientifica de datos, *ensayo y error***
>

El sistema Big Data **necesita de un sistema de computacion distribuida que permita realizar y repetir calculos complejos, iterando datos, con eficiencia y latencia acotados**.

> Debemos intentar crear sistemas que **agilicen los calculos de manera que podamos realizar complejidades tecnicas en los modelos como si trabajasemos con pocos datos**, como la herramienta ***Spark***


### 5.4 Cuarto Paso: *Almacenamiento*

**Es el proceso de garantizar la persistencia, publicacion y accesibilidad de los datos procesados, ademas de permitir ser replicados y auditados**

La complejidad es **que el formato y las estructuras de los datos es diversa**

> No solo *los datos de entrada* sino tambien *los datos procesados*
>

El sistema Big Data **necesita cumplir estas caracteristicas para poder almacenar adecuadamente**.

- **Debe ser flexible para integrar y trabajar distintos tipos de datos**
- **Debe ser capaz de gestionar grandes volumenes de datos eficientemente**
- **Debe ser escalable**
- **Debe permitir el acceso con latencia ajustada**
- **Debe facilitar un mecanismo de consulta y filtrado sencillo**
- **Debe garantizar la persistencias, disponibilidad y recuperacion de los datos**

Las ***BBDD No SQL* se han diseñado teniendo estas caracteristicas en cuenta** ya que ofrecen una gran flexibilidad y otras soluciones de alto rendimiento para modelado de datos.

### 5.5 Quinto Paso: *Publicacion*

**Es la capacidad que permite la visualizacion y aprovechamiento de los datos tras el resto de procesamientos**

> *Si hemos creado un modelo, deberia ser accesible por oitras aplicaciones que lo necesiten*
>
> > Una solucion tipica es **publicar los resultados en servicios web**

La complejidad es **la creacion de elementos de visualizacion de datos de forma interactiva**

> *Graficos,Cuadros de mando, informes,..*

El sistema Big Data **permiten realizar esta funcion y amplia aun mas con posibilidades de comparticion de los elementos de visualizacion**

## 6. Procesos de Big Data: *Por lotes o En Tiempo Real*

Otro concepto importante de conocer del Big Data es la diferencia que existe en los modos de procesamiento.

Hemos visto que los sistemas de big data deben cumplir unas caracteristicas (*escalables, tolerantes a fallos y con latencia ajustada*) y como son los distintos pasos en el procesamiento de datos, **pero tambien hay que tener en cuenta el momento en el que se procesan esos datos segun *el tiempo de respuesta***

> El big data ha sido revolucionario especialmente por **la capacidad de procesar datos en tiempo real**

Segun este *tiempo de respuesta* distinguimos 2 modos de procesamiento: ***Por lotes o En Tiempo Real***

### 6.1 Procesamiento *Por Lotes (Batch processing)*

**Son procesos planificados periodicos en los que se acumulan y ejecutan calculos sobre el volumen total de datos. Cuando concluyen los procesos los resultados son registrados para explotarlos y publicarlos**

> Se utiliza este metodo **cuando tenemos grandes volumenes de datos y no necesitamos los resultados de manera inmediata, ya que se busca ver una evolucion**
>
> > Ejemplos: *calcular las paginas/productos mas vistos en una web, terminos de busqueda mas usados,...*

### 6.2 Procesamiento *En Tiempo Real (Stream processing)*

**Son procesos con un tiempo de respuesta minimo donde los datos de entrada son procesados actualizandose para obtener resultados de forma continua**

> Se utiliza este metodo **cuando necesitamos que los nuevos datos se procesen inmediatamente conforme llegan al sistema**
>
> > Ejemplos: *Programas de deteccion de fraude, recomendadores de actividades,...*

## 7. Tecnologias de Big Data: *Hadoop, Spark, NoSQL*

En ultima instancia solo quedarian ver las herramientas de Big Data **que permiten procesar datos en grandes cantidades, cumpliendo con los requerimientos y procesos explicados anteriormente**

Aunque los vamos a ver detenidamente en el resto de modulos, vamos a mencionar y explicar brevemente cada tecnologia o herramienta.

### 7.1 *Hadoop*: Tecnologia para el almacenaje en Big Data

***Apache Hadoop* es una plataforma software para trabajar con grandes archivos de datos en un sistema distribuido**

> Especialmente diseñada para trabaja con *clusters*, facilitando la escalabilidad y la tolerancia a fallos.

Tiene 2 elementos principales: ***HDFS (Sistema de ficheros distribuido)*** y su motor ***MapReduce***

Resumidamente el proceso que realiza esta tecnologia es:
- **1º Planificar los trabajos necesarios y lanzarlo a los nodos disponibles**
- **2º Almacenarlos en HDFS repartidos entre los nodos**
- **3º Ejecucion de calculos en cada nodo con *map***
- **4º Agrupar los resultados de los nodos mediante el uso de claves y se usa *reduce***
- **5º Devolver los valores finales y combinarlo en un unico conjunto de resultados**

> Esta tecnologia es muy adecuada para **procesos *por lotes***

#### 7.1.1 *HDFS (Sistema de ficheros distribuido)*

Es un sistema que distribuye archivos de gran tamaño en multiples maquinas, **los ficheros se *trocean y reparten* en bloques entre los nodos. Ademas genera copias de cada bloque en nodos distintos, para garantizar la integridad de los datos**

#### 7.1.2 *MapReduce*

**Es un motor para ejecutar trabajos distributivamente entre el *cluster* de nodos de calculo**. 

Este motor sigue la regla de procesamiento *Map-Reduce* que consiste en dividir la tarea en 2 pasos:

- el 1º paso *map* los datos se trocean y se realiza alun calculo (transformacion) individual sobre cada dato.
- el 2º paso *reduce* se realiza una operacion de agregacion o combinacion de los resultados haciendo uso de una clave o variable.

### 7.2 *Apache Spark*: Tecnologia para el procesamiento del Big Data

***Apache Spark* es una plataforma de computacion distribuida optimizada para ejecutar algoritmos complejos sobre gran volumen de datos**

> Utiliza un *cluster* de maquinas en paralelo para calcular y garantizar tolerancia a fallos

Esta tecnologia cuenta con un elemento propio diferenciador: ***RDDS (Resilient Distributed Dataset)***

> **Son bloques de datos inmutables, distribuidos y replicados entre los nodos, mantenidos en memoria.**
>
> Cuando se completa una operacion, **se genera otro *RDDS*, lo que permite ver el historial y traza de los pasos**

Algunas de las ventajas de esta tecnologia son:

- Trabajar con una consola o interprete interactiva
- Reducir el tiempo necesario para ejecutar un calculo individual
- Generar flujos de datos complejos de manera eficiente
- Facilita el analisis exploratorio interactivo de los datos
- Añade la implementacion de algoritmos iterativos sobre un mismo conjunto de datos
> Es la base para **el entrenamiento y validacion de modelos**

Por ultimo, **tambien ofrece posibilidad de trabajar con DataFrames** y librerias especializadas en analisis estadistico y machine learning.

### 7.3 *No SQL*: Tecnologia para el acceso y publicacion del Big Data

***No SQL* Abarca diferentes sistemas de bases de datos sin estructura tradicional tabular**

En el mundo del big data se han adoptado estos sistemas de estructuras de bases de datos por estos motivos:

- **Permite manejar datos de formatos y estructuras diversos**
- **Tienen capacidad para gestionar grandes volumenes de datos y escalar de manera sencilla**
- **Aseguran el acceso y disponibilidad de informacion, con una latencia adecuada**

Esto es debido a que **estas bases de datos estan concebidas en el origen como sistemas distribuidos y facilitar la escalabilidad, replicacion y latencia de respuesta, sin casi interaccion del ususario**
