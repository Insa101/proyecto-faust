# Apuntes Hadoop: Como programar y utilizar el sistema de ficheros de Hadoop o HDFS

Una vez vista la herramienta, ahora entenderemos y nos familiarizaremos con el sistema de HDFS, su sintaxis y logica.

Su lenguaje de programacion esta basado en sistemas operativos como *linux*

> la mayoria de ordenadores virtuales se realizan en base a un sistema operativo open source, por ser mas economico, facil de configurar y mas libre de limitaciones en el escalado de los nodos.

Es importante tener en cuenta que **utilizaremos Hadoop siempre en un maquina virtual** por lo que sera esencial configurar una e iniciarla antes de ponernos a trabajar.

## Iniciando HDFS y YARN

Para iniciar los servicios de HDFS ejcutaremos en el ordenador virtual lo siguiente:

> ***start-dfs.sh***

Con esto lo que haremos es conectarnos a los nodos de los que se compone el cluster:

- Iniciara el *NameNode* en la maquina donde lo hemos lanzado
- Iniciara los *DataNodes*
- Iniciara un *NameNode secundario* en otra maquina.

Una vez tenemos conectados los servicios HDFS, podemos iniciar el sistema que lo gestionara, el YARN, se iniciara asi:

> ***start-yarn.sh***

Con esto activaremos los elementos del YARN en el cluster creado, **pero al no haber enviado ningun trabajo aun, solo se activaran 2**:

- Iniciara el *ResourceManager*
- Iniciara los *NodeManagers*

##### Iniciar todo a la vez

Se pueden iniciar todos los sistemas a la vez mediante una unica sentencia:

> ***start-all.sh***

## Deteniendo HDFS y YARN

El modo correcto de dejar de trabajar con Hadoop y la maquina virtual es **parando los sistemas que hemos inicializado**

> Si no se hace esto, **puede ser que la proxima vez al iniciar, el sistema HDFS tenga un estado incorrecto y nos encontremos un problema**

Para detener los sistemas, **1º paramos los procesos de YARN**:

>***stop-yarn.sh***

En este caso cerrara los elementos activos en ese momento

Despues habra que cerrar el sistema HDFS:

>***stop-dfs.sh***

Despues parara los nodos del cluster que habiamos creado.

##### Detener todo a la vez

Se pueden detener todos los sistemas a la vez mediante una unica sentencia:

> ***stop-all.sh***

## Proceso de trabajo en HDFS

Una vez que tenemos inicializado el sistema en la carpeta virtual, tenemos que tener en cuenta que **deben existir una serie de pasos antes de utilizarlo para computacionar datos**

Esos pasos, que se realizan mendiante comandos de HDFS, se pueden resumir en este orden:

- **1º Hay que crear un directorio con carpetas**
- **2º Hay que importar/copiar archivos a HDFS para trabajar con ellos**

### Solicitudes en Java para Hadoop

Como ya comentamos, Hadoop esta programado en Java en origen, de modo que aunque el modo de programacion en general tendra su propia sintaxis y logica, **habra ocasiones que podremos ejecutar codigo/elementos de Java para revisar el sistema de Hadoop**

Aqui indicaremos estos comandos o solicitudes:

#### Comandos JDK en Hadoop

Los ***Comandos JDK (Java Development Kit)*** **son pequeños programas ejecutables que vienen dentro de una sentencia simple**

> Estos comandos tienen diversas funciones que ayudan al desarrollador de Java dandoles informacion.

Existen 4 esenciales:

- ***javac***: Es una sentencia que permite al programa *MapReduce* ser interpretado por Hadoop para ejecutarse
> Resumen: *Usas javac si tienes que compilar código crudo.*
- ***jar***: Es una sentencia que permite comprimir los archivos con los que trabajamos en Hadoop
> Resumen: *Usas jar para empaquetarlo y enviarlo al clúster.*
- ***java***: Es una sentencia que inicia el programa. En el caso de Hadoop, es sustituida por Hadoop
> Resumen: *No lo usas, porque usas Hadoop.*
- ***jps**: Es una sentencia que despliega una herramienta de diagnostico y te permite ver como esta el sistema
> Resumen: *Usas jps cada vez que arranques Hadoop para verificar que las piezas (NameNode, DataNode) están encendidas.*

## Comandos Tipo en HDFS

Para interacuar con el sistema HDFS, necesitaremos utilizar *la linea COMANDOS*

> Es la herramienta basica para ejcutar operaciones en el script HDFS

La linea comando utilizados es la siguiente asi:

> ***hdfs dfs -....***

A continuacion se vera una imagen que nos permitira operar los distintos HDFS que **explicaremos mas adelante**:



### Listar un directorio HDFS

**Muestra el contenido de un directorio en HDFS**

> ***hdfs dfs -ls /.../...***

En esta sentencia:

- *hdfs dfs* es la linea comando de hdfs para interactuar con el sistema
- *- ls* es la operacion de acceso al directorio de ficheros
- */.../* selecciona el **directorio raiz**
- */...* seleccionas **otros directorios dentro del directorio raiz**

Al ejecutarlo, **si no hay ficheros no aparecera nada**, pero si si hay ficheros, aparecera una sentencia similar a esta:

> *drwxr-xr-x   -   hadoop  supergroup          0   2026-01-27 10:00   /mi_datos
  |________|  |_|  |____|  |________|  |_______|   |______________|   |_______|
      1        2      3        4           5              6               7*

Esta sentencia tan criptica, tiene 7 partes y se entienden asi:

##### 1º Parte, Permisos

*drwxr-xr-x*, **es la cadena de permisos sobre el dichero** esta compuesta por 4 partes:

- *d*: Posicion 1, **Indica que tipo de archivo es** (en este caso un directorio)
- *rwx*: Posiciones 2-4, **Indican los permisos del usuario dueño**, en este caso, *r* permite leer, *w* permite escribir, *x* permite ejecutar.
- *r-x*: Posiciones 5-7, **Indican los permisos del grupo**, en este caso, *r* permite leer, *-* impide escribir, *x* permite ejecutar.
- *r-x*: Posiciones 8-10, **Indican los permisos de otros usuarios**, en este caso, *r* permite leer, *-* impide escribir, *x* permite ejecutar.

> Por tanto, en Hadoop, las letras r,w,x, tienen su significado:
>
> > ***r permite ver el contenido de un archivo***
> >
> > ***w permite modificar o borrar el contenido de un archivo***
> >
> > ***x permite acceder a la carpeta donde esta el contenido, en este caso***


##### 2º Parte, Factor de replicacion:

**Es una columna exclusiva de Hadoop e indica el nº de copias del fichero en los nodos del cluster (entre 1 a 3)**

##### 3º Parte, Usuario Propietario:

**Indica el *dueño* del fichero, es decir, quien tiene el control total en los permisos**

##### 4º Parte, Grupo propietario:

**Indica el grupo de usuarios asociados al fichero. Si perteneces al grupo, se te aplcian los permisos de grupo**

##### 5º Parte, Tamaño:

**Indica el peso en bytes**. Generalmente es 0 porque en HDFS no calcula el tamaño total del contenido automaticamente.

##### 6º y 7º Parte, Fecha y Nombre:

**Indica cuándo se modificó por última vez y la ruta del archivo.**

### Crear directorios en HDFS

**Crea directorios en HDFS**

> ***hdfs dfs -mkdir /user/...***

En esta sentencia:

- *hdfs dfs* es la linea comando de hdfs para interactuar con el sistema
- *- mkdir* es la operacion de creacion del directorio
- */user* crea una carpeta raiz a nombre del usuario
- */...* crea otras carpetas con otros nombres dentro de la carpeta raiz


### Copiar datos a HDFS

**Crearemos una copia de datos dentro del sistema HDFS que hemos generado**

> ***hdfs dfs -put /"direccion de los archivos local"/"nombre del archivo"."tipo de extension" /"directorio donde quieres copiarlo"***

En esta sentencia:

- *hdfs dfs* es la linea comando de hdfs para interactuar con el sistema
- *- put* es la operacion de copia de datos
- */"direccion de los archivos local"/* es la ubicacion de los datos en el sistema local
- */"nombre del archivo"."tipo de extension* el nombre de los archivos con los datos.
> Aqui puedes hacer uso del "*" para seleccionar todos e indicar "el tipo de extension" (.txt, .csv,...) para que seleccione los de ese tipo.
- */"directorio donde quieres copiarlo"* es la ubicacion del directorio donde quieres copiarlo en el sistema HDFS

### Ver el contenido de un fichero en HDFS

**Muestra el contenido de un fichero en HDFS, lee el fichero y muestra el contenido en pantalla**

> ***hdfs dfs -cat /"direccion donde esta el fichero"***

En esta sentencia:

- *hdfs dfs* es la linea comando de hdfs para interactuar con el sistema
- *- put* es la operacion de copia de datos
- */"direccion donde esta el fichero"* es la ubicacion exacta del archivo a leer

En esta sentencia, dado que los archivos son muy extensos, **es necesario utilizar *shells* de linux**:

##### Los *Shell*

**Son extensores del codigo que conectan un *output* con otro comando en la entrada siguiente *input***

Se compone de 3 elementos:

- **1º el comando donde sale el output, en este caso, el que accede al fichero**
> *hdfs dfs -cat /"direccion donde esta el fichero"*

- ***2º el conector para el output y el input siguiente:*** |
> *hdfs dfs -cat /"direccion donde esta el fichero" |*

- ***3º el comando que indica que realizar sobre el output, en este caso, mostrar las 10 primeras lineas*** |
> *hdfs dfs -cat /"direccion donde esta el fichero" | head*

Algunos comandos a utilizar en input final son:

- ***head***: Muestras las 10 primeras lineas de un archivo
- ***tail***: Muestras las 10 ultimas lineas de un archivo
- ***wc - l***: Cuenta las lineas, te indica cuantos registros tiene el archivo.
- ***grep "palabra buscada"***: Busca y muestra las lineas que contengan la palabra buscada.

### Copiar ficheros dentro del HDFS

**Podemos crear copias de ficheros existentes en el sistema**

> ***hdfs dfs -cp /"direccion donde esta el fichero" /"direccion donde queremos copiar el archivo"/"nombre del archivo"."tipo de extension***

En esta sentencia:

- *hdfs dfs* es la linea comando de hdfs para interactuar con el sistema
- *- cp* es la operacion de copia de datos
- */"direccion donde esta el fichero"* es la ubicacion exacta del archivo a copiar
- */"direccion donde quieres pegar el archivo/"* es la ubicacion exacta donde copiar el archivo
- */"nombre del archivo"."tipo de extension"* es el nombre que recibira el archivo y su tipo de extension.
> Las extensiones pueden ser *.txt, .csv,...*

### Salir del *Safe mode* en HDFS

Puede suceder que al interactuar con el HDFS, te salte *un cortafuegos* que te impida realizar una actividad, este puede ser el *safe mode* de HDFS.

> Generalmente esta activo al iniciar una maquina virtual y se desactiva automaticamente al verificar la conexion con todos los datanodes, **pero puede tardar un poco**

Se puede quitar a traves de la siguiente sentencia:

> ***hdfs dfsadmin -safemode leave***

### Borrar archivos y ficheros en HDFS

**Puedes borrar uno o varios archivos en HDFS**

> ***hdfs dfs -rm /"direccion donde esta el fichero"/"nombre del archivo"."tipo de extension***

En esta sentencia:

- *hdfs dfs* es la linea comando de hdfs para interactuar con el sistema
- *- rm* es la operacion de borrado de datos
- */"direccion donde esta el fichero"* es la ubicacion exacta del archivo a borrar
- */"nombre del archivo"."tipo de extension"* es el nombre del archivo y su tipo de extension.

##### Borrar directorios en HDFS

Tambien se pueden borrar directorios, casi con la misma sintaxis:

> ***hdfs dfs -rm -r /"direccion donde esta el fichero"/"nombre del archivo"."tipo de extension***

La principal diferencia es que añade *- r* es el indicativo de recursivo, para seleccionar carpetas y subcarpetas.

##### Borrado definitivo de archivos o directorios

En ambos casos y por defecto en Hadoop, **nuestras selecciones de borrado son "traladadas" a una carpeta temporal, donde podemos recuperarlas, durante un tiempo limitado configurado**

Sin embargo, existe la opcion de eliminarlo definitivamente de 1 vez:

> ***hdfs dfs -rm -skipTrash /"direccion donde esta el fichero"/"nombre del archivo"."tipo de extension***

Se realiza añadiendo  *-skipTrash* que indica que omita la opcion de enviarlo a la carpeta temporal

> Tambien se puede utilizar para directorios
>
> Es recomendable no utilizarlo salvo que domines Hadoop.

### Obtener ficheros en HDFS

**Podemos obtener un fichero  (copia) concreto de nuestro ordenador**

> ***hdfs dfs -get /"direccion donde esta el fichero"/"nombre del archivo"."tipo de extension "nombre de la copia"."tipo de extension***

En esta sentencia:

- *hdfs dfs* es la linea comando de hdfs para interactuar con el sistema
- *- get* es la operacion de obtencion
- */"direccion donde esta el fichero"* es la ubicacion exacta del archivo a borrar
- */"nombre del archivo"."tipo de extension"* es el nombre del archivo original y su tipo de extension.
- *"nombre de la copia"."tipo de extension"* es el nombre de la copia y su tipo de extension.
