# Apuntes Hadoop: Programar en Hadoop utilizando la libreria MRJob en Pyhton

> Esta es una de las 2 formas de programar directamente en Hadoop para hacer uso del motor MapReduce.
>
> En ambos casos programaremos en **Python**

**Es una libreria para poder utilizar Python en tareas *MapReduce* de Hadoop** en el que se propociona estructuras de codigo para crear funciones para los mappers y reducers de una manera sencilla en python.

> La idea era tener una API de clases y funciones sencilla y potente que simplificara el desarrollo en Python

A continuacion se ve una explicacion de que y como se configuran las aplicaciones de esta libreria para usar Hadoop:



## Iniciando la libreria MRJob

Cada vez que vayamos a usar esta libreria, primero deberemos traerla a nuestro entorno de desarrollo:

>***from mrjob.job import MRJob***

Una vez realizado eso, sera necesario **crearemos una clase a partir de la clase MRJob para aprovechar sus atributos y metodos** :

> ***class "nombre clase"(MRJob):***
> 
>> Haremos esto ya que para crear las funciones que haran mappers y reducers, **naceran a partir de modulos creados de esta libreria**.

Despues de ello definiremos nuestras funciones mapper y reducer.

- la funcion mapper recibira como parametros **una clave y un valor, para realizar un calculo que generara un par de resultados clave valor intermedios**
- la funcion reducer recibira como parametros **una clave y un conjunto de valores, para realizar un calculo que generara un par de resultados finales con claves y valores**
