```python
class Jugador:
    # Constructor para atributos fijos
    def __init__(self, nombre, edad, nacionalidad, altura, posicion, equipo):
        # 1. Asignación de atributos fijos.
        self.nombre = nombre
        self.edad = edad
        self.nacionalidad = nacionalidad
        self.altura = altura
        self.posicion = posicion
        self.equipo = equipo
        # 2. Asignacion de atributos de Rendimiento. Pondremos valor 0.
        self.puntos = 0
        self.asistencias = 0
        self.minutos = 0
        self.partidos_jugados = 0  
        self.faltas = 0
        self.partidos_lesionado = 0 
        self.robos = 0
        self.tapones = 0
        self.triples_anotados = 0
        self.tiros_libres_anotados = 0
        self.tiros_dos_anotados = 0
        self.tiros_dos_fallados = 0
        self.triples_fallados = 0
        self.tiros_libres_fallados = 0
        self.perdidas = 0
        self.rebotes = 0
    # Método para calcular la valoración personalizada
    def calcular_eficiencia_jugador(self):
        score = 0
        score += self.puntos * 1
        score += self.asistencias * 1.5
        score += self.rebotes * 1.2
        score += self.robos * 2
        score += self.tapones * 2
        score -= self.tiros_dos_fallados * 0.3
        score -= self.triples_fallados * 0.5
        score -= self.tiros_libres_fallados * 0.5
        score -= self.perdidas * 2
        return score
```
