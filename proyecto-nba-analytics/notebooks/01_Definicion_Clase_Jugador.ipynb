### Metodos de la clase

Esta clase constara de los siguientes metodos para poder convetir sus estadisticas en puntualizacion del juego.

- Punto anotado: Cada punto anotado valdra +1
- Asistencia: Cada asistencia registrada valdara +1.5
- Rebotes: Cada rebote resgistrado valdra +1.2
- Robo: Cada robo generado valdra +2
- Tapones: Cada tapon valdra +2
- Fallo de 2: Cada fallo de lanzamiento de 2 puntos valdran -0.3
- Fallo de 3: Cada fallo de lanzamiento de 3 puntos valdra -0.5
- Fallo de 1: Cada fallo de tiro libre valdra 0.5
- Perdida de balon: Cada perdida de balon valdra -2

```python
class Jugador:
    # Constructor para atributos fijos
    def __init__(self, nombre, edad, nacionalidad, altura, posicion, equipo):
        # 1. Asignación de atributos fijos.
        self.nombre = nombre
        self.edad = edad
        self.nacionalidad = nacionalidad
        self.altura = altura
        self.posicion = posicion
        self.equipo = equipo
        # 2. Asignacion de atributos de Rendimiento. Pondremos valor 0.
        self.puntos = 0
        self.asistencias = 0
        self.minutos = 0
        self.partidos_jugados = 0  
        self.faltas = 0
        self.partidos_lesionado = 0 
        self.robos = 0
        self.tapones = 0
        self.triples_anotados = 0
        self.tiros_libres_anotados = 0
        self.tiros_dos_anotados = 0
        self.tiros_dos_fallados = 0
        self.triples_fallados = 0
        self.tiros_libres_fallados = 0
        self.perdidas = 0
        self.rebotes = 0
    # Método para calcular la valoración personalizada
    def calcular_eficiencia_jugador(self):
        score = 0
        score += self.puntos * 1
        score += self.asistencias * 1.5
        score += self.rebotes * 1.2
        score += self.robos * 2
        score += self.tapones * 2
        score -= self.tiros_dos_fallados * 0.3
        score -= self.triples_fallados * 0.5
        score -= self.tiros_libres_fallados * 0.5
        score -= self.perdidas * 2
        return score

test_player = Jugador("Michael Jordan", "USA", 23, 198, "Escolta", "Bulls")
test_player.puntos = 30        
test_player.asistencias = 10   
test_player.rebotes = 5        
test_player.robos = 2          
test_player.perdidas = 5       
test_player.tiros_dos_fallados = 5 
nota_final = test_player.calcular_eficiencia_jugador()
print(f"La eficiencia de {test_player.nombre} es: {nota_final}")
```

## 2º: Prueba de creacion de objetos con una muestra de datos

Ahora vamos a crear los diferentes jugadores a partir de los datos obtenidos de nba_api que tenemos guardado en un archivo .csv

> Tenemos los datos disttibuidos en partidos, en bruto, de modo que tendremos que agruparlos para que esten asignados a cada jugador

```python
import pandas as pd
from pathlib import Path
# 1. Cargar los datos crudos que bajamos antes
# Genero un Path para la ruta donde esta el archivo
Ruta_csv = Path(r"C:\proyecto-faust\proyecto-nba-analytics\data\nba_raw_data.csv")
df_raw = pd.read_csv(Ruta_csv)

# 2. Definir qué columnas queremos sumar (Las estadísticas dinámicas)
columnas_a_sumar = [
    'PTS', 'AST', 'REB', 'STL', 'BLK', 'TOV',  # Puntos, Asist, Reb, Robos, Tapones, Perdidas
    'FG3M', 'FTM', 'FGM',                      # Triples/Tiros Libres/Campo ANOTADOS (Made)
    'FG3A', 'FTA', 'FGA',                      # Triples/Tiros Libres/Campo INTENTADOS (Attempted)
    'MIN'                                      # Minutos (Ojo, a veces viene como texto, veremos si da error)
]

# NOTA PROFESIONAL: La API a veces trae columnas extra o nombres distintos. 
# Aseguramos que solo sumamos lo numérico para evitar errores.
cols_existentes = [c for c in columnas_a_sumar if c in df_raw.columns]

# 3. LA OPERACIÓN CLAVE: Groupby + Sum
# Agrupamos por ID y Nombre para no perder quién es quién.
# .sum() suma todas las columnas numéricas de cada grupo.
df_jugadores_unicos = df_raw.groupby(['Player_ID'])[cols_existentes].sum().reset_index()

# 4. Cálculo de FALLOS (Ingeniería de Características)
# La API nos da "Anotados" (M) e "Intentados" (A), pero tu clase pide "Fallados".
# Los calculamos ahora restando: Intentados - Anotados.
df_jugadores_unicos['Tiros_Dos_Fallados'] = (df_jugadores_unicos['FGA'] - df_jugadores_unicos['FG3A']) - (df_jugadores_unicos['FGM'] - df_jugadores_unicos['FG3M'])
df_jugadores_unicos['Triples_Fallados'] = df_jugadores_unicos['FG3A'] - df_jugadores_unicos['FG3M']
df_jugadores_unicos['Tiros_Libres_Fallados'] = df_jugadores_unicos['FTA'] - df_jugadores_unicos['FTM']

print("¡Agrupación completada!")
print(f"Hemos pasado de {len(df_raw)} partidos a {len(df_jugadores_unicos)} jugadores únicos.")
display(df_jugadores_unicos.head())
```

## 3º: Creacion del listado completo de objetos con el archivo con los datos completos

Por ultimo realizaremos un script para la creacion de todos los jugadores y que se guarden en la memoria del programa, para probar que se pueden crear con las caracteristicas necesarias

> Los valores NaN se completaran con valores 0
>
> Realizaremos una gestion de excepciones por si hay alguna errata en los nombres de los jugadores

```python
from pathlib import Path
import pandas as pd

# 1. CARGA DE DATOS
Ruta_datos = Path(r"C:\proyecto-faust\proyecto-nba-analytics\data\complete_nba_final_dataset_2026.csv")
df_nba = pd.read_csv(Ruta_datos)

# fillna(0): Rellenamos huecos vacíos con ceros para evitar errores de cálculo
df_nba.fillna(0, inplace=True)

print(f"Cargados {len(df_nba)} registros. Iniciando creación de objetos...")

lista_jugadores = []

# 2. BUCLE DE INSTANCIACIÓN E INYECCIÓN
for index, fila in df_nba.iterrows():
    try:
        # --- A. Datos Identificativos (Constructor) ---
        nombre = fila['DISPLAY_FIRST_LAST']
        nacionalidad = fila['COUNTRY']
        altura = fila['HEIGHT']
        posicion = fila['POSITION']
        equipo = fila['TEAM_ABBREVIATION']
        edad = fila['Edad']

        # Instanciamos el objeto
        jugador_obj = Jugador(nombre, edad, nacionalidad, altura, posicion, equipo)

        # --- B. Mapeo de Estadísticas Directas ---
        jugador_obj.puntos = fila['PTS']
        jugador_obj.asistencias = fila['AST']
        jugador_obj.rebotes = fila['REB']
        jugador_obj.robos = fila['STL']
        jugador_obj.tapones = fila['BLK']
        jugador_obj.perdidas = fila['TOV']
        
        # Nuevas columnas críticas que añadimos en el último paso
        jugador_obj.partidos_jugados = int(fila['GP'])
        jugador_obj.faltas = fila['PF']
        jugador_obj.partidos_lesionado = int(fila['Partidos_Lesionado'])
        jugador_obj.minutos = fila['MIN'] # Puede venir como número o string, lo trataremos luego

        # --- C. Cálculo de Métricas Derivadas (Tiros Fallados) ---
        # Tu clase requiere los fallos, el CSV trae Intentos (A) y Aciertos (M)
        
        # Triples
        jugador_obj.triples_anotados = fila['FG3M']
        jugador_obj.triples_fallados = fila['FG3A'] - fila['FG3M']
        
        # Tiros Libres
        jugador_obj.tiros_libres_anotados = fila['FTM']
        jugador_obj.tiros_libres_fallados = fila['FTA'] - fila['FTM']
        
        # Tiros de 2 (Campo total - Triples)
        tiros_dos_intentados = fila['FGA'] - fila['FG3A']
        jugador_obj.tiros_dos_anotados = fila['FGM'] - fila['FG3M']
        jugador_obj.tiros_dos_fallados = tiros_dos_intentados - jugador_obj.tiros_dos_anotados

        # --- D. Ejecución de Lógica de Negocio ---
        # Calculamos la puntuación usando tu método personalizado
        # Nota: Guardamos el resultado en un atributo dinámico para poder ordenar después
        jugador_obj.valoracion_final = jugador_obj.calcular_eficiencia_jugador()

        lista_jugadores.append(jugador_obj)

    except Exception as e:
        print(f"[ERROR] Fallo al importar {fila.get('DISPLAY_FIRST_LAST', 'Desconocido')}: {e}")

print(f"Proceso completado. Se han generado {len(lista_jugadores)} objetos Jugador.")
```

## 4º: Visualizacion de tabla de prueba de los objetos

Por ultimo realizo un script de prueba para generar un dataframe para ver si puedo extraer algunos objetos con sus atributos

> Pongo alguna regla de filtrado y selecciono algunas filas concretas tanto de atributos fijos, como dinamicos

```python
# --- GENERACIÓN DE REPORTE DE MERCADO ---

datos_reporte = []

for j in lista_jugadores:
    # Filtro de Calidad: Solo analizamos jugadores con actividad mínima (ej: >5 partidos)
    # Esto elimina "ruido" de jugadores que jugaron 1 minuto y tuvieron suerte estadística
    if j.partidos_jugados > 5:
        
        fila = {
            'Nombre': j.nombre,
            'Equipo': j.equipo,
            'Posicion': j.posicion,
            'Edad': j.edad,
            'GP': j.partidos_jugados,
            'Ausencias': j.partidos_lesionado,
            'Val_Total': j.valoracion_final,
            # KPI Clave: Valoración Promedio por Partido
            'Val_Promedio': round(j.valoracion_final / j.partidos_jugados, 2)
        }
        datos_reporte.append(fila)

# Crear DataFrame de Presentación
df_ranking = pd.DataFrame(datos_reporte)

# Ordenar por tu métrica de eficiencia promedio
# (Usamos el promedio por partido porque el total castiga a quien tuvo una lesión corta)
top_mercado = df_ranking.sort_values(by='Val_Promedio', ascending=False).head(15)

print("--- TOP 15 JUGADORES (SEGÚN TU ALGORITMO) ---")
display(top_mercado)
```
