## Que es el big data en esta perspectiva?

Se considera Big Data desde la ciencia de datos al **conjunto de tecnicas y tecnologias necesarias para analizar y convertir datos enm informacion y conocimiento**

Dentro de esta definicion hay campos como:

- Data Mining
- Machine Learning
- Aprendizaje Estadistico
- Analitica de Datos
- Ciencia de Datos

**La definicion exacta de Big Data es datos que tienen un volumen, velocidad y complicacion grandes para ser tratados con herramientas tradicionales**

- Volumen: Es referido a volumenes que se mueven en petabytes (1000 Teras)
- Velocidad: Implica que aparte que se generan muy rapido han de ser tratados muy rapidos
- Complicados/Duros: Quiere decir que no encajan con herramientas de procesado tradicional o requieren de analisis de herramientas no tradicionales.
> Esto se debe a la naturaleza desestructurada de los datos


### Significados de Big Data

Puede haber 4 explicaciones:

- **1º Grandes volúmenes de datos pero "análisis pequeños": SQL analítico sobre muchos datos.**
- **2º Grandes volúmenes de datos y "análisis grandes": Minería de datos con muchos datos.**
- **3º Gran velocidad: Datos en streaming, procesamiento en tiempo real.**
- **4º Gran variedad de procesamiento de datos (ETL) con muchas fuentes de datos.**

## Tecnologias para el big Data

>Hay que realizar una matizacion en la diferencia entre ***Ciencia de datos vs Tecnologias Big data***
>
>> - Ciencia de datos **se refiere a las disciplinas** que tienen como objetivo transformar los datos en conocimiento (*machine learning, estadistica,...*)
>>
>> - Tecnologias de Big Data **se refiere a las tecnicas y herrramientas** que trabaja los problemas del big data

Algunas de estas tecnicas y tecnologias son:

- **Desarrollo de tecnología software para el cálculo distribuido: Hadoop, Spark.**
- **Desarrollo de Bases de Datos para almacenar y procesar información no estructurada: NewSQL, No SQL.**
- **Desarrollo de plataformas de computación: Cloud Computing, High Performance Computing (HPC).**
- **Nuevos análisis que antes no eran posibles, como por ejemplo modelos orientados al cliente.**
> Una empresa de marketing puede tener tantos datos asociados a un solo cliente y la capacidad de procesamiento y cálculo para ajustar modelos individuales a millones de clientes.
- **Productos de Datos en tiempo real.**
- **IoT**

## La ciencia de Datos: Que es y que implica

La ciencia de datos es **la ingenieria civil de los datos**

> Es la conjuncion de los mundos del *Hacking*, *El conocimiento matematico y estadistico* y *conocimiento experto*

Dentro de la ciencia de datos hay nuevas disciplinas a entender:

- **Business Intelligence:** los sistemas de BI se asocian a dos conceptos:
> ***El Data Warehouse (repositorio unificado de fuentes de datos).***
> ***Herramientas de reporte y cuadros de mando. Son sistemas muy rígidos para adaptarse a los requerimientos cambiantes de formato en los datos de hoy en día.***

## La ciencia de datos: Procesos

Podemos definir un proceso de datos aquel que sigue estos pasos:

- **Identifica el problema:** Cual es el objetivo que queremos tener

> *¿Quiero entender como ahorrar costes en mis procesos productivos?, ¿Como van a ser mis ventas en mayo?*

- **Instrumentalizar las fuentes de datos.:** De donde obtengo la informacion para poder tratar y sacar conclusiones

> *Logs y registros de maquinas productoras, partes, registros horarios,....*/*Ventas, acciones de marketing, registros de leads, visualizacion web,...*

- **Recolectar Datos:** Realizar acciones para obtener informacion de las bases de datos, diferentes fuentes localizadas,...
  
> *Bases de datos SQL, NoSQL, ficheros de ERP's, CRM's,...*

- **Preparar datos (integrar, transformar, limpiar, filtrar, agregar)**

> *Haciendo uso de herramientas y librerias de Python, R*

- **Construir Modelos**: Hacer modelos de aprendizaje automatico para poder dar respuesta al problema planteado y dar respuesta.

> *Generar hipotesis sobre como interaccionan las distintas variables, ver los coeficientes de correlacion, entender como se influyen los datos,...*

- **Evaluar modelos**: Entrenar los modelos con datos de entrenamiento y testearlos

> *Poner los modelos en practica e ir mejorandolos con datos*

- **Comunicar resultados**: Generar un storytelling de los datos y compartir informacion

> *Realizar un ejercicio de presentacion de resultados en forma y contenido*



## Ejemplo de como se ejecuta este proceso

- ***Momento Zero:*** En el mundo real generamos una cantidad ingente de datos, **queremos descubrir algo**. **Queremos encontrar relaciones entre diversas fuentes de datos** para dar respuesta a la pregunta

- Queremos procesar estos datos para tenerlos limpios y en un formato adecuado para el análisis. **Esto es el proceso de adquisición y preparación de datos.** Para ello utilizamos herramientas como Python, scripts de Shell, R, SQL, o todo lo anterior.

- Eventualmente **conseguimos tener los datos en un formato agradable**, algo dispuesto en columnas tipo.
> En el curso de este análisis, podemos darnos cuenta de que los datos aún no están limpios del todo debido a duplicados, valores ausentes, valores absurdos, datos que no se registraron realmente o que se registraron incorrectamente. Si ese es el caso, puede que tengamos que volver a recoger más datos, o pasar más tiempo limpiando los conjuntos de datos.

- **Diseñamos nuestros modelos para usar algún algoritmo** como k-vecinos, una regresión lineal o un árbol de decisión. El modelo que elegimos depende del tipo de problema que estamos tratando resolver, lo que podría ser un problema de clasificación, un problema de predicción o un problema de descripción básica.

- **Por último, debemos interpretar, visualizar y comunicar nuestros resultados.** Esto podría ser para comunicar los resultados a nuestros clientes, o a nuestro jefe y compañeros de trabajo, o publicar un artículo académico, o en una web de acceso público.

### Explicacion del Ciclo ETL

Este proceso permite obtener datos de calidad para poder realizar analiticas, **pero puede conllevar realizarlo varias veces**

- **Extraccion de datos de las fuentes**

- **Cargar datos en programas que los traten(*o colectores)***

- **Necesitamos transformar los datos en los colectores o en areas temporales**

> Dentro de este proceso preparatorio podemos incluir el análisis exploratorio de datos. Antes de hacer inferencias/predicciones es esencial conocer y examinar nuestras variables para entre otras cosas:

> > Ver patrones en los datos.

> > Generar hipótesis.

> > Encontrar violaciones a las suposiciones estadísticas.

## Modelos en ciencia de datos

Un modelo **es un intento de entender y representar a la realidad**a través de una lente particular, donde los detalles superfluos se han eliminado o abstraído.

Dentro de la ciencia existen modelos bilogicos y estadisticos/ciencia de datos. **Seran estos segundos en los que nos centraremos**

## Diversidad de Clasificacion de Modelos estadisticos:

Si lo vemos desde la perspectiva de **la teoría matemática** hay 2:

- **Machine Learning:** algoritmos que son **capaces de aprender a partir de los datos sin basarse en reglas preestablecidas.**
- **Modelado estadístico:** modelos basados en la **formalización de relaciones entre variables a partir de ecuaciones matemáticas y distribuciones de probabilidad.**

Si lo vemos desde la perspectiva del **problema que queremos resolver** hay 2:

- **Modelos de regresión:** el objetivo es **predecir un valor numérico:** ventas, demanda de electricidad o temperatura.
- **Modelos de clasificación:** su misión es **predecir una categoría:** Hombre/Mujer, Correo bueno o basura.

Si lo vemos desde la perspectiva del **proceso necesario para el ajuste** hay 2:

- **Aprendizaje supervisado:** a partir de datos etiquetados, valor de las ventas o el género (hombre/mujer) construimos un modelo que **es capaz de hacer predicciones sobre valores nuevos de las variables predictoras:** gasto en publicidad o dimensiones de los huesos.

- **Aprendizaje no supervisado:** es **capaz de descubrir relaciones entre las observaciones sin un etiquetado previo.** Se usan por ejemplo para hacer segmentaciones de clientes por su tipo de consumo.
