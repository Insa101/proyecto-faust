# Apuntes Modelos de Regresion Estadistica

En este cuaderno veremos diferentes modelos de prediccion **para variables numericas** estos son los modelos de regresion.

Vamos a empezar con el modelo mas general, **la regresion lineal**

## 1.0 ¿Que es la regresion lineal?

Es un tipo de modelo **de aprendizaje supervisado que asume que la relacion entre la variable objetivo y las predictoras es lineal**

> Generalmente este modelo es irreal, porque pocas variables se relacionan 1:1, pero sus resultados, **suelen ser precisos y muy interpretables** por lo que se considera un metodo muy acpetado



## 2.0 Regresion Lineal Simple

Uno de los metodos de la regresion lineal es la regresion lineal simple.

En este metodo, **disponemos de una variable predictora que relaciona 2 parametros o coeficientes desconocidos que definen la pendiente (*slope)* y el inicio (*intercept*) que tendremos que estimar para poder definir valores nuevos**

Suele tener esta representacion matematica:

`Y = P1 + P2X + e`

> Donde P1 es el inicio, P2 define la pendiente y *e* es el error irreducible

Para ello utilizamos el metodo de *ajuste a minimos cuadrados* que **consiste en minimizar la suma de las distancias al cuadrado entre los valores reales y los valores predichos**

> Consiste en la suma de los errores cuadrados de cada prediccion para determinar **que valores han de tomar los parametros para que el error se aproxime lo maximo a 0**

### 3.0 Medir la precision de estos modelos de regresion simple

En estos modelos hay 2 partes diferenciadas para determinar la precision.

- 1º Hay que determinar la precision de los coeficientes o parametros
- 2º Hay que determinar la precision del modelo en su conjunto

### 3.1 Precision de los coeficientes o parametros

> **Para determinar estos realizaremos simulaciones y comprobaremos conjuntamente la distribucion de los parametros estimados**

1. **Aqui vemos las distintas estimaciones con el mismo modelo y distintos valores de parametros**


2. **Aqui vemos la distribucion de los valores de los parametros en las estimaciones hechas y nos indica la *dispersion***


> Estas dispersiones son los errores que se pueden calcular a partir del error estandar del error irreducible

3. **Una vez determinado el error estandar podemos indicar un *intervalo de confianza***

> La cual te indica el % de valores estimados que obtendrian el valor real si se aplica el modelo

4. **Realizacion de Test de Hipotesis**, te permite determinar si existe una relacion causal o dependencia entre las variables objetivo y predictoras

> Nos permite saber que importancia tiene una variable predictora para determinar el valor valor de la variable objetivo.
>
> Para esto haremos uso del *p-valor* que determina la probabilidad de que el parametro siendo 0, sea tambien 0
>
> > **Al realizar este test, nos damos cuenta de la importancia/dependencia que tienen en el modelo**

### 3.2 Precision del modelo en su conjunto

1. ***Error estandar (RSE)***: Que es la media de los errores de la diferencia entre el valor real y el predecido. EL problema es que depende demasiado de los parametros y no adapta su escala, **no hace que el modelo se pueda utilizar en otros entornos o con otros parametros nuevos**

2. ***Coeficiente de correlacion lineal o coeficiente de correlacion de pearson (R2)***: Determina cuanto te desvias de la varianza total respecto del modelo. 

> **Cuanto mas cercano a 1 sea, mejor ajuste lineal tiene**
>
> Este modelo solo coincide en modelos de regresion lineal, en otros casos **solo podremos calcular el coeficiente de correlacion, no el coeficiente de pearson***.

## 4.0 Regresion Lineal Multiple

En este metodo, **disponemos un conjunto de variables predictoras en vez de 1 unica variable predictora, que definen varios parametros**

Suele tener esta representacion matematica:

`Y = P1 + P2X + P3X... + e`

> Donde las P's definen parametros y *e* es el error irreducible

En este caso hay que **relacionar las variables de manera matricial sobre la cantidad de valores que tengamos y para cada una de los parametros**

El error se calcula por *minimos cuadrados* otra vez

### 5.0 Medir la precision de estos modelos de regresion simple

En estos modelos hay 2 partes diferenciadas para determinar la precision.

- 1º Hay que determinar la precision de los coeficientes o parametros
- 2º Hay que determinar la precision del modelo en su conjunto

### 3.1 Precision de los coeficientes o parametros

1. **Calcular la *varianza* o el error estandar a partir de los errores**

2. **Determinar el *coeficiente*, que indica como la variable objetivo fluctua de media en la misma fluctuacion de la variable predictora**
> Si las variables predictoras no estan relacionados, **se puede calcular de manera individual el coeficiente**
>
> Si estan correlacionados, **la varianza de todos fluctua** y hace que sea mas dificil interpretar elñ peso de cada variable predictora.

2. ***Coeficiente de correlacion lineal o coeficiente de correlacion de pearson (R2)***: Determina cuanto te desvias de la varianza total respecto del modelo. 

> **Cuanto mas cercano a 1 sea, mejor ajuste lineal tiene**
