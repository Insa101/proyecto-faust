# Apuntes Modelos de Clasificacion

En este cuaderno veremos diferentes modelos de prediccion **para variables cualitativas** estos son los modelos de clasificacion

## 1.0 ¿Que es un modelo de clasificacion?

Es un tipo de modelo **de aprendizaje supervisado que dispone de variables no ordenadas que dan respuesta a variable objetivo cualitativa, sin una relacion matematicamente obvia**

En este caso toma como entrada el valor de las observaciones de los predictores y da como resultado **la categoria a la que pertenece el valor de la variable predictora, minimizando el error**

> Ademas permite indicar el % de probabilidad que tiene que el modelo acierte en la prediccion.

## 2.0 Metodo de regresion logistica

> Es el metodo mas sencillo de clasificacion

Es un metodo basado en la funcion logistica, **es una funcion que siempre da como resultado algo entre 0 y 1**

Dispone de estas caracteristicas:

> Se llama logistico ya que es el reusltado el logaritmo de las probabilidades del modelo, que dan como reusltado una regresion lineal.
>
> El valor de las probabilidades es lo que aplicaremos a las observaciones para categorizar corectamente el valor
