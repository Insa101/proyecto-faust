# Apuntes como preparar archivos Markdown en R

Una cosa que nos solicitan es realizar un ejercicio final de R en formato Markdown

Lo que debemos hacer es **utilizar RStudio** 

El proceso hace que: un script en R, .Rmd es transformado mediante un motor knitr, que formatea el script a otro formato de salida, y **lo convierte a formato md (*markdown*) que sirve para convertirlo en un archivo html de marcado**
