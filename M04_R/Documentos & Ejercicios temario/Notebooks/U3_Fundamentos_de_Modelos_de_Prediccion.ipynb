## 1.0 Que es el aprendizaje estadistico o modelo estadistico?

**Es una formulacion matematica que relaciona una variable objetivo (*target*) con las variables predictoras.(*inputs/features*)**

Sus componentes son:

- La variable objetivo es **la que queremos predecir**, generalmente la representamos matematicamente como ***Y***

- Las variables predictoras **son las variables que influyen al predecir**, se representan matemaaticamente como ***X*** y qgeneran unos **errores *e* entre el valor real y la *Y***

- El error e **captura los errores de medida de las variables predictoras** e influencia otros factores que no hemos tenido en cuenta o el azar.

La representacion matematica es asi es:

`Y = f(X) + e`

Basicamente **el aprendizaje estadistico son el conjunto de tecnicas para estimar la *f()***

### 1.1 ¿Para que sirve el modelado estadistico?

- Te permite **hacer predicciones para nuevos valores de las variables predictoras**
- Te permite diferenciar **que variables predictoras son importantes y cuales son irrelevantes**
- Te permite entender **la ponderacion en la influencia de la variable predictora sobre la variable objetivo**

### 1.2 ¿Que tipos de modelados existen?

Depende de si el objetivo a desentrañar de la variable tiene una naturaleza u otra:

- Si la variable objetivo **es numerica** entonces el modelo estadistico sera **un modelo de regresion**
- Si la variable objetivo **es cualitativa** entonces el modelo estadistico sera **un modelo de clasificacion**
- Si desconocemos de que naturaleza es la variable objetivo, el modelo estadistico tendra que ser no supervisado y **un modelo de clusterizacion**

## 2.0 ¿Como ajustamos los modelos estadisticos de regresion?

Uno de los aspectos mas importantes del modelaje es el ajuste del modelo para que mejore su capacidad de prediccion sobre la realidad, es decir, **enseñar al modelo para hacer que frente a cualquier valor o tipo de variable predictora contemplada, me diga la variable objetivo**

Para esto existen distintas tecnicas de ajuste:

### 2.1 Ajuste mediante Parametros

Este tipo de ajuste **requiere que definas que tipo de relacion disponen la variable objetivo con las variables predictivas**

Cada variable predictora tiene unos parametros **que desconocemos** y que para averiguarlos, sera necesario **utilizando los datos de entrenamiento que definen los valores de las variables predictivas**

- Los modelos parametricos son **mas sencillos de definir**

> En este categoria se encuentran **los modelos de regresion lineal, cuadratico, logaritmico, poligomico,...**
>




#### 2.1.1 Ajuste mediante Regresion simple

Este tipo de ajuste consiste en **calcular el error cuadratico de la funcion e ir buscando el minimo error soportado del valor estimado mas adecuado**

> Por lo que nos dice la teoria matematica, una regresion, **va a tener siempre un % de desvio frente a la realidad**
>
> Ya hemos visto que esto solo sirve para **variables objetivo cuantitativas**
>
> Funciona bien **cuando disponemos maximo 2 variables predictoras**

### 2.2 Ajuste mediante no-parametros

Este tipo de ajuste **busca estimar el valor objetivo cerca de los datos sin ser el ajuste de union de puntos o hacer muchas asunciones**

- Tienen la ventaja de o tener **no generar asunciones previas sobre la forma**
- Tienen la desventaja de **el resultado es dificil de interpretar**

#### 2.1.1 Ajuste mediante *K-vecinos o KNN* (K nearest neighbours)

Este tipo de ajuste consiste en **tomar en cuenta valores que estan alrededor en un intervalo concreto (%) de el valor estimado mas adecuado, y se fija el valor promedio entre estos valores tomados**

> Funciona bien **cuando disponemos de pocas variables predictoras**
>
> > Esto es debido a que **cuantas mas variables, mas dimensiones y mas dificil es de predecir/representar la realidad**

El intervalo que escoges para tomar en cuenta los valores **condiciona la calidad de la prediccion**
>
> - Si el intervalo es muy pequeño, **sin muchos valores**, haras estimaciones ***muy ruidosas o erroneas***
> 
> - Si el intervalo es muy pequeño, **pero tienes muchos valores**, haras estimaciones **muy precisas**
> 
> - Si el intervalo es muy grande, **haras una curva muy suavizada pero mas alejada de la funcion de los valores reales**




### 2.3 ¿Que tenemos que tener en cuenta para ajustar un modelo estadistico?

En general al realizar un modelo predictivo tenemos intencion de conseguir 2 cosas que son **diametralmente opuestas**:

-***Precisión***: se refiere a la capacidad del modelo de **replicar la realidad de manera mas fidedigna**
-***Interpretación***: se refiere a la capacidad del modelo de **ser representado en un plano**

> Para ello nuestra capacidad de trabajo tiene que hacer que se realicen buenos ajustes.

##### 2.4 ¿Que es un buen ajuste?

**Es el punto optimo que da una calidad de prediccion adecuada entre precision e interpretacion.**

Para ello evitaremos el sobre y sub ajustes:

- Un **sobre-ajuste** sucede si **tenemos demasiado en cuenta las variaciones de los datos de entrenamiento pero no sea capaz de realizar la misma interpretacion sobre datos de test**

- Un **sub-ajuste** sucede si **tenemos en cuenta pocas variaciones y realiza predicciones muy imprecisas de la funcion real**

Luego estan los 2 problemas del ajuste:

1. el problema ***de la caja negra*** ya que no entiendes que hace el modelo para dar el resultado, **sus resultados son poco interpretables**
> Surge si se quieren utilizar muchas variables o evitar el sub-ajuste

2. el problema ***de la simplicidad*** que hace que el modelo es poco relevante y **sus datos son poco precisos**
> Surge al buscar un menor sobreajuste, involucrando pocas variables y maximizando la interpetabilidad



## 3.0 Tipos de aprendizaje para modelos: Supervisados y No-Supervisados

Otro enfoque para el modelaje estadistico es la perspectiva de **como o quien entrena el modelo**

Todo lo mencionado hasta ahora se ha visto desde un ***aprendizaje supervisado*** ya que en este caso **para cada observacion/registro de las variables predictoras disponemos de una variable objetivo que da respuesta**

De este tipo de modelos nos encontramos con: las regresiones, k-vecinos

> El objetivo es ajustar la relacion las variables predictoras

Sin embargo existe otro tipo de aprendizaje: ***aprendizaje no-supervisado***

En estos casos **dispondremos de unas obervaciones de nuestras variables predictoras pero sin variables objetivos asociadas** por lo que solo tiene en cuenta las propias caracteristicas de las observaciones

De este tipo de modelos nos con encontramos con: las clusterizaciones

> El objetivo es encontrar relaciones entre las variables predictoras

## 4.0 Determinando la precision de los modelos: *MSE* o Error Cuadratico Medio

Ahora vamos a ver varios modos de realizar calculos que nos indiquen **la calidad del ajuste y capacidad de predecir del modelo**

El principal punto aqui es: Nosotros debemos tener la capacidad de entendimiento de los modelos **para ajustarlos en sus diferentes parametros para obtener la mayor calidad de datos** (ajustar la flexibilidad, el nº de variables,....)

Uno de los puntos mas importantes del modelado estadistico es la capacidad de este de ser preciso en la prediccion.

> Por ello hemos de ajustar adecuadamente los modelos **con nuestros datos de entrenamiento** antes de realizar una evaluacion.

Para ello podriamos obervar **el error cuadratico medio(*MSE)*** que nos indicara una medida de la bondad del ajuste, indicandonos en un parametro **cuanto se desvian de media las predicciones desde las variables objetivo**

Este parametro siempre recomendaremos calcularlo **haciendo uso del modelo con datos nuevos o datos test**, ya que no estaran los resultados tan sesgados como los de entrenamiento.

> Por tanto **siempre deberemos separa nuestros datos en entrenamiento y test para realizar este analisis**
>

#### Concepto Flexibilidad en Modelos estadisticos

**La flexibilidad es la capacidad del modelo de ajustarse para generar predicciones**

- Una flexibilidad grande o sensibilidad del modelo para la flexibilidad **provocora un sobre ajuste**
- Una felxibilidad inferior o poca sensbilidad del modelo para la flexibilidad **provocora un sub-ajuste**

Es importanteterlo en cuenta ya que **cuanto mas flexible, menor sera el error, pero menos versatil con nuevos datos sera el modelo** mientras que **si es poco flexible, el error sera mayor, pero cambiaran poco las predicciones entre datos nuevos**

Por tanto, este concepto nos da 3 escenarios para nuestros modelos:

- Modelos con flexibilidad baja: Estos modelos son **aquellos que tengan una grafica *suave* y tiene un error irreducible grande**.
> Suelen ser la regresion lineal o con 2/3 grados de libertad.



- Modelos con flexibilidad alta: Estos modelos son **aquellos que tengan una grafica *rugosa* y tiene un error irreducible pequeño**.
> 

>

- Modelos con flexibilidad intermedia: Este sera un punto entre alguna de las 2 cuestiones vistas anteriormente.

#### Varianza y Bias: Parametros para ajustar la calidad del modelo

Otros 2 conceptos a aprender son: **la varianza y el bias**

- La varianza se refiere a la **variabilidad del modelo sobre los datos de entrenamiento**

> Es decir, cuanto se alejan los datos predichos de los datos entrenados
>
> Modelos con muchas flexibilidad , alta varianza

- El bias se refiere al **error que se genera al aumentar la flexibilidad de un modelo**

> Es decir, cuanto se alejan los datos predictos debido al simplificar el modelo
>
> Modelos con poca flexibilidad , alto bias

Por tanto, **el ajuste optimo se encontrara a encontrar la varianza y el bias adecuados conjuntamente del modelo**

## 5.0 Modelos estadisticos de clasificacion

Otro de los tipos de modelos que estudiamos son los de clasificacion **donde la variable objetivo es una categoria o es cualitativa**

Tiene caracteristicas:

- Este tipo de modelos buscan construir un clasificador **al que apliques sobre datos sin etiqueta para agruparlos o definirlos**
- Busca **entender los papeles de cada una de las variables predictoras** (la importancia, la flexibilidad,...)
- **Asigna una probabilidad a cada clasificacion**

### 5.1 ¿Como son estos modelos?

Estos modelos van a asignar una probabilidad a cada observacion de que pertenezcan a un tipo de categoria, segun el valor que tenga la variable predictiora.

> El tipo de categoria puede ser Si/No, Ser/No ser, una lista de categorias,...

**Se utilizan los mismos metodos de ajuste parametricos y no parametricos**

> Sin embargo disponen de propios como **el clasificador optimo (bayes)**

### 5.2 ¿Como medimos la precision?

Utilizaremos el error de clasificacion **que compara el nº de errores contra el nº de aciertos de la clasificacion predictora y la clasificacion real**

### 5.3 Tipos de modelos de clasificacion

1. ***Clasificadores estructurados (parametricos)***: Que solo **predicen la clase/categoria** como los *support vector machines*
2. ***Regresion logistica o modelos aditivos generalizados***: Modelos estructurados que **predicen para la probabilidad de la prediccion**
3. ***No parametricos***: el K-vecinos o arboles de clasificacion, que son mas interpretables y menos precisos.
